## 12.2 — ¿Qué es un contrafactual? El contrafactual y el problema fundamental (Paso 1) (Sección 1 del cuaderno)

**Correlación ≠ causa.** Que «los clientes que recibieron la campaña se quedaron más» **no** prueba que la campaña funcione: falta saber qué habrían hecho **esos mismos clientes** sin campaña. A cada unidad se le asocian dos **resultados potenciales**: `Y(1)` (con tratamiento) e `Y(0)` (sin). El **efecto individual** `Y(1) − Y(0)` es **inobservable** (solo se ve uno de los dos): es el **problema fundamental de la inferencia causal** (Rubin 1974; Holland 1986). Por eso se estiman promedios: **ATE** `= E[Y(1) − Y(0)]` y **ATT** `= E[Y(1) − Y(0) | T=1]`.

La minidemostración construye un mundo simulado donde —solo para enseñar— se conocen **ambos** resultados potenciales, y muestra cómo la diferencia de medias observada (naive) se aleja del efecto verdadero cuando la asignación **no** es aleatoria.

#### 🧮 Matemática en el cuerpo — resultados potenciales, contrafactual y estimandos (ATE, ATT)

A cada unidad $i$ se le asocian dos **resultados potenciales**: $Y_i(1)$ si recibe el tratamiento e $Y_i(0)$ si no. El **efecto individual** es $\tau_i = Y_i(1) - Y_i(0)$, pero solo se observa

$$Y_i = T_i\,Y_i(1) + (1-T_i)\,Y_i(0),$$

es decir **uno** de los dos; el otro es el **contrafactual** (nunca observable). Ese es el **problema fundamental** (Holland 1986): $\tau_i$ es inestimable a nivel individual. Por eso se estiman **promedios**:

$$\text{ATE}=\mathbb{E}[Y(1)-Y(0)],\qquad \text{ATT}=\mathbb{E}[Y(1)-Y(0)\mid T=1].$$

La **diferencia de medias observada** solo estima el ATE si la asignación es **ignorable**; con autoselección aparece el **sesgo**:

$$\underbrace{\mathbb{E}[Y\mid T=1]-\mathbb{E}[Y\mid T=0]}_{\text{naive}}=\text{ATT}+\underbrace{\big(\mathbb{E}[Y(0)\mid T=1]-\mathbb{E}[Y(0)\mid T=0]\big)}_{\text{sesgo de selección}}.$$

Fuente: el glosario de la sesión puntos 1-3; la guía de supuestos de la sesión, Parte 1.

**❓ Qué se quiere averiguar.** ¿La diferencia de resultado entre quienes recibieron la campaña y quienes no responde a la pregunta «¿cuánto añadió la campaña?», o contesta en realidad a otra distinta?

- **Qué decide:** si esa resta bastara, ningún método de esta sesión tendría razón de ser: se leería el reporte y se decidiría. Toda la sesión depende de esta respuesta.
- **Antes de mirar el resultado:** el efecto verdadero está fijado por diseño en **+8** para cada cliente. Si el naive resulta **≈ +8**, la resta simple sí estima el efecto. Si resulta **muy por encima**, la resta contesta otra pregunta —«¿en qué se diferencian los dos grupos?»— y mezcla el efecto con lo que ya separaba a tratados de no tratados **antes** de la campaña.

**🔎 Qué hace este código.** Construye un **mundo simulado** donde —solo para enseñar— se conocen **ambos** resultados potenciales `Y(0)` e `Y(1)`, con asignación **no aleatoria**. Compara el **ATE verdadero** (conocido por diseño) con la **diferencia de medias observada (naive)** para exhibir el sesgo de selección.

```
# Mini-demo: mundo simulado donde SÍ conocemos Y(0) e Y(1) (solo con fines didácticos)
rng_demo = np.random.default_rng(7)
Nd = 2000
riesgo = rng_demo.normal(0, 1, Nd)                    # rasgo del cliente (afecta trato y resultado)
Y0 = 50 + 10 * riesgo + rng_demo.normal(0, 5, Nd)     # resultado potencial SIN tratamiento
Y1 = Y0 + 8.0                                          # efecto verdadero individual = +8 (constante)
T = (rng_demo.uniform(size=Nd) < 1 / (1 + np.exp(-riesgo))).astype(int)   # asignación NO aleatoria
Yobs = np.where(T == 1, Y1, Y0)                        # solo se observa uno de los dos

ate_real = float((Y1 - Y0).mean())                    # verdad (inobservable en la práctica)
naive = float(Yobs[T == 1].mean() - Yobs[T == 0].mean())
print(f"Efecto verdadero (ATE, conocido por diseño) = {ate_real:+.2f}")
print(f"Diferencia de medias observada (naive)      = {naive:+.2f}")
print(f"Sesgo del naive = {naive - ate_real:+.2f}  (los tratados ya partían distinto)")
print()
print("Lectura de negocio: sin aleatorización, la diferencia de medias mezcla el efecto con las")
print("diferencias PREVIAS entre quienes reciben y quienes no. Hay que construir un contrafactual")
print("creíble (aleatorización, DAG + ajuste, DiD, PSM).")
```

**📖 Cómo se lee.** El **efecto verdadero es +8** (constante de diseño), pero el **naive** se aleja de esa cifra: los tratados ya partían con mayor `riesgo`, que también eleva el resultado. El **sesgo** es exactamente la diferencia en `Y(0)` entre quienes reciben y quienes no. 💡 Sin aleatorización ni ajuste, la diferencia de medias **mezcla** el efecto con las diferencias previas: hay que construir un **contrafactual creíble** (A/B, DAG + ajuste, DiD, PSM).

## 12.3 — ¿Cómo se determinan las variables que deben controlarse? El DAG del negocio y el conjunto de puerta trasera (Paso 2, Drill 1) (Sección 2 del cuaderno)

Un **DAG** (grafo acíclico dirigido) explicita las hipótesis causales y —con el **criterio de puerta trasera** (Pearl 1995)— dice **qué controlar y qué NO**. En el caso de negocio:

- `engagement → campaign` y `engagement → value`: `engagement` (actividad previa) es un **confusor** — abre la puerta trasera `campaign ← engagement → value`.
- `tenure → value`: `tenure` (antigüedad) es **covariable de precisión** (mejora la estimación, no es necesaria para identificar).
- `campaign → value`: el efecto que se busca.

**Conjunto de puerta trasera = {engagement}**: basta controlar `engagement` para identificar el efecto de `campaign` sobre `value`.

**Cuidado (drill 1): NO todo se controla.** Si se añade una variable **posterior** al tratamiento —p. ej. `uso_post` (uso del producto **después** de la campaña), efecto a la vez de la campaña y del valor: `campaign → uso_post ← value`— se estaría condicionando en un **collider**, lo que **fabrica** una asociación espuria (sesgo de colisión). Un mediador `campaign → M → value` tampoco se controla si interesa el **efecto total**. La regla: controlar **solo el conjunto de puerta trasera**.

#### 🧮 Matemática en el cuerpo — ignorabilidad y el criterio de puerta trasera

La estimación por ajuste identifica el efecto causal bajo **ignorabilidad** (no confusión) y **positividad**:

$$\big(Y(0),Y(1)\big)\ \perp\!\!\!\perp\ T \mid X,\qquad 0<e(X)=P(T=1\mid X)<1.$$

Bajo esos supuestos, el efecto se recupera promediando el efecto **dentro** de cada perfil $X$ (ajuste / *g-formula*):

$$\text{ATE}=\mathbb{E}_X\big[\,\mathbb{E}[Y\mid T=1,X]-\mathbb{E}[Y\mid T=0,X]\,\big].$$

El **criterio de puerta trasera** (Pearl 1995) da la regla operativa de **qué** poner en $X$: un conjunto $Z$ identifica el efecto de $T$ sobre $Y$ si **(1)** bloquea todos los caminos que llegan a $T$ por detrás ($T \leftarrow \cdots \rightarrow Y$) y **(2)** **no** contiene descendientes de $T$ (ni mediadores ni colliders). En el caso de negocio, $Z=\{\texttt{engagement}\}$ cierra la única puerta trasera $\texttt{campaign} \leftarrow \texttt{engagement} \rightarrow \texttt{value}$.

Fuente: el glosario de la sesión puntos 4-8; la guía de supuestos de la sesión, Partes 2 y 5.

**🔎 Qué hace este código.** Dibuja **dos DAG** del caso de negocio con matplotlib: (a) el DAG correcto con la puerta trasera `{engagement}`, y (b) una variante con la variable **`uso_post`** (uso posterior a la campaña) que es un **collider** y **NO** se debe controlar.

```
# DAG del caso de negocio (dibujado a mano con matplotlib; no requiere graphviz)
fig, ax = plt.subplots(1, 2, figsize=(11, 4))

def flecha(a, p0, p1, color=TINTA):
    a.annotate("", xy=p1, xytext=p0,
               arrowprops=dict(arrowstyle="-|>", color=color, lw=1.8))

def nodo(a, p, txt, color="white", edge=ROJO):
    a.scatter(*p, s=2600, c=color, edgecolors=edge, linewidths=1.8, zorder=3)
    a.text(*p, txt, ha="center", va="center", fontsize=8.5, zorder=4)

# (a) DAG correcto: puerta trasera = {engagement}
pa = {"engagement": (0.5, 1.0), "campaign": (0.0, 0.0), "value": (1.0, 0.0), "tenure": (1.7, 0.6)}
flecha(ax[0], pa["engagement"], pa["campaign"], ROJO)
flecha(ax[0], pa["engagement"], pa["value"], ROJO)
flecha(ax[0], pa["tenure"], pa["value"], GRIS)
flecha(ax[0], pa["campaign"], pa["value"], TINTA)
for k, p in pa.items():
    nodo(ax[0], p, k)
ax[0].set_title("(a) DAG correcto — puerta trasera = {engagement}")

# (b) Variante con COLLIDER que NO se debe controlar
pb = {"engagement": (0.5, 1.0), "campaign": (0.0, 0.0), "value": (1.0, 0.0), "uso_post": (0.5, -0.9)}
flecha(ax[1], pb["engagement"], pb["campaign"], ROJO)
flecha(ax[1], pb["engagement"], pb["value"], ROJO)
flecha(ax[1], pb["campaign"], pb["value"], TINTA)
flecha(ax[1], pb["campaign"], pb["uso_post"], TINTA)
flecha(ax[1], pb["value"], pb["uso_post"], TINTA)
for k, p in pb.items():
    nodo(ax[1], p, k, color=(ROSA if k == "uso_post" else "white"))
ax[1].set_title("(b) 'uso_post' es COLLIDER — NO se controla")

for a in ax:
    a.set_xlim(-0.6, 2.3); a.set_ylim(-1.4, 1.5); a.axis("off")
plt.tight_layout(); fig.savefig(FIG / "S12_dag.png", dpi=150, bbox_inches="tight"); plt.show()
print("Backdoor correcto: controlar {engagement}. NUNCA controlar 'uso_post' (collider):")
print("condicionar en un collider fabrica sesgo. El backdoor lo confirmará DoWhy (paso 9).")
```

**📖 Cómo se lee.** En (a), el único camino de puerta trasera es `campaign ← engagement → value`: **`engagement`** es el **confusor** (se controla), **`tenure`** es covariable de **precisión** (opcional), y el conjunto de ajuste mínimo es **{engagement}**. En (b), **`uso_post`** es un **collider** (`campaign → uso_post ← value`): condicionar en él **abre** el camino y **fabrica** sesgo. ⚠️ Un **mediador** `campaign → visitas → value` tampoco se controla si interesa el **efecto total**. Regla: controlar **solo** el conjunto de puerta trasera (ver `SUPUESTOS_S12.md`, Parte 5).

## 12.4 — ¿Cómo se estima un efecto sin posibilidad de experimentar? El A/B como patrón oro y el poder/MDE (Paso 3, enlace S02) (Sección 3 del cuaderno)

La **aleatorización** iguala en esperanza tratados y controles en **todas** las variables (observadas y no): da **ignorabilidad por diseño** y la diferencia de medias es **insesgada**. Es el **patrón oro** que los métodos observacionales intentan imitar. Antes de lanzar un A/B hay que dimensionarlo: el **MDE** (*Minimum Detectable Effect*) es el efecto más pequeño detectable con `α`, poder y `n` dados: `MDE ≈ (z_{1−α/2} + z_{1−β})·√(2σ²/n)` por brazo. El `n` crece **cuadráticamente** al reducir el MDE (detectar la mitad del efecto exige ≈4× la muestra).

> El A/B **básico** es de la Sesión 2; aquí se **extiende** al marco causal (contrafactual, poder, MDE), no se re-enseña lo elemental.

**🔎 Qué hace este código.** Calcula el **tamaño de muestra por brazo** necesario para detectar un efecto (MDE) dado, con $\alpha=0{,}05$ y poder $0{,}80$, y traza la **curva `n` vs. MDE**. Marca el efecto de negocio (`8,0`) como referencia.

```
from scipy import stats
import statsmodels.formula.api as smf

def n_por_brazo(mde, sigma, alfa=0.05, poder=0.80):
    "Tamaño por brazo para detectar 'mde' (comparación de dos medias, dos colas)."
    z = stats.norm.ppf(1 - alfa / 2) + stats.norm.ppf(poder)
    return (z ** 2) * 2 * sigma ** 2 / mde ** 2

# La dispersión que entra en el 'n' NO es única: depende de si el A/B se analiza CRUDO o AJUSTADO.
#  - A/B CRUDO (diferencia de medias sin ajustar): usa la SD MARGINAL de 'value' (~28-29).
#  - AJUSTADO por covariables (regresión sobre engagement, tenure): la sigma EFECTIVA baja al
#    RUIDO residual del DGP, N(0,8) -> ~8, y con ella el 'n' requerido. OJO: el 8 del DGP es la
#    SD del ruido, NO la dispersión de 'value'; solo vale una vez que se ajusta por covariables.
_dfp = generar_negocio()                                              # mismo DGP del laboratorio
sigma_crudo = float(_dfp["value"].std(ddof=1))                        # SD marginal (A/B crudo) ~28-29
sigma_ajustado = float(smf.ols("value ~ campaign + engagement + tenure",
                               data=_dfp).fit().resid.std(ddof=1))    # ruido residual ~8

mdes = np.linspace(1, 12, 40)
ns_crudo = np.array([n_por_brazo(m, sigma_crudo) for m in mdes])
ns_ajust = np.array([n_por_brazo(m, sigma_ajustado) for m in mdes])

fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(mdes, ns_crudo, color=ROJO, label=f"A/B crudo (σ≈{sigma_crudo:.0f})")
ax.plot(mdes, ns_ajust, color=TINTA, ls="--", label=f"ajustado por covariables (σ≈{sigma_ajustado:.0f})")
ax.axvline(TRUE_ATE, color=GRIS, ls=":")
ax.text(TRUE_ATE + 0.2, ns_crudo.max() * 0.7, f"efecto de negocio = {TRUE_ATE:.0f}", color=GRIS)
ax.set_title("Tamaño de muestra por brazo vs. MDE (α=0,05; poder=0,80)")
ax.set_xlabel("MDE (unidades de 'value')"); ax.set_ylabel("n por brazo"); ax.legend()
plt.tight_layout(); fig.savefig(FIG / "S12_poder_mde.png", dpi=150, bbox_inches="tight"); plt.show()

print(f"A/B CRUDO   (σ≈{sigma_crudo:.0f}): detectar 8,0 exige n ≈ {n_por_brazo(8.0, sigma_crudo):.0f}/brazo; "
      f"la MITAD (4,0) ≈ {n_por_brazo(4.0, sigma_crudo):.0f}/brazo (≈4×).")
print(f"AJUSTADO    (σ≈{sigma_ajustado:.0f}): al controlar covariables, detectar 8,0 exige n ≈ "
      f"{n_por_brazo(8.0, sigma_ajustado):.0f}/brazo.")
print(f"Nota: σ≈{sigma_ajustado:.0f} es el RUIDO residual del DGP N(0,8), NO la dispersión de 'value'")
print(f"(σ marginal ≈{sigma_crudo:.0f}); el n≈{n_por_brazo(8.0, sigma_ajustado):.0f} vale SOLO para el análisis AJUSTADO, no")
print(f"para un A/B crudo (que con σ≈{sigma_crudo:.0f} exige ≈{n_por_brazo(8.0, sigma_crudo):.0f}/brazo, ~10×).")
print("Un A/B 'no significativo' con MDE mayor que el efecto de interés NO prueba 'no hay efecto':")
print("significa que el test no tenía poder para verlo.")
```

**📖 Cómo se lee.** El `n` crece **cuadráticamente** al reducir el MDE: detectar la **mitad** del efecto exige ≈ **4×** la muestra. La **dispersión importa y no es única**: un **A/B crudo** (diferencia de medias sin ajustar) usa la SD **marginal** de `value` (**σ≈28-29** → del orden de **~200 por brazo** para detectar 8,0), mientras que **ajustar por covariables** (`engagement`, `tenure`) baja la σ efectiva al **ruido residual** del DGP (**σ≈8** → **≈16 por brazo**, ~10× menos). ⚠️ El `σ=8` del DGP es la SD del **ruido**, **no** la dispersión de `value`; el `n≈16` solo vale para el análisis **ajustado**, no para un A/B crudo. 💡 Un A/B «no significativo» con MDE **mayor** que el efecto de interés **no** prueba «no hay efecto»: significa que el test **no tenía poder** para verlo. La aleatorización da **ignorabilidad por diseño** (patrón oro); el A/B básico es de S02, aquí se **extiende** al marco causal.

## 12.5 — ¿Se sostiene con datos reales? Replicación DiD — Card & Krueger (1994) (Sección 4 del cuaderno)

**Paso 0 — Contexto del paper.** Card, D. & Krueger, A. B. ### Qué preguntaban los dos papers, y por qué usaron lo que usaron — Sección 0 del paper (subsección 4.0)

**💡 Antes de tocar los datos.** Este cuaderno replica **dos** estudios, y no son dos fichas paralelas: son dos respuestas distintas al **mismo obstáculo**. Sin esa articulación la réplica se vuelve mecánica —se ejecutan celdas y se obtienen números—, y lo que hay que llevarse es el **criterio para elegir un método**, no el método. *(Desarrollo completo con las citas de los originales: la ficha de la sesión de réplica del paper, «Sección 0».)*

**Ninguno de los dos equipos quiso inventar un método.** Card & Krueger abren su artículo con la pregunta de política pública: «How do employers in a low-wage labor market respond to an increase in the minimum wage? The prediction from conventional economic theory is unambiguous» (p. 1). Y la decisión que dependía del número estaba en curso: la legislatura de Nueva Jersey votó en marzo de 1992 partir el alza en dos años, y el voto quedó apenas por debajo del margen necesario para superar el veto del gobernador (p. 4). Dehejia & Wahba planteaban otra pregunta: con cuánta fidelidad un estudio **sin** experimento estima el impacto de un programa. En juego estaba el veredicto de LaLonde (1986), para quien los estimadores no experimentales resultaban «either inaccurate […], or sensitive to the specification used in the regression» (p. 1).

**El obstáculo es el mismo: comparar tratados con no tratados no responde una pregunta causal.** De cada unidad solo se observa una de sus dos historias posibles —«Only one of Yi0 or Yi1 can be observed for any unit» (p. 6)—, así que, al sustituir la historia no observada por *otro* grupo, se introduce todo lo que separa a ambos grupos por razones ajenas al tratamiento: el **confusor**. En cada estudio tiene nombre. Los restaurantes de Nueva Jersey ya eran **más pequeños** que los de Pensilvania antes del alza. Y los participantes del programa de empleo eran personas en desventaja severa, frente a un grupo de comparación tomado de encuestas «more representative of the general US population», tan distintas que «all the mean differences are statistically significant» (p. 4). La aleatorización no produce mejores datos: produce **grupos intercambiables**, y por eso allí la resta de dos promedios ya es el efecto.

**Por qué ESE método, en cada caso.** Card & Krueger no dieron por buena su elección de control: contrastaron ambos estados **antes** del alza y verificaron que los salarios iniciales «were very similar» (p. 8); además montaron un segundo control **dentro** de Nueva Jersey —los locales que ya pagaban $5,00 o más, a los que la ley no obligaba a nada— y comprobaron que se movían como Pensilvania: «The test is clearly passed.» (p. 11). Es el antecedente del drill de tendencias paralelas de la Sección 4. Dehejia & Wahba descartaron los subconjuntos ad hoc de LaLonde, creados «in an informal manner, based on one or two pre-intervention variables» (p. 5): emparejar sobre el vector completo de covariables «would encounter the curse of dimensionality» (p. 8), mientras que el propensity score es un **escalar** cuyo ajuste «is sufficient to remove bias due to all observed covariates» (Rosenbaum & Rubin, Biometrika 70(1), p. 41). El precio queda declarado: la asignación debe depender solo de variables **observadas**, y «though this is a strong assumption» (p. 2).

**Y los dos verificaron sus supuestos ANTES de estimar.** En Dehejia & Wahba, la especificación del propensity se elige **por balance** —«if there are no significant differences between the two groups, then we accept the specification» (pp. 7-8)— y antes de emparejar se recorta al soporte común, con un coste que publicaron de forma explícita: descartaron **12 611 de las 15 992** unidades de la CPS, porque «the comparison group, although very large, contains relatively few units comparable to the treatment group» (p. 9). Un grupo de control grande no es un grupo de control comparable. Es el orden que sigue este cuaderno en las secciones 4 y 5: **verificar y después estimar**.

**La articulación de la sesión.** **DiD** no necesita observar el confusor: aprovecha un cambio de política que golpea a un grupo y no a otro comparable, y supone que todo lo que los separa actúa igual en el tiempo. **PSM** sí necesita observarlo: sin un cambio externo que aprovechar, reconstruye la comparabilidad al equilibrar las covariables medidas en un único escalar. Ninguna elimina el supuesto; cada una **elige cuál asumir** —tendencias paralelas frente a selección sobre observables—, y esa elección, no la fórmula, es lo que hay que saber defender.

**⚠️ De ahí una diferencia con el paper que aparecerá en la Sección 5.** El PSM de este cuaderno obtiene **+$1.745** donde el working paper publica **+$1.582**: Dehejia & Wahba buscaron la especificación del propensity con términos de orden superior e interacciones hasta lograr balance, mientras aquí se fija una logística lineal sobre covariables estandarizadas con emparejamiento 1-NN, para que el resultado sea reproducible al pie. Es una diferencia de **protocolo**, no un error, y la conclusión que responde la pregunta se replica intacta: el ajustado **recupera el signo positivo** y supera con holgura al naive.

### Paso 3 — Estimar el DiD: tabla 2×2 y regresión (target B2, A1)

El **DiD** es la resta de los **cambios**: `DiD = (NJ_después − NJ_antes) − (PA_después − PA_antes)`. Equivale al coeficiente de la interacción `nj:post` en `FTE ~ nj + post + nj:post` (con **errores estándar clúster** por restaurante).

#### 🧮 Matemática en el cuerpo — Diferencias en Diferencias (DiD)

Con un grupo **tratado** y uno de **control**, medidos **antes** y **después**, el DiD es la **resta de los cambios**:

$$\text{DiD}=\big(\bar Y_{\text{trat,post}}-\bar Y_{\text{trat,pre}}\big)-\big(\bar Y_{\text{ctrl,post}}-\bar Y_{\text{ctrl,pre}}\big).$$

Equivale al coeficiente $\beta_3$ de la **interacción** en la regresión

$$Y_{it}=\beta_0+\beta_1\,\text{trat}_i+\beta_2\,\text{post}_t+\beta_3\,(\text{trat}_i\times\text{post}_t)+\varepsilon_{it},$$

donde $\beta_1$ capta la diferencia de **nivel** previa y $\beta_2$ la **tendencia común**. Su supuesto de identificación es **tendencias paralelas**. Card & Krueger: $(+0{,}59)-(-2{,}16)=+2{,}75$ FTE.

Fuente: el glosario de la sesión punto 12; la guía de supuestos de la sesión, Parte 3.

**❓ Qué se quiere averiguar.** ¿Subir el salario mínimo destruye empleo? Card y Krueger resolvieron esa pregunta con cuatro medias de empleo, y su respuesta contradijo lo que el libro de texto predecía.

- **Qué decide:** el signo de esta cifra decide entre dos lecturas opuestas del mercado laboral y, dentro de la sesión, decide si la réplica reproduce el paper (target B2).
- **Antes de mirar el resultado:** si el DiD resulta **claramente negativo**, el alza destruyó empleo, tal como anticipaba el modelo competitivo. Si resulta **≈ 0**, la subida no tuvo efecto medible. Si resulta **positivo**, aparece el resultado que hizo célebre al paper. El benchmark publicado es **+2,76 (SE 1,36)**: la réplica debe caer junto a él, o algo falla en la preparación de los datos.

**🔎 Qué hace este código.** Arma la **tabla 2×2** de medias FTE (antes/después × NJ/PA) y calcula el **DiD de medias** (target B2).

```
nj_b = ck.loc[ck.STATE == 1, "FTE1"].mean(); nj_a = ck.loc[ck.STATE == 1, "FTE2w"].mean()
pa_b = ck.loc[ck.STATE == 0, "FTE1"].mean(); pa_a = ck.loc[ck.STATE == 0, "FTE2w"].mean()
tabla_did = pd.DataFrame({
    "grupo": ["NJ (tratado)", "PA (control)"],
    "antes": [nj_b, pa_b], "despues": [nj_a, pa_a],
    "cambio": [nj_a - nj_b, pa_a - pa_b]})
did_means = (nj_a - nj_b) - (pa_a - pa_b)                     # target B2

print(tabla_did.round(2).to_string(index=False))
print()
print(f"[B2] DiD (medias) = {did_means:+.3f} FTE   —   benchmark publicado +2,76 (SE 1,36) [otro pipeline]")
print("Signo POSITIVO: el alza del salario mínimo NO redujo el empleo (mensaje de Card & Krueger).")
```

**🔎 Qué hace este código.** Estima el DiD por **regresión** `fte ~ nj + post + nj:post` sobre el **panel largo balanceado** (restaurantes con ambas olas), con **errores estándar clúster** por restaurante. El coeficiente de la interacción `nj:post` **es** el DiD (target A1).

```
import statsmodels.formula.api as smf

bal = ck.dropna(subset=["FTE1", "FTE2w"]).reset_index(drop=True)   # panel balanceado (ambas olas)
bal["rid"] = bal.index
largo = pd.melt(bal, id_vars=["rid", "STATE"], value_vars=["FTE1", "FTE2w"],
                var_name="ola", value_name="fte")
largo["post"] = (largo["ola"] == "FTE2w").astype(int)
largo["nj"] = largo["STATE"].astype(int)
m_did = smf.ols("fte ~ nj + post + nj:post", data=largo).fit(
    cov_type="cluster", cov_kwds={"groups": largo["rid"]})
b_int = float(m_did.params["nj:post"]); se_int = float(m_did.bse["nj:post"])          # A1

print(f"[A1] coef nj:post (regresión, panel balanceado n={len(bal)}) = {b_int:+.3f}  (SE {se_int:.3f})")
print(f"     El coeficiente de la interacción ES el DiD. Coincide con la tabla 2x2 (~{did_means:.2f}).")
print("     El coef. de 'nj' capta la diferencia de NIVEL previa; el de 'post', la tendencia común.")
```

**📖 Cómo se lee.** El coeficiente `nj:post` ≈ **+2,75** coincide con la tabla 2×2 (puede diferir décimas por usar el panel **balanceado**). El coef. de `nj` capta la diferencia de **nivel** previa y el de `post` la **tendencia común**; solo la **interacción** es el efecto causal bajo tendencias paralelas.

#### 🖐️ Cálculo manual — el DiD como resta de cambios

**🔎 Qué hace este código.** Reconstruye **de forma manual** el DiD `(Ȳ_NJ,post − Ȳ_NJ,pre) − (Ȳ_PA,post − Ȳ_PA,pre)`, y **verifica con `assert`** que iguala al DiD de la tabla 2×2 y coincide (décimas) con el coeficiente `nj:post` de la regresión.

```
# 🖐️ HAZLO A MANO: el DiD como resta de cambios, verificado contra tabla 2x2 y regresión (assert)
d_nj = nj_a - nj_b          # cambio del grupo TRATADO (NJ): después - antes
d_pa = pa_a - pa_b          # cambio del grupo CONTROL (PA): después - antes
did_mano = d_nj - d_pa      # DiD = (cambio NJ) - (cambio PA)

print(f"cambio NJ (tratado) = {d_nj:+.4f}   cambio PA (control) = {d_pa:+.4f}")
print(f"DiD a mano = ({d_nj:+.4f}) - ({d_pa:+.4f}) = {did_mano:+.4f} FTE")
print(f"DiD de medias (tabla 2x2) = {did_means:+.4f}   |   coef nj:post (regresión) = {b_int:+.4f}")

assert abs(did_mano - did_means) < 1e-9, "el DiD a mano debe igualar la resta de cambios de la tabla 2x2"
assert abs(did_mano - b_int) < 0.6, "el DiD a mano debe coincidir (décimas) con el coef. de interacción"
print("\nassert OK: (Ȳ_NJ,post−Ȳ_NJ,pre)−(Ȳ_PA,post−Ȳ_PA,pre) = 2,754 == tabla 2x2 ≈ regresión nj:post.")
```

**📖 Cómo se lee.** El DiD **no es un artificio de la regresión**: es la simple resta de dos cambios. La identidad `DiD a mano == tabla 2×2 == coef. nj:post` demuestra que el modelo `fte ~ nj + post + nj:post` solo **empaqueta** esa resta (y le añade el error estándar clúster).

### Paso 2 — PSM: propensity, solapamiento, soporte común, matching y balance (target B5, A3, Drill 3)

Se estima el **propensity** con una **logística sobre covariables estandarizadas** (`re74`/`re75` valen decenas de miles; sin escalar la logística converge mal). Se recorta al **soporte común** (rango de `e(X)` de los tratados), se empareja **1-NN** (cada tratado con su control más cercano en `e(X)`) y se estima el **ATT**. El **balance** se mide con la **diferencia estandarizada de medias (SMD)**: buscar **|SMD| < 0,1** tras emparejar.

#### 🧮 Matemática en el cuerpo — propensity score, SMD y ATT del matching

El **propensity score** es la probabilidad de tratamiento dado el perfil, modelada con una logística:

$$e(X)=P(T=1\mid X)=\frac{1}{1+e^{-X\beta}}.$$

Rosenbaum & Rubin (1983) probaron que $e(X)$ es un **balanceador**: condicionar en ese **escalar** basta para equilibrar todas las covariables de $X$. Tras emparejar cada tratado con el control de $e(X)$ más cercano, el **ATT** es

$$\widehat{\text{ATT}}=\frac{1}{n_1}\sum_{i:\,T_i=1}\big(Y_i-Y_{j(i)}\big),$$

con $j(i)$ el control emparejado. El **balance** se verifica con la **diferencia estandarizada de medias** por covariable,

$$\text{SMD}=\frac{\bar X_{\text{trat}}-\bar X_{\text{ctrl}}}{\sqrt{\tfrac12\big(s^2_{\text{trat}}+s^2_{\text{ctrl}}\big)}},\qquad \text{objetivo: } |\text{SMD}|<0{,}1.$$

Fuente: el glosario de la sesión puntos 9-10; la guía de supuestos de la sesión, Parte 4.

**❓ Qué se quiere averiguar.** ¿Alcanza con emparejar a cada tratado con un control de perfil semejante para recuperar la respuesta que ya dio el experimento?

- **Qué decide:** es la prueba de fuego del propensity score. Si funciona con LaLonde —donde el sesgo invertía el signo, de +1.794 a −8.498—, el método resulta defendible en un caso de negocio donde no hay experimento con el cual contrastar.
- **Antes de mirar el resultado:** hay dos cifras que mirar, no una. El **balance**: el `|SMD|` medio debe caer bajo el umbral **0,1** tras el emparejamiento; si no baja, no hay comparables y lo demás sobra. Y el **ATT**: si el balance mejora y el ATT se acerca a **+$1.794**, el emparejamiento reconstruyó un contrafactual creíble; si el balance mejora pero el ATT sigue lejos, entonces las covariables observadas no eran las que gobernaban la selección.

**🔎 Qué hace este código.** Ejecuta el **flujo de PSM**: estima `e(X)` con una **logística sobre covariables estandarizadas** (`max_iter=5000` concede al solver un margen holgado de iteraciones para **converger** con estas covariables escaladas; el valor por defecto podría detenerse antes de converger y emitir un aviso), recorta al **soporte común**, empareja **1-NN** (cada tratado con el control de `e(X)` más cercano), estima el **ATT** (target B5) y mide el **balance** con la **SMD** por covariable, antes y después (target A3).

```
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import NearestNeighbors

covs = ["age", "educ", "black", "hisp", "marr", "nodegree", "re74", "re75"]
X = StandardScaler().fit_transform(obs[covs].values)          # estandarizar ANTES de la logística
t = obs["treat"].to_numpy(); y = obs["re78"].to_numpy()
ps = LogisticRegression(max_iter=5000).fit(X, t).predict_proba(X)[:, 1]     # e(X) = propensity

# Soporte común: rango de e(X) de los tratados
lo, hi = ps[t == 1].min(), ps[t == 1].max()
keep = (ps >= lo) & (ps <= hi)
psc, tc, yc = ps[keep], t[keep], y[keep]
print(f"Soporte común: se conservan {int(keep.sum())} de {len(keep)} unidades "
      f"(se descartan {int(len(keep) - keep.sum())} controles fuera del solape).")

# Matching 1-NN: cada tratado con el control de e(X) más cercano
nn = NearestNeighbors(n_neighbors=1).fit(psc[tc == 0].reshape(-1, 1))
_, idx = nn.kneighbors(psc[tc == 1].reshape(-1, 1))
emparejado = yc[tc == 0][idx.flatten()]
att_psm = float((yc[tc == 1] - emparejado).mean())            # B5
print(f"[B5] PSM (1-NN, soporte común) ATT = {att_psm:+.2f} USD   —   benchmark DoWhy IPW ~1.640 [otro pipeline]")
print(f"     Recupera un ATT POSITIVO cercano al experimental (+{att_exp:.0f}) y bate al naive ({att_naive_ll:+.0f}).")

# Guardar el solapamiento (para la gráfica desde Excel)
solapamiento = pd.DataFrame({"propensity": ps, "grupo": np.where(t == 1, "tratado", "control")})

# Balance SMD por covariable, antes (vs. todos los controles) y después (vs. emparejados)
rk = obs[covs].to_numpy()[keep]
mt, mca, mcm = rk[tc == 1], rk[tc == 0], rk[tc == 0][idx.flatten()]
balance = pd.DataFrame({
    "covariable": covs,
    "smd_pre": [abs(smd(mt[:, j], mca[:, j])) for j in range(len(covs))],
    "smd_post": [abs(smd(mt[:, j], mcm[:, j])) for j in range(len(covs))]})
smd_pre = float(balance["smd_pre"].mean()); smd_post = float(balance["smd_post"].mean())   # A3
print(f"[A3] |SMD| medio  pre = {smd_pre:.3f}  ->  post = {smd_post:.3f}  (mejor si baja de 0,1)")
balance.round(3)
```

**📖 Cómo se lee.** El **PSM recupera ATT +$1.745** — positivo y cercano al experimental (+$1.794), y **supera** al naive (−$8.498). El **balance** mejora de **|SMD| medio 0,60 → 0,06**, cruzando el umbral 0,1: el matching **logró comparables**. 💡 El propensity reduce muchas covariables a un **escalar** que basta condicionar. ⚠️ Escalar `re74`/`re75` (decenas de miles) **antes** de la logística es imprescindible.

#### 🖐️ Cálculo manual — la SMD de una covariable

**🔎 Qué hace este código.** Calcula **de forma manual** la diferencia estandarizada de medias `|μ_t − μ_c| / s_pooled` de la covariable `re74` (la más desbalanceada) y **verifica con `assert`** que iguala al helper `smd()`.

```
# 🖐️ HAZLO A MANO: la SMD de una covariable |μ_t − μ_c| / s_pooled, verificada contra el helper (assert)
rk_all = obs[covs].to_numpy()[keep]        # covariables crudas en el soporte común
mt_all = rk_all[tc == 1]                    # tratados
mc_all = rk_all[tc == 0]                    # controles (antes de emparejar)
j = covs.index("re74")                       # ingresos 1974: la covariable más desbalanceada

mu_t, mu_c = mt_all[:, j].mean(), mc_all[:, j].mean()
s_pool = np.sqrt((mt_all[:, j].var(ddof=1) + mc_all[:, j].var(ddof=1)) / 2.0)
smd_re74_mano = abs((mu_t - mu_c) / s_pool)
smd_re74_lib = abs(smd(mt_all[:, j], mc_all[:, j]))

print(f"re74: μ_tratado = {mu_t:,.1f}   μ_control = {mu_c:,.1f}   s_pooled = {s_pool:,.1f}")
print(f"|SMD| a mano = {smd_re74_mano:.6f}   |   helper smd() = {smd_re74_lib:.6f}")
assert abs(smd_re74_mano - smd_re74_lib) < 1e-12, "la SMD a mano debe igualar al helper smd()"
print("\nassert OK: |μ_t − μ_c| / s_pooled a mano == la librería (misma definición de balance).")
```

**📖 Cómo se lee.** La SMD es **adimensional**: expresa la brecha de medias en unidades de desviación **combinada**, por eso se puede comparar entre covariables de escalas dispares. 💡 A diferencia del p-valor de una t (que depende del tamaño muestral), la SMD es el criterio recomendado de balance.

## 12.8 — ¿Qué decisión habilita? Laboratorio de negocio — DoWhy end-to-end (Paso 9) (Sección 6 del cuaderno)

Sobre el **caso de negocio** (DGP con efecto verdadero conocido) se aplica el flujo de DoWhy: **`CausalModel`** con el DAG → **`identify_effect`** (encuentra la puerta trasera) → **`estimate_effect`** (`backdoor.linear_regression`). Se compara **naive vs. ajustado vs. verdadero** (target B6/B7/B8).

**🔎 Qué hace este código.** Genera el **caso de negocio** (DGP, semilla fija) y calcula el **naive** = diferencia de medias de `value` entre quienes recibieron la campaña y quienes no (target B6). Muestra que los tratados tienen mayor `engagement`.

```
dfb = generar_negocio()      # DGP punto 5, autocontenido (semilla 20260719, N=5000)
print(f"Caso de negocio: {dfb.shape[0]} clientes  |  recibieron la campaña: {int(dfb.campaign.sum())}")
print(f"engagement medio   tratados = {dfb.loc[dfb.campaign == 1, 'engagement'].mean():+.2f}   "
      f"no tratados = {dfb.loc[dfb.campaign == 0, 'engagement'].mean():+.2f}   "
      f"(los más activos, más contactados -> confusión)")

naive_b = float(dfb.loc[dfb.campaign == 1, "value"].mean()
                - dfb.loc[dfb.campaign == 0, "value"].mean())          # B6
print(f"[B6] naive (diferencia de medias) = {naive_b:+.3f}   (sesgado AL ALZA por 'engagement')")
```

**📖 Cómo se lee.** El **naive ≈ +23,6** sobreestima el efecto verdadero (**8,0**): los clientes más activos (`engagement` alto) son a la vez **más contactados** y **más valiosos** — confusión al alza. Falta **cerrar la puerta trasera** `{engagement}`.

**❓ Qué se quiere averiguar.** ¿Cuánto valor añade realmente la campaña de retención, una vez descontado que se envió justamente a los clientes que ya eran más activos?

- **Qué decide en el negocio:** esta cifra sostiene el presupuesto de la campaña. Con el naive se justificaría un gasto que más tarde alguien descubriría que no rinde, y la corrección llegaría con el año fiscal cerrado.
- **Antes de mirar el resultado:** el DGP fija el efecto verdadero en **8,0** y el naive dio **≈ +23,6**. Si el ajuste por la puerta trasera `{engagement}` devuelve **≈ 8**, el cierre del confusor bastó y el procedimiento queda validado. Si quedara cerca de **23**, el ajuste no habría bloqueado el camino trasero. Si cayera **muy por debajo de 8**, se habría controlado de más —un mediador o un collider— y el efecto quedaría amputado.

**🔎 Qué hace este código.** Aplica **DoWhy end-to-end** sobre el negocio: `CausalModel` con el DAG → `identify_effect` (encuentra el conjunto de puerta trasera) → `estimate_effect` (`backdoor.linear_regression`, target B7). El argumento **`proceed_when_unidentifiable=True`** le indica a DoWhy que **continúe** aunque no pueda **garantizar formalmente** la identificación (asume que el backdoor `{engagement}` del grafo basta) en vez de detenerse a pedir confirmación. El valor del contrato **B7 es el estimador de DoWhy** y es **determinista**: para que el Excel sea **reproducible** y no dependa del entorno, esta celda **exige** `dowhy` (si no está la 0.14, **falla de forma ruidosa** en lugar de sustituir en silencio otro estimador —una OLS simple— con un número distinto).

```
# El efecto ajustado del CONTRATO (B7) es el estimador DoWhy backdoor.linear_regression:
# es DETERMINISTA y NO debe depender del entorno. Por eso se EXIGE dowhy: si no está la 0.14,
# esta celda falla de forma RUIDOSA en lugar de escribir en silencio otro estimador (p. ej. una
# OLS simple), que daría un número distinto y rompería la reproducibilidad del Excel de contrato.
from dowhy import CausalModel
grafo = ("digraph { engagement -> campaign; engagement -> value; "
         "tenure -> value; campaign -> value; }")
modelo = CausalModel(data=dfb, treatment="campaign", outcome="value", graph=grafo)
estimando = modelo.identify_effect(proceed_when_unidentifiable=True)
print("Conjunto de puerta trasera identificado por DoWhy:", estimando.get_backdoor_variables())
est = modelo.estimate_effect(estimando, method_name="backdoor.linear_regression")
adj_b = float(est.value)                                            # B7 (operativo = venv)
print(f"[B7] DoWhy backdoor.linear_regression = {adj_b:+.3f}   (operativo del venv; recupera la verdad)")
print(f"[B8] efecto VERDADERO del DGP = {TRUE_ATE:+.3f}   (constante de diseño, ancla dura)")
print()
print(f"naive {naive_b:+.2f}  >>  ajustado {adj_b:+.2f}  ≈  verdadero {TRUE_ATE:+.2f}")
print("Lectura: el naive sobreestima porque 'engagement' infla el valor de los tratados;")
print("controlar la puerta trasera {engagement} recupera el efecto real ≈ 8,0.")
```

**📖 Cómo se lee.** DoWhy identifica el backdoor **{engagement}** (confirma la lectura manual del DAG) y estima **+7,954** (Excel **B7**), que **recupera** el efecto verdadero **8,0**. La cadena `naive +23,6 ≫ ajustado +7,954 ≈ verdadero 8,0` es el núcleo pedagógico: controlar la puerta trasera correcta corrige el sesgo. ⚠️ El **operativo del contrato** es este estimador **DoWhy `backdoor.linear_regression` = +7,954** (regla del curso: **operativo = venv**); una **regresión simple** `value ~ campaign + engagement` es **otro estimador** —da un número **algo distinto** (≈ +7,85/+8,0), misma conclusión— y **no** es el valor de B7.

### Paso 10 — Refutación: pruebas de robustez (target A4, A5)

Un efecto observacional **no puede probarse**; se **refuta** sometiéndolo a pruebas que debería pasar. **Placebo treatment**: reemplazar el tratamiento por uno falso (permutado) → el efecto **debe caer a ≈0**. **Random common cause**: añadir un confusor aleatorio → el efecto **debe permanecer casi inalterado**. Pasar las refutaciones **sube la confianza**; no demuestra causalidad.

**🔎 Qué hace este código.** Somete la estimación a **dos refutaciones** de DoWhy: **placebo** (reemplaza el tratamiento por uno falso → el efecto debe caer a ≈ 0, target A4) y **random common cause** (añade un confusor aleatorio → el efecto debe quedar casi inalterado, target A5). Argumentos: **`placebo_type="permute"`** fabrica el tratamiento falso **permutando** la columna del trato (rompe su vínculo con el resultado); **`num_simulations=20`** repite la refutación 20 veces y promedia; **`random_seed=42`** fija esa aleatoriedad para que los valores sean **reproducibles**. Como el contrato **exige** `dowhy` (celda anterior), estas A4/A5 son **operativas del venv**, sin fallback manual.

```
# Refutaciones DoWhy (DETERMINISTAS por random_seed): estresan la estimación del contrato.
# placebo_type="permute": tratamiento FALSO permutando 'campaign' (rompe el vínculo con 'value')
#   -> el nuevo efecto debe caer a ~0. num_simulations=20: repeticiones que se promedian.
#   random_seed=42: fija la aleatoriedad -> A4/A5 REPRODUCIBLES. Sin fallback casero: el contrato
#   exige dowhy (celda B7), así que A4/A5 son también operativas del venv. NO escribe en el Excel aquí.
r_pl = modelo.refute_estimate(estimando, est,
                              method_name="placebo_treatment_refuter",
                              placebo_type="permute", num_simulations=20, random_seed=42)
r_cc = modelo.refute_estimate(estimando, est,
                              method_name="random_common_cause", num_simulations=20, random_seed=42)
placebo_eff = float(r_pl.new_effect); rcc_eff = float(r_cc.new_effect)     # A4, A5 (operativos)
print(f"[A4] placebo (tratamiento falso)  nuevo efecto = {placebo_eff:+.3f}   (esperado ~0)")
print(f"[A5] random common cause          nuevo efecto = {rcc_eff:+.3f}   (esperado ~{adj_b:.2f})")
print()
print("Placebo ≈ 0: al falsear el tratamiento, el efecto desaparece (buena señal).")
print("Random common cause ≈ inalterado: el efecto resiste un confusor aleatorio.")
```

**📖 Cómo se lee.** El **placebo ≈ 0**: con un tratamiento falso el efecto **desaparece** — señal **buena** (no al revés). El **random common cause ≈ +7,95**: el efecto **resiste** un confusor aleatorio. ⚠️ Pasar las refutaciones **sube la confianza**, no **prueba** causalidad (ver `SUPUESTOS_S12.md`, Parte 6).

## 12.4 en profundidad — Variables instrumentales (introducción, Paso 11) (Sección 7 del cuaderno)

Cuando hay **confusión no observada** y no hay experimento, un **instrumento** `Z` permite identificar el efecto usando **solo la variación de `T` inducida por `Z`** (dos etapas, 2SLS). Requiere tres supuestos: (i) **relevancia** (`Z` afecta a `T`), (ii) **restricción de exclusión** (`Z` no afecta a `Y` salvo a través de `T`) y (iii) **exogeneidad** (`Z` no comparte confusores con `Y`). Estima un efecto **local** (LATE, sobre los *compliers*). Ejemplo de negocio: usar la **asignación aleatoria de un cupón** (aunque no todos lo canjeen) como instrumento del **uso efectivo**. En esta sesión se **introduce**; el laboratorio se centra en DiD y PSM.

La demo siguiente (sintética, ilustrativa — **no** es una réplica) muestra la mecánica: con un confusor **no observado**, la regresión OLS queda sesgada y el **2SLS** con un instrumento válido recupera el efecto.

**🔎 Qué hace este código.** Demo **sintética e ilustrativa** (no es réplica) de **variables instrumentales**: con un confusor **no observado** `U`, el OLS queda sesgado y el **2SLS** con un instrumento válido `Z` (asignación aleatoria de un cupón) recupera el efecto.

```
from linearmodels.iv import IV2SLS

rng_iv = np.random.default_rng(2026)
Ni = 4000
U = rng_iv.normal(0, 1, Ni)                                       # confusor NO observado
Z = rng_iv.binomial(1, 0.5, Ni)                                  # instrumento (asignación del cupón)
D = ((0.8 * Z + 0.7 * U + rng_iv.normal(0, 1, Ni)) > 0.5).astype(float)   # tratamiento (uso efectivo)
beta = 3.0                                                        # efecto causal verdadero
Yiv = 1.0 + beta * D + 2.0 * U + rng_iv.normal(0, 1, Ni)          # U afecta a Y (confusión no observada)

dat = pd.DataFrame({"Yiv": Yiv, "D": D, "Z": Z})
ols = IV2SLS.from_formula("Yiv ~ 1 + D", dat).fit()              # OLS (ignora U -> sesgado)
iv = IV2SLS.from_formula("Yiv ~ 1 + [D ~ Z]", dat).fit()        # 2SLS con instrumento Z
print(f"Efecto verdadero      = {beta:+.2f}")
print(f"OLS (sesgado por U)   = {float(ols.params['D']):+.2f}")
print(f"2SLS (instrumento Z)  = {float(iv.params['D']):+.2f}")
print("Lectura: el instrumento recupera el efecto AUN con confusión no observada; el OLS no.")
```

**📖 Cómo se lee.** El **OLS** se aleja del efecto verdadero (+3) por la confusión no observada; el **2SLS** lo recupera usando solo la variación de `D` **inducida por `Z`**. 💡 IV es la palanca cuando la ignorabilidad es insostenible; estima un efecto **local** (LATE). En esta sesión se **introduce** (los tres supuestos: relevancia, exclusión, exogeneidad); el ajuste 2SLS fino es material avanzado (`SUPUESTOS_S12.md`, Parte 7.2).

## 12.8 (continúa) — Del efecto a la decisión con incertidumbre (Paso 12, entregable) (Sección 8 del cuaderno)

La recomendación combina tres piezas: (1) **magnitud** del efecto frente al **umbral de negocio** (¿el efecto supera el costo por cliente?); (2) **incertidumbre** (¿el IC 95 % excluye el umbral?); (3) **credibilidad causal** (soporte común, balance, refutaciones superadas). Si las tres se alinean → **escalar**; si el IC incluye el umbral → **A/B confirmatorio** antes de invertir a escala.

**❓ Qué se quiere averiguar.** ¿Se escala la campaña a toda la base de clientes, o no?

- **Qué decide en el negocio:** la inversión. Cada cliente contactado cuesta, y el umbral de rentabilidad del caso es **5,00 USD por cliente**: por debajo de esa cifra la campaña destruye margen aunque el efecto sea real y estadísticamente significativo.
- **Antes de mirar el resultado:** si el extremo **inferior** del IC 95 % queda por encima de **5**, corresponde escalar. Si el extremo **superior** queda por debajo de **5**, corresponde no escalar. Si el IC **contiene el 5**, ninguna de las dos decisiones está respaldada y toca un A/B confirmatorio: la pregunta del negocio no es si el efecto existe, sino si supera el costo con margen suficiente.

**🔎 Qué hace este código.** Traduce el efecto ajustado en una **decisión**: estima el efecto con su **IC 95 %** (`value ~ campaign + engagement + tenure`) y lo compara con un **umbral de rentabilidad** por cliente para recomendar **escalar / no escalar / A/B confirmatorio**.

```
m_adj = smf.ols("value ~ campaign + engagement + tenure", data=dfb).fit()
ic = m_adj.conf_int().loc["campaign"]
ic_lo, ic_hi = float(ic[0]), float(ic[1])
umbral = 5.0     # costo / umbral de rentabilidad por cliente (USD) — supuesto de negocio

print(f"Efecto ajustado (backdoor) = {float(m_adj.params['campaign']):+.2f}  "
      f"IC 95% = [{ic_lo:.2f}, {ic_hi:.2f}]")
print(f"Umbral de rentabilidad = {umbral:.2f} USD por cliente")
print()
if ic_lo > umbral:
    print("DECISIÓN: ESCALAR. El IC 95% está por ENCIMA del umbral y las refutaciones se superaron.")
elif ic_hi < umbral:
    print("DECISIÓN: NO escalar. El IC 95% está por DEBAJO del umbral.")
else:
    print("DECISIÓN: A/B CONFIRMATORIO. El IC incluye el umbral: la evidencia observacional no basta.")
print("Recordatorio: el efecto es OBSERVACIONAL (supuestos de identificación), no experimental.")
```

**📖 Cómo se lee.** La recomendación combina **magnitud** (¿supera el umbral?), **incertidumbre** (¿el IC excluye el umbral?) y **credibilidad causal** (soporte común, balance, refutaciones). Si el IC **incluye** el umbral, la acción prudente es un **A/B confirmatorio** antes de invertir a escala. ⚠️ El efecto es **observacional** (supuestos de identificación), no experimental (`SUPUESTOS_S12.md`, Parte 7.4).

## 12.3 en profundidad — Construcción desde cero (descomposición del alumno) (Sección 12 del cuaderno)

Para probar que se entiende el **flujo de PSM** —no solo la llamada a la librería— se reconstruye el pipeline completo **sin el helper `smd()`** y se reproduce el contrato:
- **🧱 Flujo de PSM:** estandarizar covariables → estimar el **propensity `e(X)`** → recortar al **soporte común** → **emparejar 1-NN** → **verificar el balance (SMD)** → estimar el **ATT**. Partiendo del **naive −$8.498**, recupera **+$1.745** (== Excel B5).

**🔎 Qué hace este código.** Ejecuta ese flujo paso a paso sobre el montaje observacional (`obs`), computa la SMD **inline** (sin helper) y **verifica con `assert`** que reproduce el ATT del contrato, que **supera** al naive y que el balance cae bajo 0,1.

```
# 🧱 CONSTRUYE DESDE CERO: el flujo de PSM sin el helper smd() (propensity → soporte común → 1-NN → balance → ATT). NO toca el Excel.
cd_xl2 = pd.read_excel(XLSX, sheet_name="causal_did")
psm_contrato, naive_contrato = float(cd_xl2["valor"].iloc[3]), float(cd_xl2["valor"].iloc[2])   # B5, B4

Xs = StandardScaler().fit_transform(obs[covs].values)                              # 1) estandarizar
tt, yy = obs["treat"].to_numpy(), obs["re78"].to_numpy()
e_hat = LogisticRegression(max_iter=5000).fit(Xs, tt).predict_proba(Xs)[:, 1]      # 2) propensity e(X)
lo_s, hi_s = e_hat[tt == 1].min(), e_hat[tt == 1].max()                            # 3) soporte común
k_s = (e_hat >= lo_s) & (e_hat <= hi_s)
e_s, t_s, y_s = e_hat[k_s], tt[k_s], yy[k_s]
nn_s = NearestNeighbors(n_neighbors=1).fit(e_s[t_s == 0].reshape(-1, 1))           # 4) emparejar 1-NN
_, ix_s = nn_s.kneighbors(e_s[t_s == 1].reshape(-1, 1))
att_cero = float((y_s[t_s == 1] - y_s[t_s == 0][ix_s.flatten()]).mean())           # 5) ATT

# balance SMD inline (a mano, sin el helper smd())
rk_s = obs[covs].to_numpy()[k_s]
A_t, C_all, C_match = rk_s[t_s == 1], rk_s[t_s == 0], rk_s[t_s == 0][ix_s.flatten()]
def _smd_inline(a, b):
    sp = np.sqrt((a.var(ddof=1) + b.var(ddof=1)) / 2.0)
    return abs((a.mean() - b.mean()) / sp) if sp > 0 else 0.0
smd_pre_cero = np.mean([_smd_inline(A_t[:, j], C_all[:, j]) for j in range(len(covs))])
smd_post_cero = np.mean([_smd_inline(A_t[:, j], C_match[:, j]) for j in range(len(covs))])

print(f"naive de partida = {naive_contrato:+.2f} USD  ->  PSM reconstruido = {att_cero:+.2f} USD")
print(f"|SMD| medio  pre = {smd_pre_cero:.3f}  ->  post = {smd_post_cero:.3f}  (umbral < 0,1)")
assert abs(att_cero - psm_contrato) < 1e-6,  "el PSM reconstruido debe reproducir el contrato (Excel B5)"
assert att_cero > naive_contrato + 5000,     "el PSM debe batir con holgura al naive (recupera el efecto)"
assert smd_post_cero < 0.1 and smd_post_cero < smd_pre_cero, "el matching debe lograr balance (|SMD|<0,1)"
print(f"\nassert OK: partiendo del naive {naive_contrato:+.0f}, el PSM reconstruido recupera {att_cero:+.0f} == Excel; balance {smd_pre_cero:.2f}→{smd_post_cero:.2f}.")
```

**📖 Cómo se lee.** El pipeline reconstruido de forma manual —estandarizar, estimar el propensity, recortar, emparejar y promediar— reproduce **exactamente** el ATT del contrato (+$1.745) y el balance (0,60 → 0,06), partiendo del naive −$8.498. Entender el flujo, no solo llamarlo, es lo que permite **diagnosticar** cuándo el PSM falla.

**🔎 Qué hace este código (Diagnóstico 13.4 — collider vs. mediador).** Demo **autocontenida**: genera un DGP con un **collider `uso_post`** (`campaign → uso_post ← value`) y un **mediador `visitas`** (`campaign → visitas → value`), y compara el efecto estimado con el backdoor correcto `{engagement}` frente a controlar **de más**. No escribe en el Excel.

```
# Diagnóstico 13.4 (collider vs. mediador): controlar de MÁS hace daño. Demo autocontenida. NO escribe en el Excel.
rng_d = np.random.default_rng(4321)
nd = 6000
eng_d = rng_d.normal(0, 1, nd)                                        # confusor
camp_d = rng_d.binomial(1, 1 / (1 + np.exp(-(0.9 * eng_d - 0.3))))    # campaña (selección por engagement)
visitas = 1.5 * camp_d + 0.5 * eng_d + rng_d.normal(0, 1, nd)         # MEDIADOR: campaign -> visitas -> value
value_d = 50 + 2.0 * camp_d + 4.0 * visitas + 20.0 * eng_d + rng_d.normal(0, 8, nd)   # efecto total = 2 + 4·1,5 = 8
uso_post = 0.8 * camp_d + 0.10 * value_d + rng_d.normal(0, 1, nd)     # COLLIDER: campaign -> uso_post <- value
demo = pd.DataFrame({"eng": eng_d, "camp": camp_d, "visitas": visitas, "value": value_d, "uso_post": uso_post})

b_backdoor = float(smf.ols("value ~ camp + eng", data=demo).fit().params["camp"])             # correcto {engagement}
b_collider = float(smf.ols("value ~ camp + eng + uso_post", data=demo).fit().params["camp"])  # + collider uso_post
b_mediador = float(smf.ols("value ~ camp + eng + visitas", data=demo).fit().params["camp"])   # + mediador visitas

print(f"efecto total verdadero              = 8,00")
print(f"backdoor {{engagement}} (correcto)    = {b_backdoor:+.3f}")
print(f"+ controlar el COLLIDER 'uso_post'  = {b_collider:+.3f}   (se ALEJA de 8: fabrica sesgo)")
print(f"+ controlar el MEDIADOR 'visitas'   = {b_mediador:+.3f}   (SUBESTIMA: bloquea el canal indirecto)")
assert abs(b_backdoor - 8.0) < 1.0, "el backdoor {engagement} recupera el efecto total ≈ 8"
assert abs(b_collider - 8.0) > abs(b_backdoor - 8.0) + 1.0, "condicionar en el collider ABRE sesgo"
assert b_mediador < b_backdoor - 3.0, "controlar el mediador subestima el efecto total"
print("\nassert OK: solo el backdoor {engagement} identifica el efecto; el collider 'uso_post' y el")
print("mediador 'visitas' —ambos posteriores a la campaña— NO se controlan (ver SUPUESTOS_S12.md, Parte 5).")
```

**📖 Cómo se lee.** Solo el backdoor **{engagement}** recupera el efecto total (**≈ 8**). Añadir el **collider `uso_post`** —posterior a la campaña— **abre** un camino espurio y **aleja** el efecto de 8; añadir el **mediador `visitas`** **bloquea** el canal indirecto y **subestima** el efecto total. 💡 El reflejo de «incluir todas las variables en el modelo» es **incorrecto y a veces peligroso**: la regla es controlar **solo** el conjunto de puerta trasera.

**🔎 Qué hace este código (Diagnóstico 13.5 — refutaciones).** Relee los resultados de las dos refutaciones de DoWhy (placebo y random common cause) ya calculadas y **verifica con `assert`** que el placebo anula el efecto (≈ 0) y que el efecto **resiste** un confusor aleatorio (≈ inalterado). No escribe en el Excel.

```
# Diagnóstico 13.5 (refutaciones DoWhy): estresar la estimación — placebo ≈ 0, random common cause ≈ inalterado. NO escribe en el Excel.
print(f"efecto ajustado (backdoor)           = {adj_b:+.3f}")
print(f"[placebo] tratamiento falso          -> nuevo efecto = {placebo_eff:+.3f}   (esperado ≈ 0)")
print(f"[random common cause] confusor azar  -> nuevo efecto = {rcc_eff:+.3f}   (esperado ≈ {adj_b:.2f})")
assert abs(placebo_eff) < 1.5, "el placebo debe anular el efecto (≈ 0): señal BUENA"
assert abs(rcc_eff - adj_b) < 1.5, "el efecto debe resistir un confusor aleatorio (≈ inalterado)"
print("\nassert OK: placebo ≈ 0 (el efecto real se desvanece con tratamiento falso) y el efecto resiste")
print("un confusor aleatorio. Refutar SUBE la confianza, no prueba causalidad (ver SUPUESTOS_S12.md, Parte 6).")
```

**📖 Cómo se lee.** El **placebo ≈ 0** es la señal **buena**: al falsear el tratamiento, el efecto se desvanece (⚠️ **no** significa «no hay efecto real», es al revés). El **random common cause ≈ inalterado** muestra estabilidad de la especificación. Pasar las refutaciones **sube la confianza**; **fallar** cualquiera indica que hay que revisar el DAG, las covariables o la positividad.
