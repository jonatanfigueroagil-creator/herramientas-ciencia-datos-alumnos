### Intuición del valor p (sin fórmula, sin H₀ formal) (subsección 0.2)

Si en realidad no pasara nada —no hay diferencia real entre dos grupos, no hay relación real entre dos variables—, ¿qué tan sorprendente sería un resultado tan extremo como el que se observó? Eso es lo que responde el p-valor: un valor pequeño dice que el resultado sería raro si no pasara nada; uno grande dice que es compatible con que no pase nada. Aquí **no se calcula ningún p-valor** y **no se define H0 formalmente** — eso es contenido de la S02.

**Fuente:** Wasserstein, R. L. & Lazar, N. A. (2016). *The ASA Statement on Statistical Significance and P-Values*. The American Statistician 70(2):129–133 — misma fuente y misma idea central que usa el glosario de la sesión. Ver también el material de referencia de la sesión, punto 7.

**Cierre del Bloque 0.** El cómputo formal de estas medidas, las pruebas de hipótesis y el diseño de un experimento A/B se enseñan en la **S02**. ## 1.2 — ¿Qué es un proyecto de datos y en qué se distingue de un reporte? El proceso KDD (Sección 3 del cuaderno)

Fayyad, Piatetsky-Shapiro y Smyth (1996) formalizaron el descubrimiento de conocimiento en bases de datos como un proceso de cinco etapas, donde la «minería de datos» es solo una de ellas:

1. **Selección** — elegir los datos relevantes para el objetivo planteado.
2. **Preprocesamiento** — limpiar los datos: faltantes, duplicados, inconsistencias.
3. **Transformación** — dar a los datos la forma que exige el análisis (codificación, agregación, nuevas variables).
4. **Minería de datos** — aplicar el algoritmo que extrae los patrones.
5. **Interpretación / evaluación** — traducir los patrones a conocimiento útil y validarlo con el negocio.

Idea central: el valor no está en el algoritmo aislado sino en el proceso completo. La celda siguiente dibuja el flujo.

**❓ Qué se quiere averiguar.** ¿Qué parte del trabajo de un proyecto de datos ocupa realmente el algoritmo?

- **Qué decide:** cómo se reparten el esfuerzo, el presupuesto y el equipo. La expectativa habitual —«contratemos a alguien que sepa modelos y el proyecto sale»— se sostiene o se cae con la respuesta.
- **Antes de mirar el resultado:** si la minería de datos fuese la mayor parte del recorrido, el resto sería preparación menor y el proyecto se jugaría en la elección del algoritmo. Si ocupa **una sola de las cinco etapas**, cuatro quintas partes del proceso pasan antes y después del modelo — y es en esas fases donde se decide el éxito o el fracaso de un proyecto.

🔎 **Qué hace este código.** La celda siguiente dibuja el proceso KDD como cinco cajas encadenadas por flechas, de *Selección* a *Interpretación / Evaluación*. No calcula nada: es un diagrama de apoyo cuyo fin es fijar visualmente que la minería de datos ocupa **una sola** de las cinco etapas y que el valor surge del proceso completo, no del algoritmo aislado.

📄 **Procedencia.** El esquema de cinco etapas proviene de Fayyad, Piatetsky-Shapiro y Smyth (1996), «From Data Mining to Knowledge Discovery in Databases», *AI Magazine* 17(3), pp. 37–54.

```python
import matplotlib.pyplot as plt

etapas = [
    "Selección",
    "Preprocesamiento",
    "Transformación",
    "Minería\nde datos",
    "Interpretación /\nEvaluación",
]

fig, ax = plt.subplots(figsize=(11, 2.8))
ax.set_xlim(0, 11)
ax.set_ylim(0, 3)
ax.axis("off")  # sin ejes: es un esquema, no una gráfica con escalas

for i, etapa in enumerate(etapas):
    x = 0.4 + i * 2.15  # x de cada caja: ancho 1.8 + separación 0.35 = paso 2.15
    caja = plt.Rectangle((x, 1.0), 1.8, 1.0, facecolor="#FCEDEE",
                         edgecolor="#9B1B30", linewidth=1.5)
    ax.add_patch(caja)
    ax.text(x + 0.9, 1.5, etapa, ha="center", va="center",  # +0.9, +0.5 = centro de la caja
            fontsize=10, color="#2D2D2D")
    if i < len(etapas) - 1:  # flecha hacia la siguiente caja (todas salvo la última)
        # xytext = borde derecho de esta caja (x+1.8); xy = borde izquierdo de la siguiente (x+2.15)
        ax.annotate("", xy=(x + 2.15, 1.5), xytext=(x + 1.8, 1.5),
                    arrowprops=dict(arrowstyle="->", color="#6E6E6E", lw=1.8))

ax.set_title("Proceso KDD (Fayyad, Piatetsky-Shapiro y Smyth, 1996)",
             fontsize=12, color="#2D2D2D")
plt.tight_layout()
plt.show()
```

📖 **Cómo leer este diagrama.** Se recorre de izquierda a derecha: cada flecha indica que la etapa siguiente **depende** de la anterior. La lectura clave es de proporción, no de tamaño de las cajas: aunque «Minería de datos» suele acaparar la atención, es apenas un quinto del recorrido; las cuatro etapas restantes (seleccionar, limpiar, transformar e interpretar) son las que convierten un patrón en conocimiento útil para el negocio.

⚠️ **Riesgo — GIGO (garbage in, garbage out) / calidad de datos.** Si los datos que entran en «Selección» y «Preprocesamiento» arrastran errores, faltantes o duplicados, ningún algoritmo posterior los corrige: el modelo aprenderá y amplificará esos defectos. Por eso la calidad se audita **antes** de modelar, no después.

## 1.2 (continúa) — CRISP-DM: la hoja de ruta del curso (Sección 4 del cuaderno)

CRISP-DM (*Cross-Industry Standard Process for Data Mining*; Chapman et al., 2000) organiza cualquier proyecto de datos en **seis fases iterativas**:

1. **Comprensión del negocio** — objetivos, decisión a habilitar, criterios de éxito.
2. **Comprensión de los datos** — inventario de fuentes, exploración inicial, calidad.
3. **Preparación de los datos** — limpieza, integración y construcción de variables. Aquí suele concentrarse la mayor parte del esfuerzo del proyecto: **entre ~45 % y ~80 %** del tiempo según la encuesta que se cite (la *Anaconda State of Data Science* lo ubica cerca del **45 %**; la cifra clásica de la industria ronda el **80 %**).
4. **Modelado** — selección y ajuste de las técnicas analíticas.
5. **Evaluación** — verificar que el modelo responde la pregunta de negocio y que sus conclusiones son válidas.
6. **Despliegue** — llevar el resultado a la decisión o a producción, y monitorearlo.

Las fases no son lineales: es normal volver atrás (de modelado a preparación de datos, o de evaluación a comprensión del negocio). El diagrama siguiente muestra el ciclo.

💡 **Intuición — KDD frente a CRISP-DM.** Describen el mismo viaje del dato al conocimiento, pero con acentos distintos: KDD (académico, 1996) detalla las **etapas técnicas** sobre los datos, mientras que CRISP-DM (industrial, 2000) envuelve ese trabajo con el **marco de negocio** al inicio (comprensión del negocio) y el **despliegue** al final. En el curso, KDD explica «qué se le hace a los datos» y CRISP-DM ordena «cómo se gestiona el proyecto».

📄 **Procedencia.** Las seis fases son de Chapman, Clinton, Kerber, Khabaza, Reinartz, Shearer y Wirth (2000), «CRISP-DM 1.0: Step-by-step data mining guide», SPSS / consorcio CRISP-DM.

📄 **Procedencia de la cifra de esfuerzo.** El rango ~45 %–~80 % combina dos encuestas distintas: la *Anaconda State of Data Science* (2020) ubica la carga de carga y limpieza de datos cerca del **45 %**, mientras que la cifra clásica de la industria —difundida por el *CrowdFlower Data Science Report* (2016) y notas de prensa afines— la sitúa en torno al **80 %**. Se cita como rango porque cada encuesta mide de forma diferente.

**❓ Qué se quiere averiguar.** ¿Un proyecto de datos avanza en línea recta de la pregunta al resultado, o vuelve sobre sus propios pasos?

- **Qué decide:** cómo se planifica el trabajo y qué fecha se compromete. Una secuencia de seis etapas encadenadas se contrata como una obra con hitos cerrados; un ciclo con retornos se contrata por iteraciones y reserva holgura para rehacer.
- **Antes de mirar el resultado:** si el diagrama fuese una fila de cajas con una sola dirección, cada fase se cerraría antes de abrir la siguiente. Si cierra un círculo con flechas de retorno, un hallazgo en modelado obliga a volver a los datos —o incluso a replantear la pregunta de negocio— y el cronograma debe preverlo desde el inicio.

🔎 **Qué hace este código.** La celda siguiente dibuja las **seis fases de CRISP-DM** como un ciclo de círculos recorrido en sentido horario, con «Datos» en el centro y una anotación sobre la fase de preparación. Las flechas curvas remarcan que el proceso es **iterativo**: es normal volver de «Modelado» a «Preparación de los datos» o de «Evaluación» a «Comprensión del negocio».

```python
import numpy as np
import matplotlib.pyplot as plt

fases = [
    "Comprensión\ndel negocio",
    "Comprensión\nde los datos",
    "Preparación\nde los datos",
    "Modelado",
    "Evaluación",
    "Despliegue",
]
n = len(fases)
# Ángulo de cada fase: se parte de 90° (arriba) y se restan pasos iguales de 360/n
# grados; restar (no sumar) recorre el círculo en sentido horario.
angulos = [90 - i * 360 / n for i in range(n)]
radio = 1.0
# Polar -> cartesiano: cada fase se ubica sobre un círculo de radio 1 (cos = x, sin = y).
coords = [(radio * np.cos(np.radians(a)), radio * np.sin(np.radians(a)))
          for a in angulos]

fig, ax = plt.subplots(figsize=(7.5, 7.5))
ax.set_xlim(-1.7, 1.7)
ax.set_ylim(-1.7, 1.7)
ax.set_aspect("equal")  # círculo real, no elipse
ax.axis("off")

for (x, y), fase in zip(coords, fases):
    borde = "#E4002B" if fase.startswith("Preparación") else "#9B1B30"  # resalta la fase clave
    ax.add_patch(plt.Circle((x, y), 0.30, facecolor="#FCEDEE",
                            edgecolor=borde, linewidth=2))
    ax.text(x, y, fase, ha="center", va="center", fontsize=9, color="#2D2D2D")

# Flechas entre fases consecutivas (ciclo en sentido horario)
for i in range(n):
    x1, y1 = coords[i]
    x2, y2 = coords[(i + 1) % n]  # % n cierra el ciclo: de la última vuelve a la primera
    dx, dy = x2 - x1, y2 - y1
    d = float(np.hypot(dx, dy))
    ux, uy = dx / d, dy / d  # vector unitario de un centro al siguiente
    # 0.36 ≈ radio del círculo (0.30) + holgura: la flecha arranca y termina en el
    # borde de cada círculo, no en su centro, para no pisar el texto.
    ax.annotate("", xy=(x2 - ux * 0.36, y2 - uy * 0.36),
                xytext=(x1 + ux * 0.36, y1 + uy * 0.36),
                # connectionstyle "arc3,rad=-0.25": curva la flecha (rad<0 la abomba hacia
                # afuera del círculo) para sugerir el giro del ciclo en vez de una recta.
                arrowprops=dict(arrowstyle="-|>", color="#6E6E6E", lw=1.8,
                                connectionstyle="arc3,rad=-0.25"))

ax.text(0, 0, "Datos", ha="center", va="center", fontsize=12,
        color="#6E6E6E", style="italic")

# Nota del esfuerzo en preparación de datos (nodo índice 2).
# xy = punto señalado (junto al nodo "Preparación"); xytext = dónde se coloca el texto (más abajo).
xp, yp = coords[2]
ax.annotate("≈45–80% del esfuerzo\ndel proyecto", xy=(xp + 0.22, yp - 0.22),
            xytext=(xp + 0.10, yp - 0.88), fontsize=9, color="#E4002B",
            ha="center", arrowprops=dict(arrowstyle="->", color="#E4002B"))

ax.set_title("Ciclo CRISP-DM: seis fases iterativas", fontsize=12, color="#2D2D2D")
plt.tight_layout()
plt.show()
```

📖 **Cómo leer este diagrama.** El círculo se lee en sentido horario desde «Comprensión del negocio», pero **no es una línea de montaje**: las flechas cierran un ciclo y en la práctica se salta hacia atrás cuando una fase revela un problema de la anterior. «Datos» está en el centro porque toda fase los consulta o los modifica. La fase resaltada («Preparación de los datos») es la que más tiempo consume —de ahí la anotación del rango ~45 %–~80 %—, un recordatorio de dónde se irá realmente el esfuerzo del proyecto.

## 1.4 — ¿Cómo se identifica el tipo de problema que se enfrenta? Taxonomías del análisis (Sección 5 del cuaderno)

### Por el tipo de pregunta que responden

- **Descriptivo** — ¿qué pasó y cómo se ven los datos? Resume, agrupa y caracteriza (reportes, segmentaciones, patrones).
- **Predictivo** — ¿qué va a pasar? Estima un valor o una probabilidad futura para cada unidad de análisis.
- **Prescriptivo** — ¿qué conviene hacer? Recomienda una acción que combina predicciones con reglas de negocio u optimización.

### Por la forma de aprender de los datos

- **Supervisado** — existe una variable objetivo con ejemplos históricos etiquetados; el modelo aprende a predecirla.
- **No supervisado** — no hay variable objetivo; se buscan estructuras ocultas (grupos, asociaciones, dimensiones latentes).
- **Por refuerzo** — un agente aprende mediante prueba y error, con recompensas; se menciona por completitud y queda fuera del alcance del curso.

### Dos ejes ortogonales, no equivalentes

Los dos ejes son **independientes**: el primero clasifica por lo que la pregunta *pide* (describir, predecir o prescribir) y el segundo por *cómo* aprende el modelo (si existe o no una variable objetivo etiquetada). Por eso cada problema recibe **una etiqueta de cada eje**, y no toda tarea descriptiva es no supervisada: un reporte de ventas es **descriptivo y sin aprendizaje** (pura agregación), mientras que una segmentación es **descriptiva y no supervisada**. Confundir «descriptivo» con «no supervisado» es un error frecuente que conviene evitar desde el primer día.

La celda siguiente clasifica seis ejemplos de negocio según ambos ejes y hace visible esa independencia.

**❓ Qué se quiere averiguar.** ¿Saber qué tipo de pregunta responde un análisis —descriptiva, predictiva o prescriptiva— basta para saber cómo aprende el modelo?

- **Qué decide:** qué se le pide al equipo y qué datos hay que reunir. Un enfoque supervisado exige una variable objetivo etiquetada; uno no supervisado no la tiene. Confundir los dos ejes lleva a pedir datos que no existen, o a prometer una predicción donde solo cabe una descripción.
- **Antes de mirar el resultado:** si cada tipo de pregunta arrastrase siempre la misma forma de aprender, un eje sobraría y bastaría con clasificar por pregunta. Si entre las filas **descriptivas** aparecen etiquetas distintas del segundo eje, los dos ejes son independientes y todo caso necesita las dos etiquetas para quedar bien planteado.

## 1.3 — ¿Cómo se traduce una pregunta de negocio en una pregunta analítica? (Sección 6 del cuaderno)

Todo proyecto de datos serio recorre la cadena **decisión → pregunta de negocio → pregunta analítica → KPI → unidad de análisis → variable objetivo**:

- **Decisión** — la acción concreta que alguien tomará con el resultado; si nada cambia con el análisis, el proyecto no tiene valor.
- **KPI** — la métrica de negocio con la que se medirá el éxito de esa decisión.
- **Unidad de análisis** — el «quién» o «qué» sobre el que se describe o predice (cliente, tienda, producto–semana).
- **Variable objetivo** — la columna que el modelo aprende a predecir; solo existe en los enfoques supervisados.

La celda siguiente aplica la cadena a los tres casos de apertura de la sesión.

💡 **Intuición — de la pregunta de negocio a la analítica.** El salto clave del curso es dejar de preguntar «¿por qué se van los clientes?» (conversación) y pasar a «¿qué clientes cancelarán en los próximos 90 días?» (pregunta analítica con **unidad de análisis**, **ventana temporal** y **variable objetivo** medibles). Una pregunta bien traducida ya contiene, en su redacción, la columna que el modelo deberá predecir.

**❓ Qué se quiere averiguar.** ¿Todo proyecto de datos tiene una variable objetivo que predecir?

- **Qué decide:** si el problema se puede plantear como una predicción o si es de otra naturaleza. De la respuesta dependen los datos que hay que reunir y la forma de evaluar el trabajo: donde no hay variable objetivo, no hay error de predicción que medir ni acierto que reportar.
- **Antes de mirar el resultado:** si los tres casos —churn, segmentación y forecasting— trajesen su variable objetivo, todo proyecto sería supervisado y la cadena terminaría siempre igual. Si alguno queda **sin** variable objetivo, la cadena decisión → KPI → unidad de análisis se sostiene sin ella, y ese proyecto se evalúa por otro camino.

## 1.6 — ¿De dónde provienen los datos y qué confiabilidad ofrecen? Fuentes y riesgos éticos (Sección 7 del cuaderno)

### Fuentes internas de una organización

- Sistemas transaccionales: ERP, facturación, punto de venta (POS).
- CRM y canales de atención al cliente (tickets, llamadas, chats).
- Web y aplicaciones: *clickstream*, logs de navegación y de uso.
- Operaciones y sensores (IoT), logística, recursos humanos y finanzas.

### Fuentes externas

- Datos abiertos de gobiernos y organismos multilaterales.
- APIs comerciales y redes sociales.
- Repositorios académicos y de referencia (UCI, Rdatasets, Kaggle).
- Datos adquiridos a terceros (paneles, burós, geodatos).

### Riesgos éticos y de privacidad

- **Datos personales**: requieren consentimiento, minimización y anonimización, conforme a la normativa local de protección de datos personales.
- **Sesgo histórico**: si los datos reflejan discriminaciones pasadas, el modelo puede aprenderlas y amplificarlas.
- **Reidentificación**: cruzar fuentes «anónimas» puede volver identificables a las personas.
- **Uso secundario no consentido**: emplear datos recolectados para un fin distinto del autorizado.
- **Opacidad**: decisiones automatizadas que el afectado no puede entender ni apelar.

Regla práctica: en la fase de comprensión del negocio debe preguntarse **quién puede resultar afectado** por el uso de los datos, y documentarlo en la ficha del proyecto.

⚠️ **Riesgo — sesgo de muestreo.** Que los datos existan no significa que **representen** a la población de interés. Una encuesta de satisfacción respondida por el 8 % de los clientes recoge sobre todo a los muy satisfechos o muy molestos; y si solo se registran las compras hechas con tarjeta de fidelidad, el comportamiento de quien no la presenta queda invisible. Un modelo entrenado sobre esa muestra sesgada generaliza mal.

⚠️ **Error de interpretación frecuente — correlación no es causalidad.** Un modelo predictivo puede descubrir que una variable acompaña al abandono sin **causarlo**; actuar sobre ella no cambiará el resultado. Distinguir predecir de explicar el porqué es el terreno de la inferencia causal (S12): predecir quién se irá y **decidir qué hacer para evitarlo** son problemas distintos.

## 1.7 — ¿Cómo se evalúa la viabilidad de un proyecto antes de iniciarlo?

Un proyecto de datos se detiene **antes** de empezar cuando falla cualquiera de tres condiciones. No son preferencias metodológicas: cada una, por sí sola, basta para que el resultado nunca llegue a usarse. Evaluarlas cuesta una reunión; descubrirlas a mitad del proyecto cuesta el proyecto entero.

### 1.7.1 ¿Se dispone de datos suficientes y accesibles?

Son dos exigencias distintas y ambas deben cumplirse.

**Suficientes** no se refiere al número de filas sino a la **cobertura**: que los datos contengan el fenómeno que se quiere modelar, con suficientes casos del evento raro. Un modelo de fraude entrenado sobre un histórico con tres fraudes confirmados no aprende a detectar fraude; aprende a decir «no hay fraude», y acierta el 99,9 % de las veces sin servir para nada. La misma lógica aplica al horizonte temporal: pronosticar demanda estacional exige al menos dos ciclos completos, porque con uno solo no hay forma de distinguir la estación de la tendencia.

**Accesibles** se refiere a que el equipo pueda leerlos **a tiempo y con permiso**. Un dato que existe en un sistema al que nadie del equipo tiene acceso, o que el área propietaria entrega en tres meses, equivale a un dato que no existe. Aquí se incluyen las restricciones legales de la sesión anterior: si el dato es personal y no hay base legal para tratarlo, la respuesta es no.

### 1.7.2 ¿Resulta medible el KPI seleccionado?

Un KPI es medible cuando existe una **definición operativa** que dos personas distintas aplican por separado y obtienen el mismo número. «Mejorar la satisfacción del cliente» no lo es; «reducir en dos puntos el porcentaje de tickets reabiertos dentro de los siete días» sí.

Tres condiciones lo hacen operativo:

- **Fórmula explícita**: qué se cuenta en el numerador y qué en el denominador. La «tasa de abandono» cambia por completo según se cuente sobre clientes activos al inicio del mes o sobre el total de la cartera.
- **Ventana temporal fijada**: el mismo indicador medido a 30 y a 90 días describe dos fenómenos diferentes.
- **Línea base conocida**: sin el valor actual no hay forma de afirmar que el proyecto mejoró algo. Si nadie sabe cuánto es hoy, la primera entrega del proyecto es medirlo, no modelarlo.

### 1.7.3 ¿Quién utilizará el resultado y con qué propósito?

Un modelo sin destinatario identificado no cambia ninguna decisión. La pregunta se responde con un **nombre de rol**, una **decisión concreta** y una **frecuencia**: «el jefe de retención decide cada lunes a qué 500 clientes llama el equipo comercial esa semana».

De esa respuesta se desprende todo lo demás: la frecuencia fija cada cuánto debe ejecutarse el modelo, el volumen de la acción (500 llamadas) fija el punto de corte del puntaje, y el rol determina en qué formato se entrega. Cuando la respuesta es «para tenerlo», el proyecto todavía no está listo para empezar.



🔎 **Qué hace este código.** La celda siguiente aplica las tres preguntas a cuatro proyectos y deriva un veredicto de una regla simple: **un solo «No» detiene el proyecto**, sin importar cuán bien puntúe en lo demás. Las condiciones no se compensan entre sí — un KPI impecable no repone los datos que no existen.


```python
import pandas as pd

pd.set_option("display.max_colwidth", None)

viabilidad = pd.DataFrame(
    {
        "Proyecto": [
            "Churn en telco",
            "Segmentación en retail",
            "Forecasting de demanda",
            "Detección de fraude interno",
        ],
        "1.7.1 Datos": ["Sí", "Sí", "Sí", "No"],
        "1.7.2 KPI": ["Sí", "Parcial", "Sí", "Parcial"],
        "1.7.3 Destinatario": ["Sí", "Sí", "Sí", "Sí"],
        "Qué falta": [
            "Nada: 24 meses de historia, tasa de churn definida, jefe de retención decide cada lunes",
            "El KPI no es directo: no hay variable objetivo, el éxito se mide por el ingreso del segmento",
            "Nada: tres años por producto-tienda-semana, MAPE con línea base conocida",
            "Solo 3 casos confirmados en 5 años: no hay evento que aprender",
        ],
    }
)

CRITERIOS = ["1.7.1 Datos", "1.7.2 KPI", "1.7.3 Destinatario"]
PESO = {"Sí": 2, "Parcial": 1, "No": 0}


def veredicto(fila):
    respuestas = [fila[c] for c in CRITERIOS]
    if "No" in respuestas:
        return "NO PROCEDE — falta una condición"
    if "Parcial" in respuestas:
        return "PROCEDE CON RESERVAS"
    return "PROCEDE"


viabilidad["Veredicto"] = viabilidad.apply(veredicto, axis=1)
viabilidad["Puntaje (0-6)"] = viabilidad[CRITERIOS].apply(lambda col: col.map(PESO)).sum(axis=1)

viabilidad[["Proyecto"] + CRITERIOS + ["Puntaje (0-6)", "Veredicto", "Qué falta"]]
```

📖 **Cómo leer esta tabla.** El puntaje ordena, el veredicto decide. «Detección de fraude interno» suma 3 sobre 6 y queda fuera, mientras que «Segmentación en retail» suma 5 y entra con reservas: la diferencia no está en el total sino en **cuál** de las tres condiciones falla. Un «No» en datos es terminal; un «Parcial» en KPI se trabaja.

💡 **Intuición — el veredicto no juzga la técnica.** Ninguno de los cuatro proyectos se descarta por ser difícil de modelar. «Detección de fraude interno» se detiene porque en cinco años hay tres casos confirmados: con ese histórico no existe modelo, por sofisticado que sea, capaz de aprender el patrón. La decisión de no empezar es un resultado del análisis, no un fracaso.

⚠️ **Error de interpretación frecuente — confundir «no procede aún» con «no procede nunca».** El veredicto vale para las condiciones de hoy. Un proyecto detenido por falta de datos puede habilitarse en seis meses si se instrumenta la captura, y esa instrumentación suele ser la recomendación más valiosa del análisis de viabilidad. Lo que no procede es empezar a modelar mientras la condición siga sin cumplirse.

### Drill 2 — Riesgos en la fase «comprensión de los datos» (CRISP-DM)

Para ello dispone de: (a) las transacciones de caja de los últimos dos años, registradas en **dos sistemas POS distintos** porque en 2024 la empresa migró de proveedor; (b) el **padrón de afiliados**, donde el número de documento se digita manualmente en tienda; (c) **encuestas de satisfacción** respondidas por cerca del 8% de los afiliados; y (d) una tabla de **direcciones y teléfonos** que solo se actualiza cuando el cliente presenta un reclamo. **Consigna:** listar **al menos 6 riesgos** que deberían detectarse en la fase «comprensión de los datos» de CRISP-DM antes de avanzar a la preparación. ### Drill 3 — Variable objetivo y KPI

Para cada uno de los dos problemas siguientes, redactar: la **unidad de análisis**, la **variable objetivo** (con su ventana temporal y su tipo: binaria o continua) y el **KPI de negocio** con el que se mediría el éxito de la iniciativa (con su fórmula en palabras).

**Problema A — Churn:** un servicio de streaming por suscripción mensual quiere reducir las cancelaciones de sus suscriptores de pago.

**Problema B — Demanda:** una panificadora central abastece cada madrugada a sus 35 tiendas y debe decidir cuántas unidades de cada producto enviar a cada tienda; hoy sufre a la vez merma (producto no vendido) y quiebres de stock.

**Criterio de calidad:** una variable objetivo bien redactada permite responder, para cada registro histórico, «¿cuál fue el valor observado?» sin ambigüedad; un KPI bien redactado puede calcularse antes y después de la iniciativa para comparar.

**Proyección — ¿qué pasaría si...?** Sobre el Problema A (churn): la empresa de streaming compara dos versiones de un correo de retención, A y B, y observa que el grupo B cancela ligeramente menos que el grupo A. ¿Alcanza esa diferencia para concluir que el correo B funciona mejor, o podría deberse al azar de qué clientes cayeron en cada grupo? ¿Qué habría que verificar antes de decidir? *(No se resuelve aquí — el diseño de experimentos A/B y el contraste de diferencias son el tema de la S02.)*
