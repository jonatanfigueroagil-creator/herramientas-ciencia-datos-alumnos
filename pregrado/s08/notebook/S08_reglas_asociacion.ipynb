## 8.2 a 8.4 — ¿Qué es una regla de asociación, cómo se identifica sin evaluar todas las combinaciones y cómo se mide su valor? Teoría guiada (Sección 3 del cuaderno)

### Canasta de mercado, itemset frecuente y one-hot por ticket — capítulo 8.2 (subsección 3.1)

Una **transacción** (o *ticket*, *canasta*) es el conjunto de ítems que un cliente compra junto en una misma operación. El **análisis de canasta de mercado** estudia qué ítems tienden a aparecer **juntos** para descubrir patrones de co-ocurrencia. Un **itemset** es cualquier subconjunto de ítems; es **frecuente** cuando su **soporte** alcanza un umbral mínimo `minsup`.

Antes de minar, los datos deben llevarse a una **matriz booleana one-hot**: **una fila por transacción**, **una columna por ítem posible**, con `True`/`False` según el ítem esté o no en esa canasta. `mlxtend.preprocessing.TransactionEncoder` hace esta conversión. La matriz resultante es **muy dispersa** (casi todo `False`).

> **⚠️ Error de preparación más frecuente** (`DEFINICIONES.md`, Sección 10): agrupar **por cliente** en vez de **por ticket** (mezcla canastas distintas e infla el soporte), o usar **cantidades** en lugar de presencia/ausencia. **Una fila = una transacción.**

**🔎 Qué hace este código.** Convierte una lista de 10 tickets de ejemplo en la **matriz one-hot booleana por transacción** con `TransactionEncoder`, y muestra el soporte (fracción de tickets) de cada ítem.

```python
# Mini-demo: de una lista de tickets a la matriz one-hot booleana por transacción
tickets_demo = [
    ["pan", "leche"],
    ["pan", "pañales", "cerveza", "huevos"],
    ["leche", "pañales", "cerveza", "refresco"],
    ["pan", "leche", "pañales", "cerveza"],
    ["pan", "leche", "pañales", "refresco"],
    ["pañales", "cerveza"],
    ["pan", "leche"],
    ["cerveza", "refresco"],
    ["pan", "pañales", "cerveza"],
    ["leche", "refresco"],
]
te_demo = TransactionEncoder()
oh_demo = pd.DataFrame(te_demo.fit(tickets_demo).transform(tickets_demo), columns=te_demo.columns_)
print("One-hot por ticket (10 transacciones x", oh_demo.shape[1], "ítems, dtype", oh_demo.dtypes.iloc[0], "):")
print(oh_demo.astype(int).to_string())
print("\nSoporte de cada ítem (fracción de tickets que lo contienen):")
print((oh_demo.mean().sort_values(ascending=False).round(2)).to_string())
```

### Soporte, confianza, lift, leverage y conviction — capítulo 8.4 (subsección 3.2)

#### 🧮 Matemática en el cuerpo

Para una regla $X \rightarrow Y$ (antecedente $X$, consecuente $Y$), con $N$ transacciones:

- **Soporte:** $\ \text{sop}(X\rightarrow Y) = \text{sop}(X \cup Y) = \dfrac{\#\{\text{tickets con } X \text{ e } Y\}}{N}$ — cuán **común** es el patrón.
- **Confianza** (probabilidad condicional $P(Y\mid X)$): $\ \text{conf}(X\rightarrow Y) = \dfrac{\text{sop}(X\cup Y)}{\text{sop}(X)}$ — de las canastas con $X$, qué fracción también lleva $Y$.
- **Lift:** $\ \text{lift}(X\rightarrow Y) = \dfrac{\text{conf}(X\rightarrow Y)}{\text{sop}(Y)} = \dfrac{\text{sop}(X\cup Y)}{\text{sop}(X)\,\text{sop}(Y)}$ — cuántas veces **supera al azar** la co-ocurrencia. Es **simétrico** y es **la métrica de decisión**:
  - $\text{lift} > 1$ → **asociación positiva** (candidata a cross-sell);
  - $\text{lift} = 1$ → **independencia** (regla sin valor);
  - $\text{lift} < 1$ → **asociación negativa / sustitución** (p. ej. marcas rivales).
- **Leverage:** $\ \text{lev}(X\rightarrow Y) = \text{sop}(X\cup Y) - \text{sop}(X)\,\text{sop}(Y)$ — el **exceso absoluto** de co-ocurrencia sobre lo esperado bajo independencia. Rango $[-1,1]$ (documentado por mlxtend; el realmente alcanzable para reglas binarias es [-0,25; 0,25]); **0 = independencia**. A diferencia del lift (una razón), es una **diferencia**: un lift muy elevado sobre soportes ínfimos puede tener leverage minúsculo (poco volumen).
- **Conviction:** $\ \text{conv}(X\rightarrow Y) = \dfrac{1-\text{sop}(Y)}{1-\text{conf}(X\rightarrow Y)}$ — cuán **implicación real** es (penaliza contraejemplos). **1 = independencia**, **> 1 = informativa**, **→ ∞** cuando la confianza → 1. Es **direccional**: $\text{conv}(X\rightarrow Y) \neq \text{conv}(Y\rightarrow X)$.

A continuación se resuelve **de forma manual** la regla $\{\text{pañales}\} \rightarrow \{\text{cerveza}\}$ sobre las 10 canastas de arriba y se contrasta con `mlxtend`.

**❓ Qué se quiere averiguar.** ¿Quien compra pañales se lleva cerveza porque exista una asociación real entre ambos productos, o simplemente porque la cerveza se vende mucho en esa tienda?

- **Qué decide:** si la asociación es real, el par justifica una acción de surtido —proximidad en góndola, sugerencia en el carrito—. Si solo refleja popularidad, la misma acción valdría para cualquier producto y no aportaría nada.
- **Antes de mirar el resultado:** la confianza sola no puede responder, porque nunca se compara con el nivel base del consecuente. Quien decide es el **lift**: si resulta **> 1**, la co-ocurrencia supera al azar y hay asociación positiva, candidata a cross-sell; si resulta **= 1**, las dos compras son independientes y la regla carece de valor por alta que sea su confianza; si resulta **< 1**, los productos se sustituyen — información útil, pero de otro tipo.

**🔎 Qué hace este código (🖐️ Cálculo manual).** Cuenta transacciones sobre `oh_demo` y calcula **manualmente** soporte, confianza, lift, leverage y conviction de $\{\text{pañales}\}\rightarrow\{\text{cerveza}\}$; luego **verifica con `assert`** que coinciden exactamente con `association_rules` de mlxtend.

```python
# 🖐️ HAZLO A MANO: soporte, confianza, lift, leverage y conviction de {pañales} -> {cerveza}
# (transacciones ya cargadas en oh_demo) y verificación contra mlxtend con assert
N = len(oh_demo)
sop_pan = oh_demo["pañales"].mean()                          # soporte del antecedente
sop_cer = oh_demo["cerveza"].mean()                          # soporte del consecuente
sop_amb = (oh_demo["pañales"] & oh_demo["cerveza"]).mean()     # soporte del itemset {pañales, cerveza}
conf = sop_amb / sop_pan
lift = conf / sop_cer
leverage = sop_amb - sop_pan * sop_cer
conv = (1 - sop_cer) / (1 - conf)

print("Cálculo A MANO de la regla  {pañales} -> {cerveza}:")
print(f"  soporte(pañales)          = {int(sop_pan*N)}/{N} = {sop_pan:.2f}")
print(f"  soporte(cerveza)          = {int(sop_cer*N)}/{N} = {sop_cer:.2f}")
print(f"  soporte(pañales u cerveza)= {int(sop_amb*N)}/{N} = {sop_amb:.2f}")
print(f"  confianza  = sop(u)/sop(pañales)        = {sop_amb:.2f}/{sop_pan:.2f} = {conf:.3f}")
print(f"  lift       = conf/sop(cerveza)          = {conf:.3f}/{sop_cer:.2f} = {lift:.3f}   (>1 -> asociación positiva)")
print(f"  leverage   = sop(u) - sop(p)*sop(c)     = {sop_amb:.2f} - {sop_pan*sop_cer:.2f} = {leverage:.3f}")
print(f"  conviction = (1-sop(cerveza))/(1-conf)  = {(1-sop_cer):.2f}/{(1-conf):.3f} = {conv:.3f}")

# Verificación con mlxtend (mismo resultado) + assert
fi_demo = apriori(oh_demo, min_support=0.2, use_colnames=True)
reglas_demo = association_rules(fi_demo, metric="confidence", min_threshold=0.5)
fila_df = reglas_demo[(reglas_demo["antecedents"] == frozenset({"pañales"})) &
                      (reglas_demo["consequents"] == frozenset({"cerveza"}))]
print("\nVerificación con mlxtend (support, confidence, lift, leverage, conviction):")
print(fila_df[["support", "confidence", "lift", "leverage", "conviction"]].round(3).to_string(index=False))
fila = fila_df.iloc[0]
assert (np.isclose(sop_amb, fila["support"]) and np.isclose(conf, fila["confidence"])
        and np.isclose(lift, fila["lift"]) and np.isclose(leverage, fila["leverage"])
        and np.isclose(conv, fila["conviction"])), "el cálculo a mano NO coincide con mlxtend"
print("\nassert OK: soporte, confianza, lift, leverage y conviction a mano == mlxtend.")
```

**📖 Lectura de negocio** (`INTERPRETACION_RESULTADOS.md`, Sección 2). El lift **corrige el sesgo de popularidad de la confianza**: separa asociaciones reales de coincidencias por volumen. Un lift de 1.39 se lee "quien lleva pañales tiene un 39 % más de probabilidad de llevar cerveza que un cliente al azar" — candidata a **cross-sell**. La conviction > 1 confirma que romper la regla (pañales sin cerveza) es más raro de lo esperado por azar. El **lift retoma el lift de A/B testing (S02)**, ahora aplicado a itemsets.

### Reglas triviales: confianza alta puede tener lift < 1 — capítulo 8.4 (subsección 3.3)

Este es el malentendido más importante de la sesión (`INTERPRETACION_RESULTADOS.md`, Sección 3). Una **regla trivial** tiene **confianza alta pero lift ≈ 1 (o < 1)**: la confianza es alta **solo porque el consecuente es muy popular**, no porque exista asociación. Ejemplo clásico: en una tienda donde casi toda canasta lleva **bolsa**, la regla $\{\text{leche}\}\rightarrow\{\text{bolsa}\}$ tiene confianza muy alta, pero el lift la delata.

Se construye un caso donde **bolsa** aparece en 9 de 10 tickets y se calcula la regla.

**❓ Qué se quiere averiguar.** ¿Una regla con confianza del 80 % es, por eso solo, una buena regla?

- **Qué decide:** de la respuesta depende el criterio con que se ordena **cualquier** lista de reglas. Quien seleccione por confianza llenará el informe de reglas dirigidas al producto más vendido de la tienda, que es justamente el que menos información aporta.
- **Antes de mirar el resultado:** la bolsa aparece en 9 de cada 10 tickets, así que su soporte base es **0,90**. Si la confianza de `{leche} → {bolsa}` resultara **por encima de 0,90**, llevar leche elevaría la probabilidad de llevar bolsa. Si resulta **igual a 0,90**, la leche no aporta nada. Y si resulta **por debajo**, quien lleva leche compra bolsa **menos** que el cliente promedio: una confianza de 0,80 esconde entonces un lift menor que 1, y la regla se descarta.

**🔎 Qué hace este código.** Construye 10 tickets donde `bolsa` es casi ubicua (9 de 10) y calcula la regla `{leche} -> {bolsa}`: su confianza parece alta, pero el lift resulta < 1. (Escenario **ilustrativo** con su propio mini-conjunto: aquí sale conf 0,80 / lift 0,889, **producto de ejecutar este código** sobre estos 10 tickets. El caso «0,88 / 0,98» que citan la guía y el control corto es un **ejemplo didáctico de libro** — no una salida de ningún cómputo sobre T10I4D100K ni Online Retail II — usado para redondear la lección con números limpios. Misma lección, dos procedencias distintas: no tratarlas como el mismo cálculo.)

```python
# Mini-demo: regla trivial -> confianza alta (0.80) pero lift < 1
tickets_bolsa = [
    ["leche", "bolsa"], ["leche", "pan", "bolsa"], ["leche", "huevos", "bolsa"],
    ["leche", "pan", "bolsa"], ["pan", "bolsa"], ["huevos", "bolsa"],
    ["pan", "huevos", "bolsa"], ["pan", "bolsa"], ["huevos", "bolsa"],
    ["leche"],   # único ticket SIN bolsa
]
te_b = TransactionEncoder()
oh_b = pd.DataFrame(te_b.fit(tickets_bolsa).transform(tickets_bolsa), columns=te_b.columns_)
sop_bolsa = oh_b["bolsa"].mean()
sop_leche = oh_b["leche"].mean()
conf_lb = (oh_b["leche"] & oh_b["bolsa"]).mean() / sop_leche
lift_lb = conf_lb / sop_bolsa
print(f"soporte(bolsa) = {sop_bolsa:.2f}  (bolsa es casi ubicua)")
print(f"regla  {{leche}} -> {{bolsa}}:  confianza = {conf_lb:.2f}  (¡parece altísima!)")
print(f"lift = confianza / soporte(bolsa) = {conf_lb:.2f} / {sop_bolsa:.2f} = {lift_lb:.3f}  ->  < 1  =>  TRIVIAL")
print("\nLectura: la confianza es alta solo porque la bolsa es ubicua; el lift < 1 dice que")
print("quien lleva leche compra bolsa un POCO MENOS que el cliente promedio. Se descarta.")
```

**📖 Regla de lectura** (`INTERPRETACION_RESULTADOS.md`, Sección 3 y Sección 8). **Nunca** ordenar ni seleccionar reglas por confianza sola: la confianza alta es condición necesaria pero **no suficiente**; se exige además **lift > 1** (con margen). Confundir `lift < 1` con "regla débil" también es un error: no es débil, es **negativa** (sustitución, p. ej. dos marcas rivales) — información útil de otro tipo, no candidata a cross-sell.

### Apriori: anti-monotonía, generación y poda — capítulo 8.3 (subsección 3.4)

**Apriori** (Agrawal & Srikant, 1994) halla todos los itemsets frecuentes apoyándose en la **propiedad de anti-monotonía del soporte** (*downward closure*):

> **Todo subconjunto de un itemset frecuente es también frecuente.** Equivalentemente: **si un itemset es infrecuente, todos sus superconjuntos son infrecuentes.**

Procede **nivel por nivel**: (1) cuenta los 1-itemsets frecuentes; (2) **genera candidatos** de tamaño $k{+}1$ combinando pares de frecuentes de tamaño $k$; (3) **poda** todo candidato con algún subconjunto no frecuente (aplicando la anti-monotonía **sin tocar los datos**); (4) recorre la base para contar el soporte de los supervivientes. La poda es lo que hace **factible** el problema (evita contar los $2^m$ itemsets posibles).

**🔎 Qué hace este código (🖐️ Cálculo manual).** Mina el demo por niveles y verifica la **anti-monotonía** (los subconjuntos de un itemset frecuente son frecuentes); luego ejecuta **manualmente un paso de poda**: `huevos` es infrecuente (soporte 0.1 < 0.3), así que el candidato `{huevos, cerveza}` se descarta **sin contarlo** — y se comprueba con `assert` que en efecto es infrecuente y mlxtend tampoco lo reporta.

```python
# 🖐️ Apriori por niveles + PODA por anti-monotonía HECHA A MANO (con assert)
fi_ap_demo = apriori(oh_demo, min_support=0.3, use_colnames=True)
fi_ap_demo["k"] = fi_ap_demo["itemsets"].apply(len)
print("Itemsets frecuentes (minsup = 0.3), por tamaño:")
for k in sorted(fi_ap_demo["k"].unique()):
    print(f"  {k}-itemsets: {int((fi_ap_demo['k']==k).sum())}")

frecuentes = set(fi_ap_demo["itemsets"])
# (1) anti-monotonía: un 2-itemset frecuente tiene TODOS sus subconjuntos de 1 ítem frecuentes
par = next(s for s in frecuentes if len(s) == 2)
subconj_ok = all(frozenset({i}) in frecuentes for i in par)
print(f"\nAnti-monotonía: el 2-itemset frecuente {como_texto(par)!r} tiene sus 1-subconjuntos frecuentes -> {subconj_ok}")
assert subconj_ok

# (2) PODA a mano: 'huevos' es infrecuente (sop 0.1 < 0.3); todo candidato que lo contenga se DESCARTA sin contar
infrecuente = "huevos"
sop_infrec = oh_demo[infrecuente].mean()
candidato = frozenset({infrecuente, "cerveza"})                          # candidato 2-itemset
se_poda = frozenset({infrecuente}) not in frecuentes                     # su subconjunto {huevos} no es frecuente
sop_candidato_real = (oh_demo[infrecuente] & oh_demo["cerveza"]).mean()  # se comprueba a posteriori
print(f"\nPODA: soporte({infrecuente}) = {sop_infrec:.2f} < 0.3  ->  {infrecuente} es INFRECUENTE.")
print(f"El candidato {como_texto(candidato)!r} se PODA sin contar (tiene un subconjunto infrecuente).")
print(f"Comprobación a posteriori: su soporte real = {sop_candidato_real:.2f} < 0.3 (en efecto, infrecuente).")
assert se_poda and sop_candidato_real < 0.3 and candidato not in frecuentes
print("assert OK: la anti-monotonía autoriza a descartar el candidato sin recorrer los datos; mlxtend concuerda.")
```

**📖 Lectura de negocio.** Apriori es el algoritmo **fundacional** del área y el **paper de la réplica** (sección 4). Su límite —**generar y contar muchos candidatos** con múltiples pasadas sobre los datos— es justo lo que motiva a FP-Growth. Fijar `minsup` es una **decisión de negocio**: un patrón en el 0,1 % de las ventas puede ser irrelevante para un supermercado pero clave para un artículo de alto margen.

### FP-Growth: FP-tree, sin generación de candidatos — capítulo 8.3 (subsección 3.5)

**FP-Growth** (Han, Pei & Yin, 2000) halla **exactamente los mismos** itemsets frecuentes que Apriori, pero **sin generar candidatos**. Comprime la base en un **FP-tree** (árbol de prefijos): las canastas parecidas comparten ramas y el árbol queda compacto. Con solo **dos pasadas** sobre los datos y **FP-trees condicionales** por ítem, extrae todos los patrones. Como produce el **mismo resultado exacto**, la elección Apriori vs. FP-Growth es puramente de **eficiencia**, no de correctitud. En `mlxtend`, `fpgrowth` es un *drop-in replacement* de `apriori`.

**🔎 Qué hace este código.** Ejecuta `fpgrowth` sobre el demo al mismo `minsup = 0.3` y comprueba con `assert` que el conjunto de itemsets es **idéntico** al de Apriori (diferencia = 0).

```python
# FP-Growth devuelve EXACTAMENTE el mismo conjunto de itemsets que Apriori (diferencia = 0)
fi_fp_demo = fpgrowth(oh_demo, min_support=0.3, use_colnames=True)
set_ap_demo = set(fi_ap_demo["itemsets"])
set_fp_demo = set(fi_fp_demo["itemsets"])
print(f"Apriori: {len(set_ap_demo)} itemsets  |  FP-Growth: {len(set_fp_demo)} itemsets")
print(f"¿Conjuntos idénticos? -> {set_ap_demo == set_fp_demo}   (diferencia = {len(set_ap_demo ^ set_fp_demo)})")
assert set_ap_demo == set_fp_demo and len(set_ap_demo ^ set_fp_demo) == 0
print("\nMismos itemsets, mismos soportes: solo cambia el TIEMPO de cómputo (se mide en la sección 4).")
```

**📖 Lectura de negocio.** Al eliminar la generación de candidatos, FP-Growth suele ser **órdenes de magnitud más rápido** que Apriori cuando `minsup` es bajo o hay patrones largos — los escenarios costosos. Si Apriori y FP-Growth devolvieran conjuntos **distintos**, no sería un hallazgo: sería un **error de preparación** (`INTERPRETACION_RESULTADOS.md`, Sección 8). La coincidencia exacta es una **verificación de correctitud**.

### Qué preguntaba el paper, y por qué usó lo que usó — Sección 0 del paper (subsección 4.0)

**💡 Antes de tocar los datos.** Una réplica sin esta pregunta se vuelve mecánica: se ejecutan celdas y sale un número. Lo que sigue explica **qué querían averiguar los autores** y **por qué eligieron cada pieza de su método**, que es de donde proviene el criterio para elegir un método propio mañana. *(Desarrollo completo con las citas del original: la ficha de la sesión de réplica del paper, «Sección 0».)*

**El objetivo no era descubrir qué se compra junto.** Eso ya estaba planteado y definido un año antes por Agrawal, Imieliński y Swami (SIGMOD 1993), a quienes el artículo cita y de quienes toma la formulación literal del problema (p. 487). Lo que Agrawal y Srikant aportan cabe en la primera palabra del título: *Fast*. La pregunta era **si generar todas las reglas era computacionalmente posible** sobre las bases que el código de barras ya acumulaba: «Progress in bar-code technology has made it possible for retail organizations to collect and store massive amounts of sales data» (p. 487). Los datos existían; la manera de explotarlos, no. El objetivo declarado llega enseguida: «The databases involved in these applications are very large. It is imperative, therefore, to have fast algorithms for this task» (p. 487).

**Por qué era difícil — la palabra «todas»** (p. 487). El problema exige **todas** las reglas por encima de los umbrales, no las más interesantes ni una muestra: esa exhaustividad prohíbe cualquier atajo. Con m ítems hay 2^m − 1 itemsets no vacíos, y los autores configuran 1000 ítems en sus experimentos (p. 493), de modo que la fuerza bruta no queda descartada por costosa, sino por inejecutable. El equipo de la época —IBM RS/6000 a 33 MHz, 64 MB de memoria, disco a 2 MB por segundo (p. 492)— hace además que **cada pasada sobre el fichero** cueste tiempo real.

**La decisión de método central — de dónde provienen los candidatos** (pp. 488-489). AIS y SETM los generaban al vuelo mientras leían cada transacción, con el defecto que los autores enuncian sin rodeos: así se generan y cuentan «too many candidate itemsets that turn out to be small». Apriori invierte el origen y los construye **solo desde los itemsets frecuentes de la pasada anterior, sin mirar las transacciones**. La licencia es la **anti-monotonía del soporte** —«any subset of a large itemset must be large» (p. 489)—, cuya forma contrapuesta es la operativa: si un itemset no llega al soporte mínimo, ningún superconjunto suyo llegará. Por eso un candidato se descarta **sin contarlo y sin abrir el fichero**. Es exactamente lo que el Paso 5 de esta sección cuantifica nivel por nivel.

**Por qué un dataset sintético** (pp. 492-493). «We generated synthetic transactions to evaluate the performance of the algorithms over a large range of data characteristics»: el objetivo no era describir una tienda, sino **medir algoritmos**, y para eso hay que variar de forma controlada lo que determina el coste. `T10I4D100K` es literalmente una fila de la malla del artículo —|T| = 10, |I| = 4, |D| = 100 000—, y sus 100 000 transacciones son el techo que imponía SETM, que «could not be run for larger values». Que los ítems sean números anónimos no es una carencia: es la condición para que la medición sea sobre el algoritmo y no sobre la tienda. Los nombres legibles llegan en la Sección 5, con Online Retail II.

**⚠️ De ahí la diferencia entre lo que mide el paper y lo que mide este cuaderno.** El artículo publica segundos de una máquina de 33 MHz con una implementación en C que nunca se distribuyó: **esos segundos no se reproducen aquí, y no por descuido**. Lo que este cuaderno replica es la condición que el argumento entero da por supuesta —que dos algoritmos exactos devuelvan el **mismo** conjunto de itemsets, con diferencia 0— y la **estructura** del experimento: la misma malla de soportes decrecientes sobre la que el artículo observa cómo se ensancha la brecha (Tabla 4, p. 495). Un algoritmo mucho más rápido que devuelve otra respuesta no es una mejora: es un error.

**Paso 3 — Minar itemsets y verificar la identidad (targets #1 y #2).** Se ejecutan `apriori` y `fpgrowth` al mismo `minsup = 0.01` (soporte absoluto 1000/100000), se cronometra cada uno, y se comprueba que devuelven el **mismo conjunto exacto** (diferencia 0).

**🔎 Qué hace este código.** Ejecuta ambos algoritmos a `minsup = 0.01`, mide sus tiempos, cuenta itemsets por tamaño y calcula la diferencia de conjuntos (target #2).

```python
# Pasos 3: Apriori y FP-Growth a minsup=0.01 (cronometrados) + identidad exacta
MINSUP = 0.01  # 1% = soporte absoluto 1000 sobre 100 000 transacciones

t0 = time.perf_counter()
fi_ap = apriori(df, min_support=MINSUP, use_colnames=True)
t_ap_001 = time.perf_counter() - t0

t0 = time.perf_counter()
fi_fp = fpgrowth(df, min_support=MINSUP, use_colnames=True)
t_fp_001 = time.perf_counter() - t0

n_itemsets_apriori = len(fi_ap)
n_itemsets_fpgrowth = len(fi_fp)
diff_apriori_fpgrowth = abs(n_itemsets_apriori - n_itemsets_fpgrowth)

tam = fi_ap["itemsets"].apply(len)
set_ap = set(frozenset(s) for s in fi_ap["itemsets"])
set_fp = set(frozenset(s) for s in fi_fp["itemsets"])
conjuntos_iguales = (set_ap == set_fp)

print(f"[#1] Itemsets frecuentes (OPERATIVO, venv) = {n_itemsets_apriori}   "
      f"| benchmark comunitario ~= 385")
print(f"      reparto por tamaño -> 1-itemsets: {int((tam==1).sum())}, "
      f"2-itemsets: {int((tam==2).sum())}, 3-itemsets: {int((tam==3).sum())}")
print(f"[#2] Apriori = {n_itemsets_apriori}  |  FP-Growth = {n_itemsets_fpgrowth}  "
      f"|  conjuntos idénticos = {conjuntos_iguales}  |  diferencia = {diff_apriori_fpgrowth}")
print(f"      tiempo Apriori = {t_ap_001:.2f} s   vs   FP-Growth = {t_fp_001:.2f} s")
```

**📖 Interpretación (target #2, el núcleo de la sesión).** Apriori y FP-Growth devuelven el **mismo conjunto de itemsets** (diferencia = 0): son algoritmos **exactos**, así que la elección entre ellos es solo de **eficiencia**. **❓ Qué se quiere averiguar.** Cuando alguien afirma «esta es la regla más fuerte del dataset», ¿qué afirma exactamente?

- **Qué decide:** la regla que encabeza la lista es la que obtiene el espacio en góndola, el cupón o la sugerencia en pantalla. Si el orden depende del criterio, el criterio se declara **antes**, no se elige después de ver qué regla queda primera.
- **Antes de mirar el resultado:** si las tres métricas —lift, confianza y conviction— situaran en primer lugar la **misma** regla, el superlativo sería inofensivo. Si sitúan en primer lugar reglas **distintas**, entonces «la más fuerte» sin métrica declarada es una afirmación vacía, y la regla que suele citarse del benchmark (`{39,704}→{825}`, confianza ≈ 0,935) no tiene por qué ser la primera por lift.

**🔎 Qué hace este código (Paso 3, cont. 2 — Target #3b).** Sobre las mismas 7 reglas ya minadas, identifica la de **mayor lift**, la de **mayor confianza** y la de **mayor conviction**, y las compara con la regla que suele citarse como «la más fuerte» del benchmark (`{39,704}→{825}`). Certifica con `assert` que **«la más fuerte» depende de la métrica elegida**.

```python
# Target #3b: "la regla más fuerte" depende de la métrica -> se certifican las tres
def _fila_regla(antecedente, consecuente):
    m = reglas[(reglas["antecedents"] == frozenset(antecedente)) &
               (reglas["consequents"] == frozenset(consecuente))]
    assert len(m) == 1, f"regla {antecedente}->{consecuente} no está entre las {n_reglas_conf50} reglas"
    f = m.iloc[0]
    return {"antecedente": como_texto(f["antecedents"]), "consecuente": como_texto(f["consequents"]),
            "support": float(f["support"]), "confidence": float(f["confidence"]),
            "lift": float(f["lift"]), "conviction": float(f["conviction"])}

fila_publicada = _fila_regla({"39", "704"}, {"825"})     # la que suele citarse como "la más fuerte"
fila_max_lift  = _fila_regla({"39", "825"}, {"704"})     # máximo lift real del benchmark
fila_max_conf  = _fila_regla({"704", "825"}, {"39"})     # máxima confianza real del benchmark

trio_benchmark_df = pd.DataFrame([
    {"etiqueta": "publicada (citada como 'la más fuerte')", **fila_publicada},
    {"etiqueta": "máximo lift real",      **fila_max_lift},
    {"etiqueta": "máxima confianza real", **fila_max_conf},
])
print("Tres candidatos a 'la regla más fuerte' del benchmark -- cada uno gana por una métrica distinta:")
print(trio_benchmark_df.round(4).to_string(index=False))

assert np.isclose(fila_max_lift["lift"], reglas["lift"].max())
assert np.isclose(fila_max_conf["confidence"], reglas["confidence"].max())
assert fila_max_lift["lift"] > fila_publicada["lift"]
assert fila_max_conf["confidence"] > fila_publicada["confidence"]
print("\nassert OK: la regla publicada NO es la de mayor lift ni la de mayor confianza del "
      "benchmark -- 'la más fuerte' exige declarar la métrica.")
```

**📖 Cómo se lee esta salida.** Las tres filas comparten los mismos datos; solo cambia el criterio de orden. `{39,704}→{825}` (confianza ≈0,935) es real y su aritmética es correcta, pero es la **3.ª por lift** (30,3 vs. 48,6 de `{39,825}→{704}`) y la **2.ª por confianza** (0,935 vs. 0,939 de `{704,825}→{39}`): no es un superlativo válido sin declarar la métrica. Regla de lectura de la sesión (checklist #1 de `INTERPRETACION_RESULTADOS.md`): **ordenar por lift** (con conviction como desempate), nunca por confianza sola. Las tres filas quedan certificadas en `resultados/S08_resultados.xlsx` (hoja `top_reglas_benchmark`).

**Paso 4 — Tendencia de tiempo Apriori vs. FP-Growth (target T, drill 3).** Se cronometran ambos a `minsup` decreciente. El número de itemsets crece al bajar el umbral; **Apriori se degrada más** (genera y cuenta muchos candidatos), mientras **FP-Growth escala mejor** (dos pasadas, sin candidatos). Se usan tres niveles seguros; el nivel `0.01` reutiliza los tiempos ya medidos arriba.

> **⚠️ Aviso de procedencia (reloj de PARED).** Los segundos de esta celda son tiempo de **reloj de pared** (`time.perf_counter`): dependen de la máquina, la carga del sistema y la versión de las librerías en el momento de ejecutar. **El valor absoluto NUNCA será idéntico entre ejecuciones o máquinas** y por eso NO es lo que se certifica. Lo que SÍ es robusto y se certifica (el material de referencia de la sesión) es la **relación**: FP-Growth es más rápido que Apriori a **todo** `minsup` probado, y la aceleración **crece** al bajar `minsup`. Los segundos de esta celda son los de **esta ejecución, en esta máquina, en la fecha que se imprime abajo**, y se vuelcan tal cual al Excel de contrato (hoja `tiempos`, columna `fecha_corrida`) — sin restaurar ni fijar ningún valor de una ejecución anterior.

**🔎 Qué hace este código.** Cronometra Apriori y FP-Growth a `minsup ∈ {0.02, 0.015, 0.01}`, tabula nº de itemsets y tiempos, y calcula la aceleración de FP-Growth. Imprime la fecha de esta ejecución (se guarda en el Excel junto a los tiempos).

```python
# Paso 4: comparación de tiempos Apriori vs FP-Growth a minsup decreciente
filas_tiempo = []
for ms in [0.02, 0.015]:                       # niveles adicionales (0.01 ya medido arriba)
    t0 = time.perf_counter(); a = apriori(df, min_support=ms, use_colnames=True); ta = time.perf_counter() - t0
    t0 = time.perf_counter(); f = fpgrowth(df, min_support=ms, use_colnames=True); tf = time.perf_counter() - t0
    filas_tiempo.append({"minsup": ms, "n_itemsets": len(a),
                         "t_apriori_s": round(ta, 3), "t_fpgrowth_s": round(tf, 3),
                         "iguales": len(a) == len(f)})
filas_tiempo.append({"minsup": MINSUP, "n_itemsets": n_itemsets_apriori,
                     "t_apriori_s": round(t_ap_001, 3), "t_fpgrowth_s": round(t_fp_001, 3),
                     "iguales": conjuntos_iguales})

tiempos = pd.DataFrame(filas_tiempo).sort_values("minsup", ascending=False).reset_index(drop=True)
tiempos["aceleracion_fp"] = (tiempos["t_apriori_s"] / tiempos["t_fpgrowth_s"]).round(1)

FECHA_CORRIDA_TIEMPOS = date.today().strftime("%d/%m/%Y")
print(f"Tiempo (s) por algoritmo al bajar minsup -- CORRIDA de HOY ({FECHA_CORRIDA_TIEMPOS}), "
      f"reloj de pared, machine-dependent (los itemsets crecen; FP-Growth se separa por debajo):")
print(tiempos.to_string(index=False))
print("\nLa identidad Apriori == FP-Growth se mantiene a TODO umbral (columna 'iguales').")
```

**📖 Interpretación (target T).** Al bajar `minsup`, el nº de itemsets crece y el tiempo de **Apriori aumenta de forma pronunciada**, mientras **FP-Growth se mantiene bajo**: la aceleración de FP-Growth **aumenta** conforme el problema se hace costoso. Lo **robusto y evaluable** es el **orden** (FP-Growth ≤ Apriori a `minsup` bajo), no un factor exacto — que depende de la máquina, la versión de `mlxtend` y el dataset. En negocio: para minería exploratoria con umbrales bajos (recomendación de nicho), **FP-Growth** es la opción por defecto.

Esta relación (nunca el segundo exacto) es lo que certifica el material de referencia de la sesión: FP-Growth más rápido que Apriori a los tres `minsup`, y aceleración no decreciente al bajar el umbral — verificado tanto sobre la ejecución guardada en el Excel como recomputando en vivo en la máquina que ejecuta el validador.

**Paso 5 — Cuantificar la poda por anti-monotonía: candidatos generados vs. itemsets frecuentes.** El argumento central de Agrawal & Srikant (1994) es que la propiedad de anti-monotonía (Sección 3.4) hace **factible** el problema: en vez de contar los `2^m` subconjuntos posibles de los `m` ítems, Apriori solo **genera candidatos** combinando frecuentes del nivel anterior (**join**) y **poda** los que tengan algún subconjunto infrecuente, ANTES de gastar una pasada por los datos contándolos. Hasta aquí el argumento se enunció en prosa (Sección 3.6); esta celda lo **cuantifica** con números reales sobre el benchmark.

**🔎 Qué hace este código.** Implementa el paso **join** (unir frecuentes del nivel `k-1` que comparten los primeros `k-2` ítems, en orden canónico) y el paso **prune** (descartar el candidato si algún subconjunto de tamaño `k-1` no es frecuente) tal como los describe el paper, y cuenta candidatos / tras-poda / frecuentes en los niveles 1, 2 y 3 sobre `T10I4D100K` a `minsup = 0.01`. El conteo de soporte de los ~70 000 candidatos de nivel 2 se vectoriza con una multiplicación de matrices booleanas (mismo resultado que recorrer la base candidato por candidato, solo más rápido); los del nivel 3 (unos pocos tras la poda) se cuentan de forma directa. Verifica con `assert` que el nº de frecuentes por nivel coincide con el desglose ya conocido de `fi_ap`.

```python
# Paso 5: candidatos generados vs. podados vs. frecuentes por nivel (join + poda, Apriori real)
import itertools, math

L1 = sorted(fi_ap.loc[tam == 1, "itemsets"].apply(lambda s: next(iter(s))).tolist())
n_L1 = len(L1)
n_items_totales = df.shape[1]

# --- Nivel 2: JOIN = combinaciones de 2 frecuentes de nivel 1 (prefijo compartido de longitud 0) ---
n_C2 = math.comb(n_L1, 2)
# PODA nivel 2: cada candidato tiene 2 subconjuntos de tamaño 1, y ambos son frecuentes por
# construcción (se formaron uniendo dos miembros de L1) -> la poda no descarta nada en este nivel
n_C2_podado = n_C2

M = df[L1].to_numpy(dtype=np.float32)                  # matriz ya cargada (Paso 2), recortada a L1
co = M.T @ M                                             # co-ocurrencia de cada par: recorre la base UNA vez, vectorizado
iu, ju = np.triu_indices(n_L1, k=1)
N = len(df)
soporte_pares = co[iu, ju] / N
mask_frecuentes2 = soporte_pares >= MINSUP
L2 = [frozenset({L1[i], L1[j]}) for i, j, ok in zip(iu, ju, mask_frecuentes2) if ok]
n_L2 = len(L2)

# --- Nivel 3: JOIN de L2 (pares que comparten el primer ítem, orden canónico) + PODA por anti-monotonía ---
L2_tuplas = sorted(tuple(sorted(p)) for p in L2)
por_prefijo = {}
for a, b in L2_tuplas:
    por_prefijo.setdefault(a, []).append(b)
C3 = set()
for a, sufijos in por_prefijo.items():
    for x, y in itertools.combinations(sorted(sufijos), 2):
        C3.add(frozenset({a, x, y}))
n_C3 = len(C3)

L2_set = set(L2)
C3_podado = {c for c in C3 if all(frozenset(s) in L2_set for s in itertools.combinations(c, 2))}
n_C3_podado = len(C3_podado)
n_C3_descartado = n_C3 - n_C3_podado

L3 = [c for c in C3_podado if df[list(c)].all(axis=1).mean() >= MINSUP]
n_L3 = len(L3)

candidatos_apriori_df = pd.DataFrame([
    {"nivel": 1, "candidatos": n_items_totales, "tras_poda": n_L1, "frecuentes": n_L1,
     "descartados_por_poda": 0},
    {"nivel": 2, "candidatos": n_C2, "tras_poda": n_C2_podado, "frecuentes": n_L2,
     "descartados_por_poda": n_C2 - n_C2_podado},
    {"nivel": 3, "candidatos": n_C3, "tras_poda": n_C3_podado, "frecuentes": n_L3,
     "descartados_por_poda": n_C3_descartado},
])
print("Candidatos generados vs. podados vs. frecuentes, por nivel (T10I4D100K, minsup=0.01):")
print(candidatos_apriori_df.to_string(index=False))

universo_pares = math.comb(n_items_totales, 2)
print(f"\nNivel 2: el join solo combina frecuentes de L1 (n={n_L1}), generando {n_C2} candidatos de los "
      f"{universo_pares} pares posibles entre los {n_items_totales} ítems "
      f"({n_C2/universo_pares:.1%} del universo) -- un recorte del {1 - n_C2/universo_pares:.1%} "
      f"ANTES de contar un solo soporte, solo por restringir el join a L1.")
if n_C3:
    print(f"Nivel 3: de los {n_C3} candidatos que arma el join de L2, la poda descarta {n_C3_descartado} "
          f"({n_C3_descartado}/{n_C3}) por tener un subconjunto de tamaño 2 infrecuente -- exactamente "
          f"el paso 'prune' del paper, actuando ANTES de gastar una pasada de conteo sobre ellos.")

assert n_L1 == int((tam == 1).sum())
assert n_L2 == int((tam == 2).sum())
assert n_L3 == int((tam == 3).sum())
print(f"\nassert OK: frecuentes por nivel (join+poda a mano) == desglose de mlxtend "
      f"({n_L1} / {n_L2} / {n_L3}).")
```

**📖 Cómo se lee esta salida.** El argumento de Agrawal & Srikant (1994) deja de ser solo prosa: en el nivel 2, el *join* combina únicamente los `375` ítems ya frecuentes, así que genera **70 125** candidatos en vez de los **378 015** pares posibles entre los 870 ítems (un recorte del **81,4 %** antes de contar un solo soporte). En el nivel 3, el *join* de los 9 pares frecuentes arma **2** candidatos de tamaño 3, y la **poda** por anti-monotonía descarta **1** de ellos (`{368, 682, 829}`, porque su subconjunto `{682, 829}` no es frecuente) sin gastar ninguna pasada de conteo sobre él; solo el candidato superviviente (`{39, 704, 825}`) se cuenta y resulta frecuente. Esa poda —no contar candidatos condenados de antemano— es lo que hace **escalable** a Apriori: sin ella, el nivel 3 tendría que evaluar los `math.comb(375, 3)` ≈ **8,7 millones** de tríos posibles entre los ítems de `L1`. Estas cifras quedan certificadas en `resultados/S08_resultados.xlsx` (hoja `candidatos_apriori`).

**Paso d — Estimar el impacto de negocio de una regla** (`INTERPRETACION_RESULTADOS.md`, Sección 5). Para una regla $X\rightarrow Y$:

$$\text{transacciones\_incrementales}(Y) \approx N \cdot \text{sop}(X) \cdot (\text{conf}(X\rightarrow Y) - \text{sop}(Y)) \cdot \text{tasa\_adopción}$$

donde $(\text{conf} - \text{sop}(Y))$ es el **exceso** sobre el nivel base (ligado al lift) y la tasa de adopción es un **supuesto explícito**. Se reporta como **rango**, nunca como cifra exacta.

**🔎 Qué hace este código.** Elige la mejor regla 1→1 por lift y estima las ventas incrementales y el impacto en libras (GBP) como **rango** (adopción 5-15 %), sobre la muestra y sobre la base completa. Guarda ambos escenarios (muestra y base completa) en una tabla para el Excel de contrato (hoja `impacto_negocio`). Salvaguarda muestra→población: el soporte(X) proviene de la muestra de 800 facturas y se aplica al N completo (≈36 969) como supuesto de representatividad, a validar con la base completa antes de escalar. **Margen SUPUESTO ILUSTRATIVO** (`MARGEN_Y = £2,5/u`): Online Retail II reporta precio de venta, no costo, así que el margen NO se deriva de un dato real; el ancla de plausibilidad es que el precio unitario MEDIANO real de la taza Regency en el dataset completo es **£2,95** (verificado en vivo 24/08/2026), por lo que el margen declarado no puede exceder ese precio.

```python
# Paso d: impacto de cross-selling de la regla top (rango con supuestos declarados)
# Se elige la mejor regla 1->1 (antecedente y consecuente de un solo ítem) por lift
uno_a_uno = top_retail[(top_retail["antecedents"].apply(len) == 1) &
                       (top_retail["consequents"].apply(len) == 1)]
r = (uno_a_uno if len(uno_a_uno) else top_retail).sort_values("lift", ascending=False).iloc[0]

N_facturas = len(canastas)          # base de la muestra
N_base_completa = 36969             # facturas del dataset completo (diccionario_datos.md)
sop_X = r["antecedent support"]
sop_Y = r["consequent support"]
conf_XY = r["confidence"]
exceso = conf_XY - sop_Y            # porción "no trivial" ligada al lift
MARGEN_Y = 2.5                      # SUPUESTO ILUSTRATIVO (GBP/u), NO de costo real -- ver nota en la celda de arriba (precio mediano real de la taza = £2,95)
ADOPCION = (0.05, 0.15)            # SUPUESTO: 5%-15% de adopción de la sugerencia

print(f"Regla evaluada:  {{{r['antecedente']}}} -> {{{r['consecuente']}}}")
print(f"  soporte(X) = {sop_X:.3f} | confianza = {conf_XY:.2f} | soporte(Y) = {sop_Y:.3f} | lift = {r['lift']:.1f}")
print(f"  exceso sobre base (conf - sop(Y)) = {exceso:.3f}\n")

filas_impacto = []
for etiqueta, N in [("muestra_800_facturas", N_facturas), ("base_completa_36969_facturas", N_base_completa)]:
    inc_lo = N * sop_X * exceso * ADOPCION[0]
    inc_hi = N * sop_X * exceso * ADOPCION[1]
    print(f"  {etiqueta}: entre {inc_lo:.0f} y {inc_hi:.0f} ventas incrementales de '{r['consecuente']}'/período")
    print(f"     -> impacto ~ £ {inc_lo*MARGEN_Y:,.0f} a £ {inc_hi*MARGEN_Y:,.0f} (margen supuesto £ {MARGEN_Y:.1f}/u, adopción 5-15%)")
    filas_impacto.append({
        "escenario": etiqueta, "N": int(N), "antecedente": r["antecedente"], "consecuente": r["consecuente"],
        "soporte_X": round(float(sop_X), 4), "confianza": round(float(conf_XY), 4),
        "soporte_Y": round(float(sop_Y), 4), "lift": round(float(r["lift"]), 3),
        "exceso_conf_menos_sopY": round(float(exceso), 4),
        "adopcion_baja": ADOPCION[0], "adopcion_alta": ADOPCION[1], "margen_gbp_unidad": MARGEN_Y,
        "ventas_incrementales_baja": round(inc_lo, 1), "ventas_incrementales_alta": round(inc_hi, 1),
        "impacto_gbp_bajo": round(inc_lo * MARGEN_Y, 1), "impacto_gbp_alto": round(inc_hi * MARGEN_Y, 1),
    })
impacto_df = pd.DataFrame(filas_impacto)

print("\nAdvertencia causal: el lift mide ASOCIACIÓN, no causalidad. La cifra es una HIPÓTESIS")
print("a validar con un experimento A/B (S02); la causalidad formal es S12.")
```

**🔎 Qué hace este código (🧱 a).** Con un diccionario `soporte[itemset]` construido desde `fi_ap`, genera **todas** las reglas de los itemsets frecuentes de tamaño ≥ 2 (para cada partición antecedente/consecuente: `conf = sop(Z)/sop(A)`, `lift = conf/sop(B)`), filtra por confianza ≥ 0.5 y **verifica con `assert`** que reproduce las 7 reglas del contrato y el **mismo conjunto** que `association_rules`.

```python
# 🧱 (a) Generar reglas A MANO desde los itemsets frecuentes, sin association_rules
from itertools import combinations
soporte = {frozenset(s): float(v) for s, v in zip(fi_ap["itemsets"], fi_ap["support"])}

reglas_manual = []
for Z, sop_Z in soporte.items():
    if len(Z) < 2:
        continue
    for k in range(1, len(Z)):
        for A in combinations(sorted(Z), k):          # antecedente
            A = frozenset(A)
            B = Z - A                                  # consecuente
            confianza = sop_Z / soporte[A]             # conf(A->B) = sop(A u B)/sop(A)
            lift_m = confianza / soporte[B]            # lift = conf/sop(B)
            if confianza >= 0.5:
                reglas_manual.append((A, B, sop_Z, confianza, lift_m))

n_reglas_manual = len(reglas_manual)
print(f"Reglas (confianza >= 0.5) generadas A MANO desde los itemsets frecuentes: {n_reglas_manual}")
mejores = sorted(reglas_manual, key=lambda t: t[4], reverse=True)[:3]
for A, B, sZ, cf, lf in mejores:
    print(f"  {como_texto(A)} -> {como_texto(B)}   sop={sZ:.3f}  conf={cf:.3f}  lift={lf:.2f}")

# cruce con el contrato del Excel y con el conjunto exacto de mlxtend
n_reglas_xl = int(pd.read_excel(XLSX, sheet_name="asociacion_apriori")
                  .set_index("metrica").loc["n_reglas_conf50", "valor"])
pares_manual = {(A, B) for A, B, *_ in reglas_manual}
pares_mlx = {(frozenset(a), frozenset(c)) for a, c in zip(reglas["antecedents"], reglas["consequents"])}
assert n_reglas_manual == n_reglas_xl == n_reglas_conf50 == 7
assert pares_manual == pares_mlx
print("\nassert OK: la generación de reglas a mano reproduce el contrato (7 reglas) "
      "y el conjunto exacto (antecedente, consecuente) de mlxtend.")
```

**🔎 Qué hace este código (🧱 b).** Implementa un **Apriori artesanal** sobre el demo: genera candidatos uniendo frecuentes de tamaño `k-1`, **poda** los que tengan algún subconjunto no frecuente (anti-monotonía) y solo entonces cuenta el soporte. Verifica con `assert` que su conjunto de itemsets es **idéntico** al de `apriori` de mlxtend, y reafirma la identidad Apriori == FP-Growth del benchmark (diferencia 0).

```python
# 🧱 (b) Apriori DESDE CERO (generar + podar + contar) y verificación contra mlxtend
def apriori_desde_cero(oh, minsup):
    "Devuelve el conjunto de itemsets frecuentes (frozensets) por generación + poda anti-monotónica."
    items = list(oh.columns)
    def soporte_de(itemset):
        return oh[list(itemset)].all(axis=1).mean()
    L = {frozenset({i}) for i in items if soporte_de({i}) >= minsup}   # L1
    frecuentes = set(L)
    k = 2
    while L:
        # GENERAR candidatos de tamaño k uniendo pares de L_{k-1}
        Lk = list(L); candidatos = set()
        for a in range(len(Lk)):
            for b in range(a + 1, len(Lk)):
                u = Lk[a] | Lk[b]
                if len(u) == k:
                    candidatos.add(u)
        # PODAR: descartar candidatos con algún (k-1)-subconjunto no frecuente (anti-monotonía)
        podados = {c for c in candidatos if all((c - frozenset({x})) in L for x in c)}
        # CONTAR soporte solo de los supervivientes
        L = {c for c in podados if soporte_de(c) >= minsup}
        frecuentes |= L
        k += 1
    return frecuentes

frec_cero = apriori_desde_cero(oh_demo, 0.3)
frec_mlx = set(frozenset(s) for s in apriori(oh_demo, min_support=0.3, use_colnames=True)["itemsets"])
print(f"Apriori desde cero: {len(frec_cero)} itemsets  |  mlxtend: {len(frec_mlx)} itemsets  "
      f"|  idénticos = {frec_cero == frec_mlx}")
assert frec_cero == frec_mlx
# y la identidad Apriori == FP-Growth del benchmark (reusando lo ya minado)
assert (set_ap ^ set_fp) == set() and diff_apriori_fpgrowth == 0
print("assert OK: el Apriori artesanal (generar + podar + contar) reproduce EXACTO el de mlxtend; "
      "e identidad Apriori == FP-Growth con diferencia 0.")
```

**📖 Cómo se lee esta salida.** El Apriori artesanal reproduce los **mismos** itemsets que la librería porque implementa la **misma** lógica (generar → podar por anti-monotonía → contar). Y como FP-Growth es exacto, sobre el benchmark ambos coinciden (diferencia 0): la elección entre algoritmos es de **eficiencia**, nunca de resultado.

## 8.6 — ¿Cuándo se puede confiar en una regla? Supuestos: decisiones y condiciones de validez (Sección 8 del cuaderno)

Las reglas de asociación son una técnica **no supervisada** (como el clustering de S07): **no** tienen los supuestos distribucionales del OLS/GLM (normalidad, homocedasticidad; esos empiezan en S09). Lo que sí tienen son **decisiones y condiciones de validez** que gobiernan el resultado. La **fuente canónica** es la guía de supuestos de la sesión; aquí se ejecutan cuatro **diagnósticos** que la ilustran. **Ninguno escribe en el Excel de contrato.**

| Diagnóstico | Qué revela | Fuente |
|---|---|---|
| 8.1 Barrido de soporte mínimo | La explosión combinatoria: al bajar `minsup`, los itemsets crecen (155 → 237 → 385) | `SUPUESTOS_S08.md`, Parte 2.1 |
| 8.2 Sesgo de la confianza | Confianza alta con lift ≈ 1 = regla espuria (consecuente popular) | `SUPUESTOS_S08.md`, Parte 3.1 |
| 8.3 Filtrado por lift | El lift descarta reglas triviales/sustitución hacia ítems ubicuos | `SUPUESTOS_S08.md`, Parte 3.2 y 4.2 |
| 8.4 Apriori vs. FP-Growth | Mismos itemsets a todo umbral; distinta velocidad (la aceleración crece al bajar `minsup`) | `SUPUESTOS_S08.md`, Parte 5.2 |

**❓ Qué se quiere averiguar.** El `minsup` se fija **antes** de minar y decide qué patrones llegan siquiera a existir en la salida. ¿Cuánto cambia el resultado por mover ese único número?

- **Qué decide:** un umbral alto deja una lista corta y cómoda, apta para decisiones de logística, pero borra los nichos rentables — el patrón que aparece en una fracción pequeña de las canastas y sostiene una recomendación personalizada. Un umbral bajo los recupera al precio de miles de itemsets y de un cómputo que aumenta de forma pronunciada.
- **Antes de mirar el resultado:** si al bajar el umbral de **0,02 a 0,01** el conteo de itemsets creciera de forma suave, el `minsup` sería un parámetro menor y podría fijarse de forma arbitraria. Si crece **más que en proporción** —el umbral se reduce a la mitad y los itemsets se multiplican por bastante más que eso—, entonces es el parámetro de mayor costo de la sesión: se calibra por volumen y por tiempo, con el criterio de negocio delante y no el del algoritmo.

**🔎 Qué hace este código (Diagnóstico 8.1 — barrido de soporte mínimo).** Lee del contrato `tiempos` cómo crece el nº de itemsets al bajar `minsup` (155 → 237 → 385) y cuantifica el multiplicador. Solo **lee** el contrato ya calculado; no escribe.

```python
# Diagnóstico 8.1: barrido de soporte mínimo (explosión combinatoria) — NO escribe en el Excel
barrido = tiempos[["minsup", "n_itemsets"]].sort_values("minsup", ascending=False).reset_index(drop=True)
print("Barrido de soporte mínimo (nº de itemsets frecuentes sobre T10I4D100K):")
print(barrido.to_string(index=False))
ms_alto, ms_bajo = barrido["minsup"].iloc[0], barrido["minsup"].iloc[-1]
n_alto, n_bajo = int(barrido["n_itemsets"].iloc[0]), int(barrido["n_itemsets"].iloc[-1])
print(f"\nAl bajar minsup de {ms_alto} a {ms_bajo}, los itemsets pasan de {n_alto} a {n_bajo} "
      f"(x{n_bajo/n_alto:.1f}).")
print("La literatura FIMI reporta que a minsup=0.001 el conteo llega a decenas de miles (benchmark, etiquetado).")
```

**📖 Cómo se lee esta salida** (`SUPUESTOS_S08.md`, Parte 2.1). El `minsup` es el **único parámetro numérico** antes de minar y gobierna la **explosión combinatoria**: muy bajo → miles de itemsets y cómputo lento; muy alto → se pierden nichos valiosos. Se calibra por **volumen de itemsets y tiempo**, no de forma arbitraria, y es una **decisión de negocio** (umbral alto para logística; bajo para recomendación personalizada).

**🔎 Qué hace este código (Diagnóstico 8.2 — sesgo de la confianza).** Construye un caso exacto donde el consecuente `Y` es ubicuo (soporte 0.9) y `X` es independiente de `Y`: la confianza `X → Y` parece alta (0.9) pero el **lift = 1** (y leverage = 0) — la regla es **espuria**. Se verifica con `assert`.

```python
# Diagnóstico 8.2: la trampa de la confianza (confianza alta, lift ~ 1 => espuria) — NO escribe en el Excel
# 100 tickets construidos: Y (consecuente) ubicuo en 90; X en 50, INDEPENDIENTE de Y
Y = np.array([True] * 90 + [False] * 10)
X = np.array([True] * 45 + [False] * 45 + [True] * 5 + [False] * 5)   # 45 con Y, 5 sin Y -> X y Y independientes
oh_trap = pd.DataFrame({"X": X, "Y": Y})
sX = oh_trap["X"].mean(); sY = oh_trap["Y"].mean(); sXY = (oh_trap["X"] & oh_trap["Y"]).mean()
conf_trap = sXY / sX
lift_trap = conf_trap / sY
lev_trap = sXY - sX * sY
print(f"soporte(Y) = {sY:.2f}  (Y es casi ubicuo)")
print(f"confianza(X -> Y) = {conf_trap:.2f}  (¡parece alta!)   pero:")
print(f"lift = confianza/soporte(Y) = {conf_trap:.2f}/{sY:.2f} = {lift_trap:.2f}   y   leverage = {lev_trap:.2f}")
assert abs(lift_trap - 1.0) < 1e-9 and abs(lev_trap) < 1e-9 and conf_trap >= 0.85
print("\nLa confianza alta es un ESPEJISMO de popularidad: lift = 1 (leverage = 0) => X e Y independientes.")
print("Regla espuria: nunca ordenar ni seleccionar por confianza sola.")
```

**📖 Cómo se lee esta salida** (`SUPUESTOS_S08.md`, Parte 3.1). La confianza **no** compara con el nivel base del consecuente, así que **premia a los ítems ubicuos**. La señal de alarma es `confianza ≈ soporte(Y)` → `lift ≈ 1`. Por eso el **lift** (o el leverage) es la métrica de decisión, no la confianza.

**🔎 Qué hace este código (Diagnóstico 8.3 — filtrado por lift).** Sobre un caso con una asociación genuina (`café ↔ leche`) y un ítem ubicuo (`bolsa`, en todos los tickets), genera reglas y aplica el filtro `lift > 1`: sobreviven las reales y se descartan las triviales hacia el ítem ubicuo (lift = 1).

```python
# Diagnóstico 8.3: el filtrado por lift descarta las triviales — NO escribe en el Excel
tickets_mix = [
    ["café", "leche", "bolsa"], ["café", "leche", "bolsa"], ["café", "leche", "bolsa"],
    ["café", "leche", "bolsa"], ["café", "leche", "bolsa"],
    ["té", "bolsa"], ["pan", "bolsa"], ["huevos", "bolsa"], ["pan", "bolsa"], ["huevos", "bolsa"],
]
te_m = TransactionEncoder()
oh_mix = pd.DataFrame(te_m.fit(tickets_mix).transform(tickets_mix), columns=te_m.columns_)
reglas_mix = association_rules(apriori(oh_mix, min_support=0.2, use_colnames=True),
                               metric="confidence", min_threshold=0.5)
reglas_mix["antecedente"] = reglas_mix["antecedents"].apply(como_texto)
reglas_mix["consecuente"] = reglas_mix["consequents"].apply(como_texto)
sobreviven = reglas_mix[reglas_mix["lift"] > 1]
triviales = reglas_mix[reglas_mix["lift"] <= 1]
print(f"soporte(bolsa) = {oh_mix['bolsa'].mean():.2f}  (ítem ubicuo)")
print(f"reglas antes del filtro: {len(reglas_mix)}  ->  tras exigir lift > 1: {len(sobreviven)}")
print(f"reglas triviales descartadas (lift <= 1), todas ligadas a 'bolsa': {len(triviales)}")
print("\nReglas que sobreviven (asociación genuina café<->leche, lift > 1):")
print(sobreviven.sort_values("lift", ascending=False)
      [["antecedente", "consecuente", "confidence", "lift"]].round(3).to_string(index=False))
assert len(sobreviven) >= 1 and (sobreviven["lift"] > 1).all()
assert (triviales["lift"] <= 1 + 1e-9).all()
print("\nassert OK: el filtro por lift conserva la asociación real y elimina las triviales hacia el ítem ubicuo.")
```

**📖 Cómo se lee esta salida** (`SUPUESTOS_S08.md`, Parte 3.2 y 4.2). El **lift** separa asociación de popularidad: las reglas hacia `bolsa` (ubicua) tienen lift = 1 y caen; sobrevive `café ↔ leche`. El orden de filtrado recomendado es soporte → **lift > 1** → confianza → conviction → colapsar redundantes → sentido de negocio, hasta un **top-N** accionable.

**🔎 Qué hace este código (Diagnóstico 8.4 — Apriori vs. FP-Growth).** Lee del contrato `tiempos` que la **identidad** se mantiene a todo umbral (misma cantidad de itemsets) y que la **aceleración** de FP-Growth crece al bajar `minsup`. Solo lee; no escribe.

```python
# Diagnóstico 8.4: mismos itemsets, distinta velocidad — NO escribe en el Excel
comp = tiempos[["minsup", "n_itemsets", "t_apriori_s", "t_fpgrowth_s", "iguales", "aceleracion_fp"]].copy()
print("Apriori vs. FP-Growth (contrato 'tiempos'): MISMOS itemsets, distinta velocidad.")
print(comp.to_string(index=False))
assert bool(tiempos["iguales"].all())     # identidad a todo umbral (diferencia 0 siempre)

# La RELACIÓN es lo robusto (no el segundo exacto, que es reloj de pared y machine-dependent):
comp_desc = comp.sort_values("minsup", ascending=False).reset_index(drop=True)   # minsup decreciente
assert bool((comp_desc["t_fpgrowth_s"] < comp_desc["t_apriori_s"]).all()), \
    "FP-Growth debe ser más rápido que Apriori a TODO minsup"
assert bool((comp_desc["aceleracion_fp"].diff().dropna() >= 0).all()), \
    "la aceleración de FP-Growth debe CRECER (no decrecer) al bajar minsup"
print(f"\nFP-Growth es entre {comp['aceleracion_fp'].min():.1f}x y {comp['aceleracion_fp'].max():.1f}x más rápido; "
      f"la aceleración CRECE al bajar minsup (Apriori genera y cuenta cada vez más candidatos).")
print("assert OK: FP-Growth < Apriori en tiempo a los tres umbrales, y la aceleración no decrece "
      "al bajar minsup -- la RELACIÓN se certifica; el segundo exacto es reloj de pared (machine-dependent).")
```

**📖 Cómo se lee esta salida** (`SUPUESTOS_S08.md`, Parte 5.2). Ambos algoritmos son **exactos**: si devolvieran conjuntos distintos, sería un **error de preparación** (distinto `min_support`, one-hot con enteros, ítems duplicados), **no** un hallazgo. La elección es de **eficiencia**: FP-Growth por defecto cuando `minsup` es bajo o hay patrones largos. Esta sesión certifica la **relación** (no un factor exacto): dos `assert` verifican que FP-Growth es más rápido a los tres umbrales y que la aceleración no decrece al bajar `minsup`.
