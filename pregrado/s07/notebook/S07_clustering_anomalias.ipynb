## 7.2 a 7.4 — ¿Qué es un conglomerado, cómo se construyen los grupos sin fijar k y cómo se mide la segmentación? Teoría guiada (Sección 3 del cuaderno)

Cada concepto se presenta con una **minidemostración ejecutable** (código corto) seguida de su **lectura de negocio**. Todo deriva de la investigación verificada de la sesión (el glosario de la sesión, la guía de interpretación de resultados).

### K-Means: solo punto de partida y sus límites (NO se enseña) — capítulo 7.1 (subsección 3.1)

K-Means es el clustering **por particiones** más conocido: divide *n* observaciones en un **número k fijo de grupos** elegido de antemano, minimizando la distancia al cuadrado a los centroides. Es prerequisito de IA I; **aquí solo se referencia** para entender por qué hacen falta otros métodos. Sus tres límites motivan la sesión:

- **(a)** exige **fijar k** antes de ver los datos → lo levanta el **clustering jerárquico**;
- **(b)** asume clusters **esféricos** de tamaño y densidad similares, y falla con formas alargadas → lo levanta **DBSCAN**;
- **(c)** **no tiene concepto de ruido**: fuerza cada punto a un cluster, incluso los atípicos → también lo levanta **DBSCAN**.

La minidemostración lo muestra sobre dos medias lunas: K-Means (esférico) parte mal la estructura.

🔎 **Qué hace este código.** Genera dos medias lunas, aplica K-Means con k=2 y dibuja lado a lado la estructura real y la partición de K-Means para ver cómo corta en recto.

```python
# Mini-demo: K-Means impone fronteras esféricas y falla con formas alargadas
Xm2, ym2 = make_moons(n_samples=300, noise=0.06, random_state=RANDOM_STATE)
km = KMeans(n_clusters=2, n_init=10, random_state=RANDOM_STATE).fit_predict(Xm2)

fig, axs = plt.subplots(1, 2, figsize=(9.2, 3.8))
axs[0].scatter(Xm2[:, 0], Xm2[:, 1], c=[UPC_TINTA if t == 0 else UPC_ROJO for t in ym2], s=14)
axs[0].set_title("Estructura real (dos medias lunas)")
axs[1].scatter(Xm2[:, 0], Xm2[:, 1], c=[UPC_TINTA if t == 0 else UPC_ROJO for t in km], s=14)
axs[1].set_title("K-Means (k=2): corta en línea recta y mezcla las lunas")
for ax in axs:
    ax.set_xlabel("x1"); ax.set_ylabel("x2")
mostrar(fig, FIGURAS / "S07_demo_kmeans_limite.png")
print("K-Means asigna por cercanía a un centroide -> impone grupos esféricos.")
```

📖 **Lectura de negocio.** K-Means es rápido y sirve de referencia mental, pero segmentar una base de clientes con formas no esféricas, o forzar a cada cliente a un grupo sin reconocer atípicos, produce **segmentos artificiales**. Por eso el curso trabaja con jerárquico y DBSCAN.

### Clustering jerárquico y dendrograma (enlaces single, complete, average, Ward) — capítulo 7.3 (subsección 3.2)

El clustering **jerárquico aglomerativo** **no requiere fijar k**: empieza con cada observación como su propio cluster y en cada paso **fusiona los dos clusters más cercanos**, hasta que queda uno solo. El resultado es un **dendrograma**, un árbol donde la **altura de cada fusión** indica la disimilitud a la que se unieron dos grupos. El número de clusters se decide **después**, cortando el árbol a una altura.

El **enlace** define la distancia entre dos clusters y cambia la forma y el número aparente de segmentos:

- **Single (mínimo):** distancia entre los dos puntos más cercanos. Detecta formas alargadas pero encadena (*chaining*) y es frágil al ruido.
- **Complete (máximo):** distancia entre los dos puntos más lejanos. Clusters compactos, de diámetro parecido.
- **Average (promedio):** media de todas las distancias entre pares. Compromiso entre los dos anteriores.
- **Ward:** fusiona el par que **menos incrementa la varianza intra-cluster**. Segmentos compactos y de **tamaño regular**; es el enlace más usado para segmentar clientes.

🔎 **Qué hace este código.** Sobre tres blobs estandarizados dibuja el dendrograma con enlace Ward (la altura mide la disimilitud de cada fusión) y compara cómo cambian los tamaños de grupo al cortar en 3 clusters con cada enlace.

```python
# Mini-demo: dendrograma con enlace Ward y comparación de enlaces (scipy.cluster.hierarchy.linkage)
Xb, _ = make_blobs(n_samples=60, centers=3, cluster_std=1.0, random_state=RANDOM_STATE)
Xb = StandardScaler().fit_transform(Xb)

fig, ax = plt.subplots(figsize=(8.4, 4.0))
Z = linkage(Xb, method="ward")
dendrogram(Z, ax=ax, color_threshold=0.7 * Z[:, 2].max(), above_threshold_color=UPC_GRIS)
ax.set_title("Dendrograma con enlace Ward: la altura mide la disimilitud de cada fusión")
ax.set_xlabel("observaciones (hojas)"); ax.set_ylabel("altura de fusión (distancia)")
mostrar(fig, FIGURAS / "S07_demo_dendrograma.png")

# Cómo cambia la estructura según el enlace (mismo dato, 3 clusters)
print("Tamaños de grupo con corte a 3 clusters, por enlace:")
for enl in ["single", "complete", "average", "ward"]:
    lab = AgglomerativeClustering(n_clusters=3, linkage=enl).fit_predict(Xb)
    print(f"  {enl:9s} -> {np.bincount(lab)}")
```

📖 **Lectura de negocio.** El dendrograma permite explorar la segmentación **a todos los niveles de granularidad a la vez** (2 macro-segmentos u 8 micro-segmentos) con una sola ejecución y comunicarla visualmente a áreas no técnicas. Elegir el enlace es una **decisión de negocio**: Ward da segmentos equilibrados (útil para repartir presupuesto de marketing por igual); single revela grupos irregulares pero es frágil. **Solo la altura importa** en el dendrograma; el orden horizontal de las hojas es arbitrario.

### DBSCAN: clustering basado en densidad — capítulo 7.3 (subsección 3.3)

DBSCAN (*Density-Based Spatial Clustering of Applications with Noise*) define un cluster como una **región densa** separada de otras por zonas de baja densidad. Usa dos parámetros: **`eps`** (radio de la vecindad) y **`min_samples`** (mínimo de puntos dentro de ese radio para considerar densa la zona). Clasifica cada punto en **núcleo**, **frontera** o **ruido** (etiqueta `-1`). Ventajas sobre K-Means: **descubre el número de clusters solo**, encuentra **formas arbitrarias** y **separa el ruido explícitamente** — y ese ruido conecta con la detección de anomalías.

**Elegir `eps` con el gráfico de k-distancia (codo).** Para cada punto se calcula la distancia a su **k-ésimo vecino** (`k = min_samples`), se **ordenan** de menor a mayor y se grafican. La curva crece de forma gradual y luego **aumenta de forma pronunciada** en un codo: ese valor es un buen `eps`. La plantilla `plantillas/guia_eps_dbscan.docx` sistematiza este procedimiento.

🔎 **Qué hace este código.** Estandariza las medias lunas, traza el gráfico de k-distancia (el codo sugiere `eps`) y corre DBSCAN para ver cómo recupera las dos lunas y marca el ruido como `-1`.

```python
# Mini-demo: DBSCAN recupera las medias lunas (formas arbitrarias) donde K-Means fallaba
Xs2 = StandardScaler().fit_transform(Xm2)
db = DBSCAN(eps=0.20, min_samples=5).fit_predict(Xs2)
n_clusters = len(set(db)) - (1 if -1 in db else 0)
n_ruido = int((db == -1).sum())

fig, axs = plt.subplots(1, 2, figsize=(9.6, 3.8))
# gráfico de k-distancia
nn = NearestNeighbors(n_neighbors=5).fit(Xs2)
dist, _ = nn.kneighbors(Xs2)
kdist = np.sort(dist[:, -1])
axs[0].plot(range(len(kdist)), kdist, color=UPC_TINTA, lw=2)
axs[0].axhline(0.20, color=UPC_ROJO, ls="--", lw=1.5)
axs[0].text(5, 0.22, "eps = 0.20 (codo)", color=UPC_ROJO, fontsize=9)
axs[0].set_title("Gráfico de k-distancia: el codo sugiere eps")
axs[0].set_xlabel("puntos ordenados"); axs[0].set_ylabel("distancia al 5.º vecino")
# resultado DBSCAN
cols = [UPC_GRIS if t == -1 else PALETA[t % len(PALETA)] for t in db]
axs[1].scatter(Xs2[:, 0], Xs2[:, 1], c=cols, s=14)
axs[1].set_title(f"DBSCAN: {n_clusters} clusters + {n_ruido} puntos de ruido (gris)")
axs[1].set_xlabel("x1"); axs[1].set_ylabel("x2")
mostrar(fig, FIGURAS / "S07_demo_dbscan.png")
print(f"DBSCAN encontró {n_clusters} clusters sin fijar k y marcó {n_ruido} puntos como ruido (-1).")
```

📖 **Lectura de negocio.** DBSCAN es útil cuando los segmentos tienen **formas irregulares** y cuando interesa **aislar clientes atípicos** en lugar de forzarlos a un grupo. Los puntos `-1` (ruido) son **candidatos a anomalía**. Su debilidad: falla si los clusters tienen **densidades muy dispares** y es **sensible a `eps`**; el `-1` **no es un cluster más**, es ausencia de cluster.

### Coeficiente de silueta (silhouette) — capítulo 7.4 (subsección 3.4)

La silueta evalúa **qué tan bien asignado está cada punto a su cluster**, sin etiquetas reales. Para un punto *i*, con **a(i)** = distancia media a su propio cluster (cohesión) y **b(i)** = distancia media al cluster vecino más cercano (separación), el coeficiente global es el promedio de los `s(i)`. Cercano a **+1** = punto bien dentro de su cluster; **0** = en la frontera; **negativo** = probablemente mal asignado. Bandas orientativas (Kaufman y Rousseeuw): **>0.7** fuerte, **0.5–0.7** razonable, **0.25–0.5** débil, **≤0.25** sin estructura convincente.

🧮 **Matemática en el cuerpo — distancia euclidiana, inercia/WCSS y silueta.**

Todo lo basado en distancias parte de la **distancia euclidiana** entre dos observaciones $x_i, x_j \in \mathbb{R}^p$:

$$ d(x_i, x_j) \;=\; \sqrt{\sum_{m=1}^{p} \left(x_{im} - x_{jm}\right)^2 }. $$

K-Means (y, por herencia, Ward) minimiza la **inercia** o suma de cuadrados intra-cluster (**WCSS**, *within-cluster sum of squares*): con clusters $C_1,\dots,C_k$ y centroides $\mu_k = \frac{1}{|C_k|}\sum_{i\in C_k} x_i$,

$$ \mathrm{WCSS} \;=\; \sum_{k=1}^{K} \sum_{i \in C_k} \left\lVert x_i - \mu_k \right\rVert^2 . $$

La **silueta** de un punto $i$ combina **cohesión** $a(i)$ (distancia media a su propio cluster) y **separación** $b(i)$ (distancia media al cluster vecino más cercano):

$$ s(i) \;=\; \frac{b(i) - a(i)}{\max\!\big(a(i),\, b(i)\big)} \;\in\; [-1,\, +1], \qquad a(i)=\tfrac{1}{|C_i|-1}\!\!\sum_{j\in C_i,\, j\neq i}\!\! d(x_i,x_j),\quad b(i)=\min_{C\neq C_i}\ \tfrac{1}{|C|}\!\sum_{j\in C} d(x_i,x_j). $$

El **coeficiente global** es $\bar s = \frac{1}{n}\sum_i s(i)$. Estas tres fórmulas se **aplican de forma manual** en la Sección 3.5 (silueta) y se reutilizan en la Sección 7 (`silueta_a_mano`).

🔎 **Qué hace este código.** Sobre cuatro blobs bien separados calcula la silueta media de Ward para k=2..6, para ver cómo la métrica ayuda a elegir el número de clusters (aquí, con estructura nítida, el pico cae en k=4).

```python
# Mini-demo: la silueta media ayuda a elegir el número de clusters
Xk, _ = make_blobs(n_samples=400, centers=4, cluster_std=0.8, random_state=RANDOM_STATE)
Xk = StandardScaler().fit_transform(Xk)
print("Silueta media (Ward) según k:")
for k in range(2, 7):
    lab = AgglomerativeClustering(n_clusters=k, linkage="ward").fit_predict(Xk)
    print(f"  k={k}: silueta = {silhouette_score(Xk, lab):.3f}")
```

Precaución (`INTERPRETACION_RESULTADOS.md`, Sección 3): **la métrica informa, no decide sola** — a veces k=2 da la silueta más alta pero 4 segmentos son más accionables; y en DBSCAN se calcula **excluyendo el ruido** (`-1`).

### 🖐️ Cálculo manual — la silueta y un paso de K-Means por dentro — capítulo 7.4 (subsección 3.5)

Las minidemostraciones anteriores usaron `sklearn` como caja negra. Aquí se reconstruyen **de forma manual** —solo con `numpy`— dos mecánicas centrales y se **verifican con `assert`** contra la librería: **(a)** el coeficiente de silueta `s(i)` de un punto en un ejemplo comprobable manualmente, y **(b)** un paso de K-Means (distancia → asignación → nuevo centroide). La función `silueta_a_mano` que se define aquí se reutiliza en la Sección 7 para reconstruir la segmentación desde cero.

🔎 **Qué hace este código.** Define `silueta_a_mano(X, labels)` (solo `numpy`) y la prueba en un ejemplo de dos grupos sobre una recta donde `s(0,0)` se puede calcular manualmente: `a = 1`, `b = 5.5` → `s = (5.5−1)/5.5 = 0.8182`. Verifica con `assert` que coincide punto a punto con `sklearn.silhouette_samples`.

```python
# 🖐️ Silueta s(i) "a mano" (solo numpy), verificada contra sklearn.silhouette_samples
def silueta_a_mano(X, labels):
    "Coeficiente de silueta por punto s(i)=(b-a)/max(a,b), solo con numpy."
    X = np.asarray(X, dtype=float); labels = np.asarray(labels)
    n = len(X)
    D = np.sqrt(((X[:, None, :] - X[None, :, :]) ** 2).sum(axis=2))   # distancias euclidianas
    etiquetas = np.unique(labels)
    s = np.zeros(n)
    for i in range(n):
        propia = labels == labels[i]; propia[i] = False
        if propia.sum() == 0:
            s[i] = 0.0; continue
        a = D[i, propia].mean()                                       # cohesión
        b = min(D[i, labels == c].mean() for c in etiquetas if c != labels[i])  # separación
        s[i] = (b - a) / max(a, b)
    return s

# Ejemplo verificable a lápiz: dos grupos sobre una recta
Xej = np.array([[0.0, 0], [1, 0], [5, 0], [6, 0]]); lab_ej = np.array([0, 0, 1, 1])
s_mano, s_sklearn = silueta_a_mano(Xej, lab_ej), silhouette_samples(Xej, lab_ej)
print("punto (0,0):  a = dist a (1,0) = 1.0   |   b = media dist a {(5,0),(6,0)} = 5.5")
print(f"s(0,0) a mano = (5.5-1)/5.5 = {s_mano[0]:.4f}   |   sklearn = {s_sklearn[0]:.4f}")
print("s(i) a mano        :", np.round(s_mano, 4))
print("silhouette_samples :", np.round(s_sklearn, 4))

assert abs(s_mano[0] - 0.8182) <= 1e-4                                    # (5.5-1)/5.5 = 0.8182
assert np.allclose(s_mano, s_sklearn, atol=1e-9)                          # a mano == sklearn (por punto)
assert abs(s_mano.mean() - silhouette_score(Xej, lab_ej)) <= 1e-9        # media == coeficiente global
print("OK: silueta a mano == sklearn  (y s(0,0) = 0.8182)")
```

📖 **Cómo se lee.** El `s(0,0)=0.8182` calculado manualmente coincide **exactamente** con `sklearn`, y el vector completo de siluetas también: `silhouette_samples` no encierra ningún artificio, aplica esta fórmula `(b−a)/max(a,b)`. 💡 El punto (0,0) tiene silueta alta porque su vecino de cluster está a 1 y el otro grupo, a 5.5: **muy cohesionado y muy separado**. La misma `silueta_a_mano` se usará en la Sección 7 para reconstruir la segmentación RFM sin `silhouette_score`.

🔎 **Qué hace este código.** Ejecuta **un paso de K-Means de forma manual**: calcula la distancia euclidiana de cada punto a dos centroides iniciales, asigna cada punto al más cercano y recoloca los centroides como la media de su grupo. Verifica con `assert` que la distancia coincide con `scipy.cdist`, la asignación con `pairwise_distances_argmin` y los centroides con `KMeans` a 1 iteración.

```python
# 🖐️ Un paso de K-Means "a mano": distancia -> asignacion -> nuevo centroide
from scipy.spatial.distance import cdist
from sklearn.metrics import pairwise_distances_argmin

Xkm = np.array([[0.0, 0], [1, 0], [0, 1], [10, 10], [11, 10], [10, 11]])
C0 = np.array([[0.0, 0], [10, 10]])                          # centroides iniciales

D_mano = np.sqrt(((Xkm[:, None, :] - C0[None, :, :]) ** 2).sum(axis=2))  # 1) distancia a cada centroide
asig_mano = D_mano.argmin(axis=1)                            # 2) asignar al centroide mas cercano
C1_mano = np.array([Xkm[asig_mano == k].mean(axis=0) for k in range(len(C0))])  # 3) recolocar (media)

D_scipy = cdist(Xkm, C0)                                     # testigo 1: distancia
asig_sklearn = pairwise_distances_argmin(Xkm, C0)           # testigo 2: asignacion
km1 = KMeans(n_clusters=2, init=C0, n_init=1, max_iter=1,
             algorithm="lloyd", random_state=RANDOM_STATE).fit(Xkm)  # testigo 3: 1 iteracion

print("asignacion a mano :", asig_mano, "  | sklearn:", asig_sklearn)
print("centroides nuevos a mano  :\n", np.round(C1_mano, 3))
print("centroides KMeans(1 iter) :\n", np.round(km1.cluster_centers_, 3))

assert np.allclose(D_mano, D_scipy)                                              # distancia == scipy.cdist
assert np.array_equal(asig_mano, asig_sklearn)                                   # asignacion == sklearn
assert np.allclose(np.sort(C1_mano, axis=0), np.sort(km1.cluster_centers_, 0))   # centroide == KMeans
print("OK: distancia==cdist, asignacion==pairwise_distances_argmin, centroide==KMeans(1 iter)")
```

📖 **Cómo se lee.** Los tres testigos confirman la mecánica: la matriz de distancias es la de `scipy`, la asignación es la de `sklearn` y, tras **una** iteración, los centroides de `KMeans` son exactamente las **medias** de los puntos asignados (`[0.33, 0.33]` y `[10.33, 10.33]`). Repetir asignación-actualización hasta que no cambie **es** el algoritmo de K-Means. 💡 El signo/orden de las etiquetas de cluster es arbitrario (por eso se comparan los centroides ordenados): 0 y 1 no tienen jerarquía.

### Detección de anomalías: Isolation Forest y LOF — capítulo 7.2 (subsección 3.6)

- **Isolation Forest (iForest)** *aísla* los puntos en vez de modelar la normalidad: construye muchos árboles que parten los datos con cortes aleatorios. Una **anomalía se aísla en pocos cortes** (queda sola rápido); un punto normal necesita muchos. La **longitud de camino promedio** da un score de anomalía. No usa distancias → escala a datos grandes y es **insensible a la escala**.
- **LOF (Local Outlier Factor)** compara la **densidad local** de cada punto con la de sus vecinos: un punto en una zona mucho **menos densa que sus vecinos** es un atípico **local**. Depende de `n_neighbors`.

**AUC-ROC** evalúa un detector de score continuo **sin fijar umbral** y **sin depender del desbalance**: 0.5 = azar, 1.0 = perfecto. **Cuidado con la orientación del score** (`INTERPRETACION_RESULTADOS.md`, Sección 5): en scikit-learn `score_samples` da **más alto = más normal**, así que para el AUC con "anomalía = clase positiva" se pasa el score **negado** (`-score_samples`); olvidarlo produce un AUC ≈ 1 − AUC (error silencioso).

🔎 **Qué hace este código.** Sobre una nube normal con atípicos dispersos calcula el score de anomalía (negado: más alto = más anómalo) de iForest y de LOF y los grafica en dos paneles para contrastar aislamiento global vs. densidad local.

```python
# Mini-demo: iForest (aislamiento global) vs LOF (densidad local) sobre datos con atípicos
rng = np.random.RandomState(RANDOM_STATE)
X_norm = rng.randn(200, 2) * 0.6
X_out = rng.uniform(low=-4, high=4, size=(15, 2))     # atípicos dispersos
Xtoy = np.vstack([X_norm, X_out])

if_toy = IsolationForest(random_state=RANDOM_STATE).fit(Xtoy)
score_if = -if_toy.score_samples(Xtoy)                # más alto = más anómalo
lof_toy = LocalOutlierFactor(n_neighbors=20)
lof_toy.fit_predict(Xtoy)
score_lof = -lof_toy.negative_outlier_factor_         # más alto = más anómalo

fig, axs = plt.subplots(1, 2, figsize=(9.6, 3.9))
for ax, sc, tit in [(axs[0], score_if, "Isolation Forest"), (axs[1], score_lof, "LOF")]:
    sca = ax.scatter(Xtoy[:, 0], Xtoy[:, 1], c=sc, cmap="Reds", s=22, edgecolor=UPC_GRIS, linewidth=0.3)
    ax.set_title(f"{tit}: color = score de anomalía")
    ax.set_xlabel("x1"); ax.set_ylabel("x2")
    fig.colorbar(sca, ax=ax, shrink=0.85)
mostrar(fig, FIGURAS / "S07_demo_anomalias.png")
print("Ambos scores usan la convención negada: más alto = más anómalo.")
```

La replicación de la Sección 4 los contrasta con AUC-ROC sobre los datos originales del paper.

### Qué preguntaba el paper, y por qué usó lo que usó — Sección 0 del paper (subsección 4.0)

**💡 Antes de tocar los datos.** Una réplica sin esta pregunta se vuelve mecánica: se ejecutan celdas y se obtiene un número. Lo que sigue explica **qué buscaban los autores** y **por qué eligieron cada pieza de su método**, que es de donde proviene el criterio para elegir un método propio en casos futuros. *(Desarrollo completo con las citas del original: la ficha de la sesión de réplica del paper, «Sección 0».)*

**El objetivo no era «detectar mejor»: era dejar de perfilar lo normal.** El resumen del artículo lo declara en su primera línea: «This paper proposes a fundamentally different model-based method that **explicitly isolates anomalies instead of profiles normal points**» (p. 413) —*un método basado en modelo fundamentalmente distinto, que aísla explícitamente las anomalías en vez de perfilar los puntos normales*—. La pregunta era: **¿se puede detectar anomalías sin construir antes un modelo de la normalidad y, si se puede, se obtiene a la vez una mejora en detección y en coste?** Los autores le ponen nombre a los dos defectos del paradigma anterior (p. 413): el detector «is optimized to profile normal instances, but not optimized to detect anomalies» —de ahí «too many false alarms […] or too few anomalies being detected»— y su coste computacional lo confina a pocos datos y pocas dimensiones. Por eso la contribución se mide siempre en **dos** dimensiones a la vez, AUC y segundos.

**Con qué contaban** (p. 419). Once conjuntos reales más uno sintético, elegidos porque «they contain **known anomaly classes as ground truth**», y una restricción autoimpuesta que define todo el trabajo: «anomaly labels are unavailable in the training stage. Anomaly labels are only available in the evaluation stage to compute the performance measure, AUC». El problema es **no supervisado por diseño**; la etiqueta interviene solo como criterio de evaluación —exactamente como en las celdas siguientes, donde `y` no entra en ningún `fit`—. Todo corrió «as single threaded jobs processed at 2.3GHz» (p. 418), un solo hilo, y el coste del rival estaba medido: sobre Mammography, LOF consumió **14 647 segundos** frente a **0,66 s** de iForest (Tabla 3, p. 420).

**Por qué cortes al azar y no una medida de distancia** (pp. 413-415). Dos propiedades sostienen la apuesta: las anomalías «are the minority» y «have attribute-values that are very different» —son **«few and different»**, pocas y diferentes—, «which make them more susceptible to isolation than normal points». Si son pocas y distintas, aislarlas cuesta menos trabajo, y ese trabajo se mide. El mecanismo **no se dio por supuesto**: sobre una gaussiana de 135 puntos, aislar un punto normal exigió **doce** particiones y una anomalía solo **cuatro**; con 1 000 árboles, las longitudes medias convergen a **12,82** y **4,02**. El beneficio es explícito: «iForest utilizes no distance or density measures […] This eliminates major computational cost of distance calculation» (p. 414). De ahí que **iForest sea invariante a la escala** y LOF, basado en distancias, no lo sea: es la razón de que la celda de preparación estandarice para LOF y no para iForest.

**Por qué un submuestreo pequeño, contra la intuición de la época** (pp. 416-418). Es la decisión más contraintuitiva del artículo y así está enunciada: «Contrary to existing methods where large sampling size is more desirable, isolation method works best when the sampling size is kept small». El motivo son el *swamping* —puntos normales próximos a una anomalía que se marcan por error— y el *masking* —un grupo de anomalías tan denso que se encubre a sí mismo—, con un diagnóstico común: «both swamping and masking are a result of **too many data** for the purpose of anomaly detection». Y también aquí midieron antes de decidir: sobre Mulcross, la muestra completa da **AUC 0,67** y un submuestreo de 128 instancias lo eleva a **0,91**. De ahí los dos únicos parámetros que se fijan más abajo sin discutir: **ψ = 256** (`max_samples=256`) y **t = 100** árboles (`n_estimators=100`), «because path lengths usually converge well before t = 100».

**Por qué AUC-ROC y por qué LOF como rival** (pp. 419-420). Como la etiqueta solo interviene al evaluar y el método entrega un score continuo, la pregunta correcta es si el **ordenamiento** coloca las anomalías arriba, sin fijar umbral: eso es el AUC-ROC, y es la métrica correcta con una clase anómala del **2,32 %**, porque con ese desbalance la exactitud premia al modelo que nunca alarma. LOF entra como rival por ser «a well known density based method», el representante del paradigma que el artículo discute. El artículo declara para LOF «a commonly used setting of **k = 10**» (p. 419) mientras aquí se fija `n_neighbors = 20`, y los cortes de iForest son aleatorios (de ahí la semilla). 🧮 **Matemática en el cuerpo — el score de Isolation Forest.**

iForest mide con qué facilidad se puede **aislar** un punto. En un árbol de aislamiento, $h(x)$ es la **longitud de camino** (número de cortes) hasta dejar solo a $x$; se promedia sobre los $t$ árboles: $E[h(x)]$. Para hacerla comparable entre tamaños de muestra se normaliza por $c(n)$, el **camino promedio de una búsqueda fallida en un árbol binario de búsqueda** de $n$ puntos:

$$ c(n) \;=\; 2\,H(n-1) \;-\; \frac{2\,(n-1)}{n}, \qquad H(i) \;=\; \ln(i) + \gamma \quad(\gamma \approx 0{,}5772,\ \text{constante de Euler-Mascheroni}). $$

El **score de anomalía** es entonces

$$ s(x, n) \;=\; 2^{\,-\,\dfrac{E[h(x)]}{c(n)}} \;\in\; (0,\, 1). $$

Un camino **corto** ($E[h(x)] \ll c(n)$) → exponente cercano a 0 → $s \to 1$ (**anómalo**); un camino **largo** → $s \to 0$ (**normal**). ⚠️ **Atención a la orientación en scikit-learn:** `IsolationForest.score_samples` devuelve una versión donde **más alto = más NORMAL** (el opuesto de este $s$). Por eso, para el AUC con «anomalía = positivo», se pasa el score **negado** `-score_samples(X)`. $c(n)$ se deriva de que un iTree es un árbol binario de búsqueda: aislar $x$ equivale a una búsqueda fallida, cuyo coste medio es $c(n)$.

```python
# Paso 1: cargar Mammography y Shuttle (originales del benchmark del paper)
X_mam, y_mam = cargar_mammography()
X_shu, y_shu = cargar_shuttle()
print(f"Mammography: {X_mam.shape} | anomalías = {int(y_mam.sum())} ({y_mam.mean()*100:.2f}%)")
print(f"Shuttle    : {X_shu.shape} | anomalías = {int(y_shu.sum())} ({y_shu.mean()*100:.2f}%)")
```

**❓ Qué se quiere averiguar.** Sin una sola etiqueta disponible, ¿puede un algoritmo situar al principio de la lista los casos que un experto marcaría como anómalos? Y si puede, ¿conviene el método que **aísla** lo raro o el que primero modela cómo es lo normal?

- **Qué decide:** es la elección entre dos familias con costes de cómputo muy distintos en producción. La respuesta de Liu et al. (2008) invirtió el orden de la literatura, y es lo que se replica aquí.
- **Antes de mirar el resultado:** el AUC se lee como la probabilidad de que una anomalía reciba un score más alto que un caso normal. **0,50** equivale a barajar la lista al azar; **1,00**, a ordenarla sin un solo error. Los valores publicados son 0,86 para iForest y 0,67 para LOF en Mammography: la réplica se juzga por su **cercanía a esas dos cifras**, no por lo alta que salga. Y hay un error silencioso al acecho: si el score se usara sin negar, el AUC saldría ≈ 1 − AUC y nadie recibiría un aviso.

🔎 **Qué hace este código.** Ajusta los dos detectores **no supervisados** (iForest con ψ=256/t=100; LOF con `n_neighbors=20`), calcula el AUC sobre el **score negado** (anomalía = positivo) y añade el **guardarraíl** anti-error: el AUC con el score **sin negar** debe dar ≈ 1 − AUC. LOF recibe los datos estandarizados (está basado en distancias); iForest es insensible a la escala.

```python
# Pasos 2-3: detectores no supervisados y AUC-ROC (score negado)
def evaluar_iforest(X, y):
    clf = IsolationForest(n_estimators=100, max_samples=256,
                          contamination="auto", random_state=RANDOM_STATE).fit(X)
    score = -clf.score_samples(X)              # más alto = más anómalo
    return roc_auc_score(y, score), score

def evaluar_lof(X, y, k=20):
    lof = LocalOutlierFactor(n_neighbors=k)
    lof.fit_predict(X)
    score = -lof.negative_outlier_factor_      # más alto = más anómalo
    return roc_auc_score(y, score), score

X_mam_s = StandardScaler().fit_transform(X_mam)
X_shu_s = StandardScaler().fit_transform(X_shu)

auc_if_mam, sc_if_mam = evaluar_iforest(X_mam, y_mam)
auc_lof_mam, sc_lof_mam = evaluar_lof(X_mam_s, y_mam)
auc_if_shu, sc_if_shu = evaluar_iforest(X_shu, y_shu)
auc_lof_shu, _ = evaluar_lof(X_shu_s, y_shu)

print("=== Mammography ===")
print(f"  iForest AUC (OPERATIVO, venv) = {auc_if_mam:.4f}   | benchmark publicado Liu 2008 = 0.86")
print(f"  LOF     AUC (OPERATIVO, venv) = {auc_lof_mam:.4f}   | benchmark publicado Liu 2008 = 0.67")
print("=== Shuttle ===")
print(f"  iForest AUC (OPERATIVO, venv) = {auc_if_shu:.4f}   | benchmark publicado Liu 2008 ~ 1.00")
print(f"  LOF     AUC (OPERATIVO, venv) = {auc_lof_shu:.4f}")

# Guardarraíl anti-error #1: el score SIN negar da ~ 1 - AUC (error silencioso)
# Este print es DIAGNÓSTICO (no lleva `assert`); el guardarraíl FORMAL vive en la Sección 6.1 (celda 78):
#   `assert rec_sin_negar < 0.5` — el umbral REAL del guardarraíl es 0.5 (no 0.2).
auc_mal = roc_auc_score(y_mam, IsolationForest(random_state=RANDOM_STATE).fit(X_mam).score_samples(X_mam))
print(f"\n[guardarraíl, diagnóstico] AUC con score SIN negar = {auc_mal:.4f}  (debe ser ~ 1 - AUC; confirma la orientación)")
```

📖 **Cómo se lee.** Los valores del venv caen junto al benchmark publicado (Liu 2008, etiquetado): iForest ordena las anomalías claramente mejor que LOF en Mammography (**0,8689** vs **0,7204**), y es casi perfecto en Shuttle (**0,9975**). El **guardarraíl** resulta **0,1311 ≈ 1 − 0,8689**: confirma que la orientación correcta es el score **negado** (si el AUC principal resultara 0,13, el detector no estaría fallando: estaría **sin negar**). Pequeñas diferencias con el decimal publicado provienen de la versión de scikit-learn, `max_samples` y la semilla.

🔎 **Qué hace este código.** Dibuja las curvas ROC (cálculo directo desde los scores): Mammography contrasta iForest vs. LOF; Shuttle muestra el iForest casi perfecto. Cuanto más se aproxima la curva a la esquina superior izquierda, mejor ordena el detector.

```python
# Paso 5: curvas ROC (cálculo directo desde los scores)
roc_data = {}   # (dataset, metodo) -> (fpr, tpr); reutilizado en el Anexo B.2 para exportar puntos al Excel
fig, axs = plt.subplots(1, 2, figsize=(10.0, 4.2))
# Mammography: iForest vs LOF
for sc, nombre_metodo, etiqueta, col in [(sc_if_mam, "IsolationForest", f"iForest (AUC {auc_if_mam:.3f})", UPC_ROJO),
                                          (sc_lof_mam, "LOF", f"LOF (AUC {auc_lof_mam:.3f})", UPC_TINTA)]:
    fpr, tpr, _ = roc_curve(y_mam, sc)
    roc_data[("Mammography", nombre_metodo)] = (fpr, tpr)
    axs[0].plot(fpr, tpr, lw=2, color=col, label=etiqueta)
axs[0].plot([0, 1], [0, 1], "--", color=UPC_GRIS, lw=1)
axs[0].set_title("Mammography: iForest ordena mejor las anomalías que LOF")
axs[0].set_xlabel("tasa de falsos positivos"); axs[0].set_ylabel("tasa de verdaderos positivos")
axs[0].legend(loc="lower right", fontsize=9)
# Shuttle: iForest casi perfecto
fpr, tpr, _ = roc_curve(y_shu, sc_if_shu)
roc_data[("Shuttle", "IsolationForest")] = (fpr, tpr)
axs[1].plot(fpr, tpr, lw=2, color=UPC_ROJO, label=f"iForest (AUC {auc_if_shu:.3f})")
axs[1].plot([0, 1], [0, 1], "--", color=UPC_GRIS, lw=1)
axs[1].set_title("Shuttle: iForest casi perfecto (depende del dataset)")
axs[1].set_xlabel("tasa de falsos positivos"); axs[1].set_ylabel("tasa de verdaderos positivos")
axs[1].legend(loc="lower right", fontsize=9)
mostrar(fig, FIGURAS / "S07_roc_anomalias.png")
```

📖 **Conclusión robusta (reproducible).** **iForest supera a LOF** en detección global: Mammography 0,869 vs. 0,720 (Δ ≈ 0,15) y Shuttle 0,997 vs. 0,52. El rendimiento depende del dataset (Shuttle es "sencillo", Mammography tiene solapamiento). En negocio: **iForest** para producción por velocidad; **LOF** cuando importa la anomalía **local**. Elegido el detector, el umbral operativo se fija por el **presupuesto de revisión** (p. ej. revisar solo el 1 % con score más alto), no por el AUC.

## 7.8 — ¿Qué decisión habilita? Laboratorio de negocio: segmentación RFM (Online Retail II) (Sección 5 del cuaderno)

**Paso 6 — Del paper al negocio.** Se segmenta la base de clientes de un e-commerce británico (Online Retail II, UCI id 502; Chen 2019, CC BY 4.0). Se usa la **muestra local** de 140 clientes con toda su historia (`data/online_retail_II_muestra.csv`, ≤1 MB); el conjunto completo (~1,07 M filas) se obtiene en runtime con `data/descargar_datos.py`. La guía paso a paso está en `laboratorio/GUIA_LABORATORIO_S07.docx` y las plantillas en `plantillas/rfm_perfilamiento.docx` y `plantillas/fichas_segmento.docx`.

🔎 **Qué hace este código (Paso a).** Construye la tabla RFM por cliente desde las transacciones: *Recency* (días desde la última compra respecto a una fecha de referencia), *Frequency* (número de facturas distintas) y *Monetary* (gasto total), y describe las tres variables.

```python
# Paso a: construir RFM por cliente desde las transacciones
tx = cargar_retail()
tx["Amount"] = tx["Quantity"] * tx["Price"]            # importe por línea
fecha_ref = tx["InvoiceDate"].max() + pd.Timedelta(days=1)   # fecha de referencia

rfm = (tx.groupby("Customer ID")
         .agg(Recency=("InvoiceDate", lambda s: (fecha_ref - s.max()).days),
              Frequency=("Invoice", "nunique"),
              Monetary=("Amount", "sum"))
         .reset_index())
print(f"Clientes: {len(rfm)} | Transacciones: {len(tx)}")
print(rfm[["Recency", "Frequency", "Monetary"]].describe().round(1).to_string())
```

🔎 **Qué hace este código (Paso b).** *Monetary* y *Frequency* están muy sesgados (colas largas: unos pocos clientes concentran el gasto). Se aplica `log1p` para comprimir las colas y luego `StandardScaler` (z-score), como es práctica estándar en RFM. Sin este paso, la variable de mayor magnitud dominaría la distancia. Comprueba que la matriz queda con media ≈ 0 y desviación ≈ 1 por columna.

```python
# Paso b: transformación log + estandarización (los métodos por distancia lo exigen)
rfm_log = pd.DataFrame({
    "Recency":   np.log1p(rfm["Recency"]),
    "Frequency": np.log1p(rfm["Frequency"]),
    "Monetary":  np.log1p(rfm["Monetary"]),
})
X_rfm = StandardScaler().fit_transform(rfm_log)
print("Matriz RFM estandarizada:", X_rfm.shape, "| media ~0, desv ~1 por columna")
print("medias:", X_rfm.mean(axis=0).round(3), "| desv:", X_rfm.std(axis=0).round(3))
```

🔎 **Qué hace este código (Paso c).** Traza el dendrograma Ward de los clientes (truncado a las últimas 20 fusiones) sobre la matriz RFM estandarizada: los **saltos de altura** sugieren cuántos macro-segmentos hay.

```python
# Paso c: dendrograma Ward de los clientes (scipy linkage)
Z_rfm = linkage(X_rfm, method="ward")
fig, ax = plt.subplots(figsize=(9.0, 4.2))
dendrogram(Z_rfm, ax=ax, truncate_mode="lastp", p=20, show_leaf_counts=True,
           color_threshold=0.6 * Z_rfm[:, 2].max(), above_threshold_color=UPC_GRIS)
ax.set_title("Dendrograma Ward de clientes (RFM): los saltos de altura sugieren el corte")
ax.set_xlabel("clientes agrupados (hojas truncadas)"); ax.set_ylabel("altura de fusión")
mostrar(fig, FIGURAS / "S07_dendrograma_rfm.png")
```

**❓ Qué se quiere averiguar.** ¿En cuántos segmentos se parte una base de 140 clientes cuando nadie sabe cuántos hay y ningún dato dice cuál es el número correcto?

- **Qué decide:** el número de campañas que el área comercial tendrá que diseñar, presupuestar y sostener. Dos segmentos resultan de menor costo y poco diferenciados; seis resultan detallados y probablemente imposibles de operar.
- **Antes de mirar el resultado:** si la silueta dibujara un **pico nítido** —como la demo de cuatro blobs de la Sección 3, que se mueve entre 0,57 y 0,84—, habría un k que los datos defienden y la decisión sería técnica. Si la curva resulta **plana y baja**, no existe un k verdadero: la métrica solo **acota** el rango razonable y quien decide es el negocio. Conviene comprometerse ya: si el máximo cae en un k que el equipo comercial no puede operar, ¿se sigue a la métrica o al negocio?

🔎 **Qué hace este código (Paso d).** Prueba Ward con k = 2..6, compara la silueta media y elige k. La métrica **informa**; la decisión final pondera la **accionabilidad** de negocio (`INTERPRETACION_RESULTADOS.md`, Sección 3).

```python
# Paso d: silhouette de Ward para k=2..6 y elección de k
sil_ward = {}
for k in range(2, 7):
    lab = AgglomerativeClustering(n_clusters=k, linkage="ward").fit_predict(X_rfm)
    sil_ward[k] = silhouette_score(X_rfm, lab)
for k, s in sil_ward.items():
    print(f"  k={k}: silueta = {s:.4f}")

K_ELEGIDO = 4   # silueta casi plana entre k=4..6; se elige 4 por accionabilidad (RFM clásico)
seg_ward = AgglomerativeClustering(n_clusters=K_ELEGIDO, linkage="ward").fit_predict(X_rfm)
rfm["segmento"] = seg_ward
print(f"\nk elegido = {K_ELEGIDO} (silueta = {sil_ward[K_ELEGIDO]:.4f}). Tamaños:",
      np.bincount(seg_ward))
```

📖 **Cómo se lee.** La silueta por k es **k=2: 0,3187, k=3: 0,3297, k=4: 0,3352, k=5: 0,3493, k=6: 0,3472**: el máximo está en **k=5 (0,3493)**, pero la curva es **casi plana** entre k=4 y k=6 y toda la señal es **débil** (banda 0,25–0,5). Por eso se elige **k=4 por accionabilidad** (RFM clásico de 4 segmentos), documentando k=3 como alternativa válida. ⚠️ No se maximiza la silueta de forma mecánica: la métrica **acota**, el negocio **decide** (Sección 8.3 y `SUPUESTOS_S07.md`, Parte 2.2).

**❓ Qué se quiere averiguar.** Ward asigna a **todos** los clientes a algún grupo. ¿Y si algunos no pertenecen a ninguno? ¿Cuántos clientes de esta base son sencillamente raros, y dónde queda la frontera entre «raro» y «miembro del grupo»?

- **Qué decide:** el grupo de ruido (`-1`) no es un descarte administrativo: son los clientes a los que ninguna ficha de segmento va a describir bien, y son el puente hacia la detección de anomalías del Paso h. `eps` es el hiperparámetro que fija cuántos son.
- **Antes de mirar el resultado:** el radio no se elige por inspección visual sino con el gráfico de k-distancia, cuyo codo lo acota. Con un `eps` **muy pequeño** casi toda la base cae en ruido y no hay segmentación; con uno **muy grande**, un único cluster absorbe a todos y tampoco la hay. Por eso la rejilla se recorre con dos condiciones fijadas de antemano: al menos 2 clusters y menos del 40 % de ruido. Una silueta muy alta con el 80 % de los clientes fuera no es un buen resultado, es una partición sin contenido.

🔎 **Qué hace este código (Paso e).** Fija `min_samples = 2 × dimensiones = 6`, traza la curva de k-distancia y detecta el **codo**; dentro de la región del codo afina `eps` por silueta (≥2 clusters, ruido < 40 %). El grupo de ruido (`-1`) son los clientes atípicos, puente con la detección de anomalías.

```python
# Paso e: DBSCAN con eps del gráfico de k-distancia
MIN_SAMPLES = 2 * X_rfm.shape[1]        # 2 x dimensiones = 6
nn = NearestNeighbors(n_neighbors=MIN_SAMPLES).fit(X_rfm)
dist, _ = nn.kneighbors(X_rfm)
kdist = np.sort(dist[:, -1])

# codo (Kneedle simple sobre la curva recortada al 95%, evitando la cola de atípicos)
n_trim = int(len(kdist) * 0.95); curva = kdist[:n_trim]; xs = np.arange(len(curva))
x0, y0, x1, y1 = xs[0], curva[0], xs[-1], curva[-1]
num = np.abs((y1 - y0) * xs - (x1 - x0) * curva + x1 * y0 - y1 * x0)
den = np.hypot(y1 - y0, x1 - x0)
eps_codo = round(float(curva[int(np.argmax(num / den))]), 2)

# afinar eps por silueta dentro de la región del codo; se registra la REJILLA COMPLETA (no solo el ganador)
# -> grid_resultados se reutiliza en el Anexo B.1 para exportar la rejilla entera al Excel (hoja dbscan_grid_eps)
grid_resultados = []   # (eps, n_clusters, n_ruido, pct_ruido, silueta_o_None) para TODO punto evaluado
mejor = None
for e in np.round(np.arange(max(0.2, eps_codo - 0.3), eps_codo + 0.01, 0.05), 2):
    lab = DBSCAN(eps=float(e), min_samples=MIN_SAMPLES).fit_predict(X_rfm)
    nc = len(set(lab)) - (1 if -1 in lab else 0)
    nru = int((lab == -1).sum())
    s = None
    if nc >= 2:
        m = lab != -1
        if m.sum() > nc:
            s = float(silhouette_score(X_rfm[m], lab[m]))
    grid_resultados.append((float(e), nc, nru, round(nru / len(lab) * 100, 1), s))
    if nc >= 2 and nru / len(lab) < 0.40 and s is not None:
        if mejor is None or s > mejor[1]:
            mejor = (float(e), s)
EPS = mejor[0]
seg_db = DBSCAN(eps=EPS, min_samples=MIN_SAMPLES).fit_predict(X_rfm)
n_clusters_db = len(set(seg_db)) - (1 if -1 in seg_db else 0)
n_ruido_db = int((seg_db == -1).sum())
mask = seg_db != -1
sil_db = silhouette_score(X_rfm[mask], seg_db[mask])

fig, ax = plt.subplots(figsize=(7.4, 4.0))
ax.plot(range(len(kdist)), kdist, color=UPC_TINTA, lw=2)
ax.axhline(eps_codo, color=UPC_GRIS, ls=":", lw=1.5)
ax.axhline(EPS, color=UPC_ROJO, ls="--", lw=1.6)
ax.text(2, eps_codo + 0.03, f"codo del gráfico = {eps_codo}", color=UPC_GRIS, fontsize=9)
ax.text(2, EPS - 0.10, f"eps operativo = {EPS}", color=UPC_ROJO, fontsize=9)
ax.set_title("Gráfico de k-distancia (RFM): el codo acota eps para DBSCAN")
ax.set_xlabel("clientes ordenados"); ax.set_ylabel(f"distancia al {MIN_SAMPLES}.º vecino")
mostrar(fig, FIGURAS / "S07_kdistancia_rfm.png")

print(f"eps del codo (gráfico) = {eps_codo} | eps operativo (afinado) = {EPS}")
print(f"DBSCAN -> {n_clusters_db} clusters + {n_ruido_db} clientes de ruido "
      f"({n_ruido_db/len(seg_db)*100:.0f}%) | silueta (sin ruido) = {sil_db:.4f}")
print(f"Rejilla evaluada (eps, n_clusters, n_ruido, %ruido, silueta): {grid_resultados}")
```

📖 **Cómo se lee.** El codo del gráfico da `eps ≈ 0,63`; afinado por silueta baja a **`eps = 0,53`**, con `min_samples = 6` → **3 clusters + 39 clientes de ruido (28 %)**, silueta (sin ruido) **0,3179**.

**Comparación jerárquico vs. DBSCAN (drill 2).** Ambos dan una silueta comparable (~0,33), pero difieren en filosofía: **Ward** entrega **k segmentos equilibrados** (todos los clientes asignados, útil para repartir campañas); **DBSCAN** entrega un núcleo denso más un **grupo de ruido explícito** (los atípicos que Ward disolvería dentro de un segmento). Cuando el negocio exige exactamente *k* segmentos accionables, Ward es preferible; cuando interesa **aislar atípicos**, DBSCAN.

**❓ Qué se quiere averiguar.** Los cuatro grupos ya existen como etiquetas 0, 1, 2 y 3. ¿Describen comportamientos de compra distintos, o son cuatro cajas que nadie puede accionar el lunes?

- **Qué decide:** es la prueba que separa una partición **matemáticamente válida** de una segmentación **útil**. Un segmento accionable admite la frase «a estos, esto, por esta razón»; si no la admite, no se opera aunque su silueta sea buena.
- **Antes de mirar el resultado:** las cifras se leen en unidades de negocio —días desde la última compra, número de pedidos, libras gastadas—, nunca en las z de la distancia. Si los cuatro grupos salieran con perfiles **parecidos** en R, F y M, la partición no serviría para nada. Si cada uno ocupa un **rincón distinto** del espacio RFM, cada rincón admite su propia acción. Y una cautela previa sobre el segmento más pequeño: con pocos clientes, la media y la mediana de gasto pueden separarse mucho, y quien describe al cliente típico es la **mediana**.

🔎 **Qué hace este código (Paso f).** Describe cada segmento Ward con su **tamaño** y su **valor típico en unidades de negocio** (no estandarizadas): días, número de pedidos y gasto (media y mediana). A partir de la tabla **nombra** cada segmento con una regla explícita (R baja = reciente; F/M altos = valioso).

```python
# Paso f: perfilamiento de los segmentos Ward (valores en unidades de negocio)
perfil = (rfm.groupby("segmento")
            .agg(n_clientes=("Customer ID", "size"),
                 Recency_media=("Recency", "mean"),
                 Frequency_media=("Frequency", "mean"),
                 Monetary_media=("Monetary", "mean"),
                 Monetary_mediana=("Monetary", "median"))
            .round(1).reset_index())
perfil["pct_clientes"] = (perfil["n_clientes"] / len(rfm) * 100).round(1)

# nombrar cada segmento a partir de la tabla (R baja = reciente; F/M altos = valioso)
def nombrar(r):
    if (r["Frequency_media"] >= perfil["Frequency_media"].median()
            and r["Monetary_media"] >= perfil["Monetary_media"].median()
            and r["Recency_media"] <= perfil["Recency_media"].median()):
        return "Campeones / VIP"
    if r["Recency_media"] >= perfil["Recency_media"].quantile(0.75):
        return "Hibernando / perdidos"
    if r["Monetary_media"] >= perfil["Monetary_media"].median():
        return "Valiosos en riesgo"
    return "Ocasionales / nuevos"
perfil["segmento_nombre"] = perfil.apply(nombrar, axis=1)
print(perfil[["segmento", "segmento_nombre", "n_clientes", "pct_clientes",
              "Recency_media", "Frequency_media", "Monetary_media", "Monetary_mediana"]].to_string(index=False))
```

📖 **Lectura de negocio (fichas de segmento).** Cada nombre está respaldado por la tabla (`INTERPRETACION_RESULTADOS.md`, Sección 4). Traducción típica a acción: **Campeones/VIP** → retener con programa de fidelidad; **Valiosos en riesgo** → reactivación antes de que se enfríen; **Ocasionales/nuevos** → onboarding y segunda compra; **Hibernando/perdidos** → campaña de bajo costo o descarte. El detalle por segmento se materializa en `plantillas/fichas_segmento.docx`. *Nota:* la mediana de *Monetary* describe mejor al cliente típico cuando hay colas largas.

🔎 **Qué hace este código (Paso g).** Dibuja el scatter Recency vs. Monetary (escala log) coloreado por segmento, con el tamaño del punto proporcional a la Frequency.

```python
# Paso g: scatter de los segmentos RFM (cálculo directo)
fig, ax = plt.subplots(figsize=(7.6, 4.6))
for c in sorted(rfm["segmento"].unique()):
    m = rfm["segmento"] == c
    nombre = perfil.loc[perfil["segmento"] == c, "segmento_nombre"].iloc[0]
    ax.scatter(rfm.loc[m, "Recency"], rfm.loc[m, "Monetary"],
               s=20 + rfm.loc[m, "Frequency"] * 6, alpha=0.75,
               color=PALETA[c % len(PALETA)], edgecolor="white", linewidth=0.4,
               label=f"{c}: {nombre}")
ax.set_yscale("log")
ax.set_xlabel("Recency (días desde la última compra)")
ax.set_ylabel("Monetary (gasto total, escala log)")
ax.set_title("Segmentos RFM: los recientes y de alto gasto son los más valiosos")
ax.legend(fontsize=8, title="segmento (tamaño = Frequency)")
mostrar(fig, FIGURAS / "S07_scatter_rfm.png")
```

**❓ Qué se quiere averiguar.** El detector entrega un score continuo para las 18 125 transacciones. ¿Dónde se corta esa lista? Es decir, ¿cuántas operaciones se envían a revisión manual?

- **La decisión concreta:** el umbral no lo fija la estadística, lo fija cuántas revisiones caben en el equipo. Un corte generoso multiplica los **falsos positivos** y consume horas de analista sobre operaciones legítimas; un corte estrecho deja pasar el fraude que se quería detectar, y ese **falso negativo** no aparece en ningún informe porque nadie llegó a mirarlo.
- **Antes de mirar el resultado:** el corte de esta sesión es el percentil 99 —el 1 % más anómalo—, elegido como **presupuesto de revisión** y no por optimizar métrica alguna. Si las marcadas tuvieran un perfil **idéntico** al de la transacción típica en cantidad, precio e importe, el detector no habría encontrado nada y el corte sobraría. Si se apartan con claridad en alguna de las tres, hay material que revisar. Lo que se obtiene aquí es una **lista de candidatas**, nunca una sentencia de fraude.

🔎 **Qué hace este código (Paso h).** Con Isolation Forest sobre las transacciones (cantidad, precio, importe) marca el **1 % con score de anomalía más alto** para revisión y compara el perfil (medianas) de las transacciones anómalas frente al total.

```python
# Paso h: 1% de transacciones más anómalas con Isolation Forest
# Convención de esta sesión: contamination=0.01 FIJO (no 'auto'), para dejar explícito en el propio ajuste
# que se pide "el 1% más anómalo". El marcado real siempre se hace por el umbral p99 del score (línea de abajo),
# así que el resultado (transacciones marcadas) no depende de este valor: el Diagnóstico 8.5 (celda 100)
# lo prueba variando `contamination` en {0.01, 0.0232, 0.05, 0.10, 'auto'} — el AUC/ranking no cambia,
# solo cambiaría el `predict()` binario, que aquí no se usa. Se estandariza por consistencia de pipeline
# (mismo criterio que Mammography/Shuttle); el Anexo B.4 certifica que, para Isolation Forest, el resultado
# es idéntico con y sin `StandardScaler` en este dataset.
X_tx = StandardScaler().fit_transform(tx[["Quantity", "Price", "Amount"]])
if_tx = IsolationForest(contamination=0.01, random_state=RANDOM_STATE).fit(X_tx)
score_tx = -if_tx.score_samples(X_tx)
umbral = np.quantile(score_tx, 0.99)
tx["es_anomalia"] = (score_tx >= umbral).astype(int)
anom = tx[tx["es_anomalia"] == 1]
print(f"Transacciones marcadas como anómalas: {len(anom)} ({tx['es_anomalia'].mean()*100:.2f}%)")
print("\nPerfil de las transacciones anómalas vs. el total (medianas):")
print(pd.DataFrame({
    "anómalas": anom[["Quantity", "Price", "Amount"]].median(),
    "todas":    tx[["Quantity", "Price", "Amount"]].median()}).round(2).to_string())
```

## 7.3 en profundidad — 🧱 Construcción desde cero (Sección 7 del cuaderno)

Se reconstruye el flujo **sin los ayudantes de alto nivel** para probar que se entiende la mecánica: **(a)** la segmentación RFM sin `StandardScaler`, `AgglomerativeClustering` ni `silhouette_score` —solo `numpy`, `scipy.linkage` y la `silueta_a_mano` de la Sección 3.5— y **(b)** el AUC de iForest como el estadístico **U de Mann-Whitney** sobre los scores. Ambos reproducen el **contrato** del Excel con `assert`.

🔎 **Qué hace este código.** Define `segmentar_desde_cero(df_log, k)` que estandariza de forma manual (z-score), aplica Ward con `scipy.linkage`, corta el árbol en k grupos con `fcluster` y calcula la silueta con `silueta_a_mano`. Reproduce la silueta de Ward del Excel para k=2..6 (pico en **k=5 = 0,3493**) con `assert`.

```python
# 🧱 Segmentar RFM DESDE CERO: sin StandardScaler / AgglomerativeClustering / silhouette_score
# (solo numpy + scipy.linkage + la silueta_a_mano de la Seccion 3.5) -> reproduce el contrato del Excel
from scipy.cluster.hierarchy import fcluster

def segmentar_desde_cero(df_log, k):
    M = df_log.values.astype(float)
    Z = (M - M.mean(axis=0)) / M.std(axis=0)          # z-score a mano (== StandardScaler, ddof=0)
    L = linkage(Z, method="ward")                     # Ward con scipy
    lab = fcluster(L, t=k, criterion="maxclust")      # cortar el arbol en k grupos
    return float(silueta_a_mano(Z, lab).mean())       # silueta con la funcion de la Seccion 3.5

xls_clu = pd.read_excel(XLSX, sheet_name="clustering")
ward_xls = {int(r.k_o_eps): float(r.silhouette) for r in
            xls_clu[xls_clu["metodo"] == "Ward"].itertuples()}

print(f"{'k':>3s}{'silueta desde cero':>20s}{'Excel (Ward)':>14s}")
sil_cero = {}
for k in range(2, 7):
    sil_cero[k] = segmentar_desde_cero(rfm_log, k)
    print(f"{k:>3d}{sil_cero[k]:>20.4f}{ward_xls[k]:>14.4f}")
    assert abs(sil_cero[k] - ward_xls[k]) <= 2e-3, (k, sil_cero[k], ward_xls[k])

k_pico = max(sil_cero, key=sil_cero.get)
assert k_pico == 5                                     # el pico de silueta esta en k=5
assert abs(sil_cero[5] - 0.3493) <= 2e-3               # y vale 0.3493 (contrato del Excel)
print(f"OK: pipeline reconstruido reproduce la silueta de Ward del Excel; pico en k={k_pico} ({sil_cero[5]:.4f})")
```

📖 **Cómo se lee.** Con solo `numpy` + `scipy.linkage` + `silueta_a_mano` se reproduce la **misma** curva de silueta de Ward del Excel (pico en **k=5 = 0,3493**): `AgglomerativeClustering`/`silhouette_score` no encierran ningún artificio, aplican esta secuencia estandarizar → Ward → cortar → silueta. Ahora se reconstruye la otra pieza clave, el **AUC**, para ver que también es una fórmula transparente.

🔎 **Qué hace este código.** Define `auc_desde_scores(scores, y)` que calcula el AUC como el estadístico **U de Mann-Whitney** vía rangos (`AUC = (R⁺ − n⁺(n⁺+1)/2) / (n⁺·n⁻)`) sobre los scores de iForest en Mammography, y verifica con `assert` que es igual a `sklearn.roc_auc_score` y al Excel (**0,8689**).

```python
# 🧱 AUC DESDE CERO como estadistico U de Mann-Whitney sobre los scores de iForest (Mammography)
def auc_desde_scores(scores, y):
    "AUC = P(score_anomalia > score_normal) via rangos (U de Mann-Whitney), solo numpy/scipy."
    from scipy.stats import rankdata
    y = np.asarray(y).astype(int); scores = np.asarray(scores, dtype=float)
    r = rankdata(scores)                            # rangos promedio (maneja empates)
    n_pos = int(y.sum()); n_neg = int((y == 0).sum())
    U = r[y == 1].sum() - n_pos * (n_pos + 1) / 2.0
    return float(U / (n_pos * n_neg))

auc_mano = auc_desde_scores(sc_if_mam, y_mam)       # sc_if_mam = -score_samples de iForest (Seccion 4)
auc_sklearn = float(roc_auc_score(y_mam, sc_if_mam))
xls_if = pd.read_excel(XLSX, sheet_name="anomalias_iforest").set_index("metrica")["valor"]
auc_excel = float(xls_if["auc_if_mammography"])

print(f"AUC a mano (U de Mann-Whitney) = {auc_mano:.4f}")
print(f"AUC sklearn.roc_auc_score      = {auc_sklearn:.4f}")
print(f"AUC en el Excel (contrato)     = {auc_excel:.4f}   (benchmark Liu 2008 = 0.86)")

assert abs(auc_mano - auc_sklearn) <= 1e-9          # el AUC ES el U de Mann-Whitney normalizado
assert abs(auc_mano - auc_excel) <= 1e-3            # y coincide con el Excel (0.8689)
print("OK: AUC reconstruido a mano == sklearn == Excel (0.8689)")
```

📖 **Cómo se lee.** El AUC reconstruido manualmente coincide **exactamente** con `sklearn` y con el Excel (**0,8689**): el AUC-ROC **es** la probabilidad de que una anomalía reciba un score más alto que un punto normal —el estadístico **U de Mann-Whitney** normalizado por `n⁺·n⁻`—. Por eso es **invariante al umbral** y al desbalance: solo depende del **orden** de los scores, no de dónde se corte. 💡 Es la razón por la que `contamination` no cambia el AUC (Sección 8.5).

```python
# Diagnostico 8.1 - la estandarizacion cambia el clustering (NO escribe el Excel)
X_raw = rfm[["Recency", "Frequency", "Monetary"]].values.astype(float)          # SIN escalar
lab_raw = AgglomerativeClustering(n_clusters=4, linkage="ward").fit_predict(X_raw)
lab_std = AgglomerativeClustering(n_clusters=4, linkage="ward").fit_predict(X_rfm)  # escalado (log+z)
sil_raw = silhouette_score(X_raw, lab_raw)
sil_std = silhouette_score(X_rfm, lab_std)

def _separacion(labels):
    "Rango de las medias por cluster en cada variable, como fraccion del rango total (1 = una sola variable separa)."
    cols = ["Recency", "Frequency", "Monetary"]
    g = rfm[cols].assign(c=labels).groupby("c").mean()
    rango = rfm[cols].max() - rfm[cols].min()
    return ((g.max() - g.min()) / rango).round(2).to_dict()

print(f"Silueta SIN escalar = {sil_raw:.4f}   |   escalado (log+z) = {sil_std:.4f}")
print("Separacion de las medias por variable (fraccion del rango total):")
print("  SIN escalar:", _separacion(lab_raw))
print("  escalado   :", _separacion(lab_std))
print("\nSin escalar, Monetary (miles) domina la distancia: los grupos se ordenan casi solo por gasto.")
print("Al estandarizar, R/F/M pesan por igual y aparecen segmentos de COMPORTAMIENTO, no rangos de gasto.")
```

📖 **Cómo se lee.** Sin escalar, la partición separa a los clientes casi **solo por *Monetary*** (su fracción de separación domina, las de *Recency*/*Frequency* son mínimas): son **rangos de gasto disfrazados de segmentos**. Al estandarizar (log + z-score), las tres variables pesan por igual y emergen perfiles de **comportamiento**. ⚠️ Es la **decisión central** de la sesión: estandarizar por defecto antes de cualquier método por distancias (excepto iForest, insensible a la escala). 🔎 **Qué hace este código (Diagnóstico 8.5 — contaminación).** Ajusta iForest sobre Mammography variando `contamination` (0,01 / 0,0232 / 0,05 / 0,10 / auto) con la misma semilla y compara el **AUC** (sobre el score) y el **número de marcadas** (por `predict`), para mostrar que la contaminación no cambia el AUC. **No** escribe el Excel.

```python
# Diagnostico 8.5 - contaminacion NO cambia el AUC de iForest: solo mueve el umbral (NO escribe el Excel)
print(f"{'contamination':>14s}{'AUC':>9s}{'n_marcadas':>12s}")
aucs = []
for cont in [0.01, 0.0232, 0.05, 0.10, "auto"]:
    clf = IsolationForest(n_estimators=100, max_samples=256, contamination=cont,
                          random_state=RANDOM_STATE).fit(X_mam)
    auc = roc_auc_score(y_mam, -clf.score_samples(X_mam))
    n_marc = int((clf.predict(X_mam) == -1).sum())
    aucs.append(round(auc, 4))
    print(f"{str(cont):>14s}{auc:>9.4f}{n_marc:>12d}")
print(f"\nEl AUC es identico ({aucs[0]:.4f}) para toda contaminacion: se calcula sobre el SCORE continuo,")
print("que no cambia con contamination; esta solo mueve cuantas se marcan (discusion sklearn #28172).")
```

## 7.7 — ¿Cómo se verifica que el resultado es real? Anexo A: robustez, incertidumbre de las métricas y valor en dinero de las anomalías

> Anexo técnico añadido tras el panel de revisión. **Complementario y no ejecutivo**: estas celdas **NO escriben el Excel** ni cambian ninguna cifra del contrato (Sección 6); solo cuantifican la **incertidumbre** de los números protagonistas y anclan la decisión antifraude al **dinero real del caso**. Reutilizan objetos ya calculados (`X_mam`, `y_mam`, `sc_if_mam`, `X_rfm`, `sil_ward`, `tx`).

🔎 **Qué hace este código (A.1 — incertidumbre de AUC y silueta).** El clímax de la sesión es una comparación numérica (iForest 0.8689 vs. LOF 0.7204; Ward k=5 = 0.3493), reportada a cuatro decimales. Una cifra puntual sin banda es **precisión falsa**. Aquí se separan **dos fuentes de variabilidad distintas**:
- **Error de muestreo** del AUC de iForest en Mammography: **bootstrap** sobre las parejas (score, etiqueta) con semilla fija → IC 95 %. Es la incertidumbre por tener *esta* muestra y no otra (comparable a la SE de Hanley–McNeil para ~260 anomalías).
- **Jitter de semilla** del iForest: reajustar el bosque con distintas `random_state` (el modelo es estocástico) → desviación entre semillas. Es la incertidumbre por el *azar del algoritmo*, no por los datos.

Para la **silueta de Ward k=5** se estima su **estabilidad** por submuestreo del 80 % de los clientes (Ward es determinista: no tiene jitter de semilla, solo error de muestreo). Al final se comprueba que la ventaja iForest−LOF (Δ≈0.15) **supera holgadamente** ambas bandas.

```python
# A.1 Incertidumbre: IC bootstrap del AUC de iForest (Mammography) + jitter de semilla; estabilidad de la silueta de Ward k=5
# (NO escribe el Excel; reutiliza sc_if_mam, y_mam, X_mam, X_rfm, sil_ward ya calculados)

# --- error de MUESTREO: bootstrap sobre las parejas (score, etiqueta) con semilla fija ---
rng = np.random.default_rng(RANDOM_STATE)
n_mam = len(y_mam); B = 1000
auc_boot = []
for _ in range(B):
    idx = rng.integers(0, n_mam, n_mam)               # remuestreo con reemplazo
    if 0 < int(y_mam[idx].sum()) < n_mam:             # descartar remuestras de una sola clase
        auc_boot.append(roc_auc_score(y_mam[idx], sc_if_mam[idx]))
auc_boot = np.array(auc_boot)
ic_lo, ic_hi = np.percentile(auc_boot, [2.5, 97.5])

# --- jitter de SEMILLA: reajustar iForest con 20 semillas (el contrato usa random_state=42) ---
auc_semilla = []
for s in range(20):
    clf_s = IsolationForest(n_estimators=100, max_samples=256, contamination="auto", random_state=s).fit(X_mam)
    auc_semilla.append(roc_auc_score(y_mam, -clf_s.score_samples(X_mam)))
auc_semilla = np.array(auc_semilla)

# --- estabilidad de la silueta de Ward k=5: submuestreo del 80% (Ward es determinista) ---
rng_sil = np.random.default_rng(RANDOM_STATE)
sil_sub = []
for _ in range(200):
    idx = rng_sil.choice(len(X_rfm), int(0.8 * len(X_rfm)), replace=False)
    Xs = X_rfm[idx]
    sil_sub.append(silhouette_score(Xs, AgglomerativeClustering(n_clusters=5, linkage="ward").fit_predict(Xs)))
sil_sub = np.array(sil_sub)
sic_lo, sic_hi = np.percentile(sil_sub, [2.5, 97.5])

delta = auc_if_mam - auc_lof_mam
print(f"AUC iForest Mammography (contrato, semilla 42) = {auc_if_mam:.4f}")
print(f"  error de MUESTREO (bootstrap B={len(auc_boot)}): SE = {auc_boot.std(ddof=1):.4f} | IC 95% = [{ic_lo:.4f}, {ic_hi:.4f}]")
print(f"  jitter de SEMILLA (20 refits):                 desv = {auc_semilla.std(ddof=1):.4f} | rango = [{auc_semilla.min():.4f}, {auc_semilla.max():.4f}]")
print(f"\nSilueta Ward k=5 (contrato) = {sil_ward[5]:.4f}   [Ward es determinista: sin jitter de semilla]")
print(f"  estabilidad (submuestreo 80%, 200 rep.): SE = {sil_sub.std(ddof=1):.4f} | IC 95% = [{sic_lo:.4f}, {sic_hi:.4f}]")
print(f"  (referencia: silueta Ward k=4 = {sil_ward[4]:.4f}; el IC de k=5 la solapa -> la elección k=4/k=5 cae dentro del ruido)")
print(f"\nVentaja iForest - LOF (Mammography) = {delta:.4f}: supera el error de muestreo (~{auc_boot.std(ddof=1):.3f}) y el jitter de semilla (~{auc_semilla.std(ddof=1):.3f}).")
assert delta > 5 * auc_boot.std(ddof=1)               # la conclusión no la explica el ruido del estimador
```

📖 **Cómo se lee.** El AUC de iForest (0.8689) trae ahora su **banda**: un IC 95 % de bootstrap ≈ [0.85, 0.89] (SE ≈ 0.011, coherente con Hanley–McNeil para ~260 anomalías) y un jitter de semilla de desviación ≈ 0.010 — **dos incertidumbres del mismo orden pero de origen distinto**: una viene de la muestra, la otra del azar del bosque. La silueta de Ward k=5 (0.3493) tiene un IC de estabilidad ≈ [0.32, 0.39] que **solapa** con k=4 (0.3352): la partición es estadísticamente **débil** y elegir k=4 por accionabilidad no contradice los datos. Lo que **sí** es robusto es la conclusión central: la ventaja iForest−LOF (Δ≈0.15) es **varias veces** mayor que cualquiera de las dos bandas, así que no la explica el ruido. *(Nota honesta: el AUC es in-sample —ajuste y score sobre los mismos datos—; el bootstrap acota el error de muestreo, no el optimismo in-sample.)*
